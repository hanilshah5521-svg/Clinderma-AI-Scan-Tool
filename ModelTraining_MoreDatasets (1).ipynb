```python
# ================================================================================
# CLINDERMA STAGE 8A — ROBUST MULTI-SOURCE DATASET ACQUISITION
# ================================================================================
#
# NEW EXPERIMENT ONLY
#
# Sources:
#   PIE
#   MELASMA
#   FACE SEGMENT
#   SKIN PROBLEM INSTANCE SEGMENTATION
#
# Target:
#   Build a common raw-data pool for the new multi-source experiment.
#
# IMPORTANT:
#   - No class mapping
#   - No augmentation
#   - No normalization
#   - No training
#   - No modification of old processed dataset
#   - No modification of old checkpoints
# ================================================================================

from pathlib import Path
from google.colab import userdata
from roboflow import Roboflow

import os
import json
import time
import shutil
import traceback


# ================================================================================
# 1. PATHS
# ================================================================================

WORKSPACE = Path("/content/clinderma_workspace")

RAW = WORKSPACE / "raw_multisource"
AUDIT = WORKSPACE / "audit_multisource"

RAW.mkdir(parents=True, exist_ok=True)
AUDIT.mkdir(parents=True, exist_ok=True)


# ================================================================================
# 2. ROBOFLOW API KEY
# ================================================================================

ROBOFLOW_API_KEY = userdata.get("ROBOFLOW_API_KEY")

if not ROBOFLOW_API_KEY:
    raise RuntimeError(
        "ROBOFLOW_API_KEY could not be loaded from Colab Secrets."
    )

os.environ["ROBOFLOW_API_KEY"] = ROBOFLOW_API_KEY


# ================================================================================
# 3. ROBOFLOW CONFIGURATION
# ================================================================================

ROBOFLOW_WORKSPACE = "hanil-shah-s-workspace"

DATASETS = {
    "pie": {
        "project": "pie-annotation-fvniz",
        "version": 1,
        "type": "instance-segmentation",
    },

    "melasma": {
        "project": "melasma-segmentation-z3raq-g7q8g",
        "version": 1,
        "type": "instance-segmentation",
    },

    "face_segment": {
        "project": "face-segment-l1nwn-vbyro",
        "version": 1,
        "type": "instance-segmentation",
    },

    "skin_problem": {
        "project": "skin-problem-instance-segmentation-01-o6hvd",
        "version": 1,
        "type": "instance-segmentation",
    },
}


# ================================================================================
# 4. CONNECT
# ================================================================================

print("=" * 80)
print("CLINDERMA STAGE 8A")
print("ROBUST MULTI-SOURCE DATASET ACQUISITION")
print("=" * 80)

print()
print("Workspace:", ROBOFLOW_WORKSPACE)

rf = Roboflow(api_key=ROBOFLOW_API_KEY)
workspace = rf.workspace(ROBOFLOW_WORKSPACE)

print()
print("✓ Roboflow workspace accessible")
print("Workspace:", workspace.name)
print("URL:", workspace.url)


# ================================================================================
# 5. HELPERS
# ================================================================================

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
}


def inventory_dataset(path):

    files = [
        p for p in path.rglob("*")
        if p.is_file()
    ]

    images = [
        p for p in files
        if p.suffix.lower() in IMAGE_EXTENSIONS
    ]

    coco_files = [
        p for p in files
        if p.name.endswith(".coco.json")
    ]

    return {
        "files": len(files),
        "images": len(images),
        "coco_files": len(coco_files),
        "image_paths": [str(p) for p in images],
        "coco_paths": [str(p) for p in coco_files],
    }


def request_and_download(name, project_id, version_number):

    print()
    print("=" * 80)
    print(f"{name.upper()}")
    print("=" * 80)

    destination = RAW / name / f"version_{version_number}"

    # Fresh destination for this run
    if destination.exists():
        print("Removing incomplete previous download:")
        print(destination)
        shutil.rmtree(destination)

    destination.mkdir(parents=True, exist_ok=True)

    # --------------------------------------------------------------------------
    # PROJECT
    # --------------------------------------------------------------------------

    project = workspace.project(project_id)

    print()
    print("Project:", project.name)
    print("Project ID:", project_id)
    print("Type:", project.type)

    # --------------------------------------------------------------------------
    # VERSION
    # --------------------------------------------------------------------------

    version = project.version(version_number)

    print()
    print("Version:", version.version)
    print("Images:", getattr(version, "images", None))
    print("Splits:", getattr(version, "splits", None))
    print("Exports currently listed:", getattr(version, "exports", None))

    # --------------------------------------------------------------------------
    # REQUEST EXPORT DIRECTLY
    #
    # IMPORTANT:
    # We DO NOT require version.exports to already contain COCO.
    # --------------------------------------------------------------------------

    print()
    print("Requesting COCO segmentation export directly...")

    export_result = version.export(
        model_format="coco-segmentation"
    )

    print("Export request result:", export_result)

    # --------------------------------------------------------------------------
    # WAIT + DOWNLOAD
    # --------------------------------------------------------------------------

    max_attempts = 8
    wait_seconds = 15

    dataset = None

    for attempt in range(1, max_attempts + 1):

        print()
        print("-" * 80)
        print(f"DOWNLOAD ATTEMPT {attempt}/{max_attempts}")
        print("-" * 80)

        if attempt > 1:
            print(f"Waiting {wait_seconds} seconds...")
            time.sleep(wait_seconds)

        try:

            print("Calling version.download()...")

            dataset = version.download(
                model_format="coco-segmentation",
                location=str(destination),
                overwrite=True,
            )

            print()
            print("Download returned:", type(dataset))

            inventory = inventory_dataset(destination)

            print()
            print("Files:", inventory["files"])
            print("Images:", inventory["images"])
            print("COCO JSON:", inventory["coco_files"])

            if inventory["images"] > 0 and inventory["coco_files"] > 0:

                print()
                print("✓ DATASET DOWNLOAD VERIFIED")

                return {
                    "status": "SUCCESS",
                    "project": project_id,
                    "project_name": project.name,
                    "version": version_number,
                    "version_images": getattr(version, "images", None),
                    "splits": getattr(version, "splits", None),
                    "type": project.type,
                    "destination": str(destination),
                    "inventory": inventory,
                }

            print()
            print("Download returned but required files are not present yet.")

        except Exception as exc:

            print()
            print("Download attempt failed:")
            print(repr(exc))

            if attempt == max_attempts:
                raise

    raise RuntimeError(
        f"{name}: export/download did not produce a usable COCO dataset "
        f"after {max_attempts} attempts."
    )


# ================================================================================
# 6. DOWNLOAD ALL FOUR DATASETS
# ================================================================================

results = {}

for name, cfg in DATASETS.items():

    try:

        results[name] = request_and_download(
            name=name,
            project_id=cfg["project"],
            version_number=cfg["version"],
        )

    except Exception as exc:

        print()
        print("=" * 80)
        print(f"{name.upper()} FAILED")
        print("=" * 80)

        print(repr(exc))
        traceback.print_exc()

        results[name] = {
            "status": "FAILED",
            "project": cfg["project"],
            "version": cfg["version"],
            "error": repr(exc),
        }


# ================================================================================
# 7. FINAL INVENTORY
# ================================================================================

print()
print("=" * 80)
print("FINAL MULTI-SOURCE INVENTORY")
print("=" * 80)

for name, result in results.items():

    print()
    print(name.upper())

    if result["status"] == "SUCCESS":

        inv = result["inventory"]

        print("  Status      : SUCCESS")
        print("  Version imgs:", result["version_images"])
        print("  Local files :", inv["files"])
        print("  Images      :", inv["images"])
        print("  COCO files  :", inv["coco_files"])
        print("  Destination :", result["destination"])

    else:

        print("  Status:", result["status"])
        print("  Error :", result["error"])


# ================================================================================
# 8. SAVE MANIFEST
# ================================================================================

manifest = {
    "stage": "8A",
    "description": "Robust multi-source dataset acquisition",
    "workspace": ROBOFLOW_WORKSPACE,

    "datasets": {
        name: {
            "project": cfg["project"],
            "version": cfg["version"],
            "expected_type": cfg["type"],
            "result": results[name],
        }
        for name, cfg in DATASETS.items()
    },

    "excluded_sources": {
        "pih_dark_spots": {
            "reason": "Private dataset; no permitted workspace access/export."
        }
    },

    "rules": {
        "class_mapping": False,
        "augmentation": False,
        "normalization": False,
        "resplitting": False,
        "training": False,
    },
}

manifest_path = AUDIT / "stage8a_acquisition_manifest.json"

with open(manifest_path, "w") as f:
    json.dump(
        manifest,
        f,
        indent=2,
    )


# ================================================================================
# 9. COMPLETE
# ================================================================================

print()
print("=" * 80)
print("STAGE 8A COMPLETE")
print("=" * 80)

print()
print("Raw multi-source directory:")
print(RAW)

print()
print("Manifest:")
print(manifest_path)

print()
print("NO CLASS MAPPING.")
print("NO AUGMENTATION.")
print("NO NORMALIZATION.")
print("NO RESPLITTING.")
print("NO TRAINING.")
print("OLD BASELINE WAS NOT MODIFIED.")
```

```python
# ================================================================================
# CLINDERMA STAGE 8B
# COMPREHENSIVE MULTI-SOURCE DATASET AUDIT
# ================================================================================
#
# PURPOSE:
#   Inspect all downloaded datasets before any semantic class mapping.
#
# DATASETS:
#   PIE
#   MELASMA
#   FACE SEGMENT
#   SKIN PROBLEM
#
# IMPORTANT:
#   - No data modification
#   - No class mapping
#   - No augmentation
#   - No normalization
#   - No resplitting
#   - No training
# ================================================================================

from pathlib import Path
from collections import Counter, defaultdict
import json
import math
import os

from PIL import Image


# ================================================================================
# PATHS
# ================================================================================

WORKSPACE = Path("/content/clinderma_workspace")
RAW = WORKSPACE / "raw_multisource"
AUDIT = WORKSPACE / "audit_multisource"

REPORT_PATH = AUDIT / "stage8b_dataset_audit.json"

DATASETS = {
    "PIE": RAW / "pie" / "version_1",
    "MELASMA": RAW / "melasma" / "version_1",
    "FACE_SEGMENT": RAW / "face_segment" / "version_1",
    "SKIN_PROBLEM": RAW / "skin_problem" / "version_1",
}


IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
}


# ================================================================================
# HELPERS
# ================================================================================

def find_image_files(directory):
    return sorted(
        p for p in directory.rglob("*")
        if p.is_file()
        and p.suffix.lower() in IMAGE_EXTENSIONS
    )


def find_coco_files(directory):
    return sorted(
        p for p in directory.rglob("*")
        if p.is_file()
        and p.name.lower().endswith(".coco.json")
    )


def load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def polygon_area(coords):
    """
    Shoelace formula for COCO polygon coordinates.
    """
    if len(coords) < 6:
        return 0.0

    points = [
        (coords[i], coords[i + 1])
        for i in range(0, len(coords) - 1, 2)
    ]

    if len(points) < 3:
        return 0.0

    area = 0.0

    for i in range(len(points)):
        x1, y1 = points[i]
        x2, y2 = points[(i + 1) % len(points)]

        area += x1 * y2
        area -= x2 * y1

    return abs(area) / 2.0


def inspect_image_dimensions(image_path):

    try:
        with Image.open(image_path) as img:

            width, height = img.size

            # Force image read to detect corrupt files
            img.verify()

        return {
            "valid": True,
            "width": width,
            "height": height,
        }

    except Exception as exc:

        return {
            "valid": False,
            "width": None,
            "height": None,
            "error": repr(exc),
        }


# ================================================================================
# START
# ================================================================================

print("=" * 80)
print("CLINDERMA STAGE 8B")
print("COMPREHENSIVE MULTI-SOURCE DATASET AUDIT")
print("=" * 80)

print()
print("Raw directory:")
print(RAW)

print()
print("Datasets:")
for name, path in DATASETS.items():
    print(f"  {name:15} {path}")

print()
print("No class mapping will be performed.")
print("No data will be modified.")
print("No training will be performed.")


# ================================================================================
# AUDIT FUNCTION
# ================================================================================

def audit_dataset(dataset_name, dataset_path):

    print()
    print("=" * 80)
    print(dataset_name)
    print("=" * 80)

    if not dataset_path.exists():

        print()
        print("ERROR: Dataset directory does not exist.")
        print(dataset_path)

        return {
            "status": "MISSING",
            "path": str(dataset_path),
        }


    # --------------------------------------------------------------------------
    # Locate files
    # --------------------------------------------------------------------------

    image_files = find_image_files(dataset_path)
    coco_files = find_coco_files(dataset_path)

    print()
    print("Images found:", len(image_files))
    print("COCO files :", len(coco_files))


    # --------------------------------------------------------------------------
    # Image validation
    # --------------------------------------------------------------------------

    valid_images = 0
    corrupt_images = 0

    image_dimensions = Counter()
    image_dimension_details = {}

    for image_path in image_files:

        info = inspect_image_dimensions(image_path)

        if info["valid"]:

            valid_images += 1

            dimension = (
                info["width"],
                info["height"],
            )

            image_dimensions[dimension] += 1

            image_dimension_details[
                image_path.name
            ] = {
                "width": info["width"],
                "height": info["height"],
            }

        else:

            corrupt_images += 1

            image_dimension_details[
                image_path.name
            ] = {
                "error": info["error"],
            }


    # --------------------------------------------------------------------------
    # COCO audit
    # --------------------------------------------------------------------------

    coco_reports = []

    total_annotations = 0
    total_empty_images = 0
    total_invalid_segmentations = 0
    total_invalid_bboxes = 0
    total_orphan_annotations = 0

    raw_class_annotation_counts = Counter()
    raw_class_image_counts = defaultdict(set)

    segmentation_types = Counter()

    class_area = Counter()

    annotation_image_map = defaultdict(list)

    all_image_ids = set()

    for coco_path in coco_files:

        print()
        print("-" * 80)
        print("COCO FILE")
        print(coco_path.relative_to(dataset_path))

        try:
            coco = load_json(coco_path)

        except Exception as exc:

            print("ERROR loading JSON:", repr(exc))

            coco_reports.append({
                "file": str(coco_path),
                "status": "INVALID_JSON",
                "error": repr(exc),
            })

            continue


        # ----------------------------------------------------------------------
        # Categories
        # ----------------------------------------------------------------------

        categories = {
            int(c["id"]): c.get("name", f"class_{c['id']}")
            for c in coco.get("categories", [])
        }

        print()
        print("Categories:")

        for cid, cname in categories.items():
            print(f"  {cid}: {cname}")


        # ----------------------------------------------------------------------
        # Images
        # ----------------------------------------------------------------------

        images = coco.get("images", [])

        image_ids = {
            int(img["id"])
            for img in images
            if "id" in img
        }

        all_image_ids.update(image_ids)

        image_id_to_info = {
            int(img["id"]): img
            for img in images
            if "id" in img
        }


        # ----------------------------------------------------------------------
        # Local image lookup
        # ----------------------------------------------------------------------

        local_by_name = {
            p.name: p
            for p in image_files
        }


        local_referenced = set()


        # ----------------------------------------------------------------------
        # Annotations
        # ----------------------------------------------------------------------

        annotations = coco.get("annotations", [])

        file_class_counts = Counter()
        file_class_images = defaultdict(set)

        invalid_segmentations = 0
        invalid_bboxes = 0

        for ann in annotations:

            total_annotations += 1

            ann_id = ann.get("id")
            category_id = ann.get("category_id")

            category_name = categories.get(
                category_id,
                f"UNKNOWN_{category_id}"
            )

            raw_class_annotation_counts[
                category_name
            ] += 1

            image_id = ann.get("image_id")

            if image_id in image_id_to_info:

                file_class_images[
                    category_name
                ].add(image_id)

                raw_class_image_counts[
                    category_name
                ].add(
                    (
                        dataset_name,
                        str(coco_path),
                        image_id,
                    )
                )

                filename = image_id_to_info[
                    image_id
                ].get("file_name")

                if filename:

                    filename_only = Path(filename).name

                    if filename_only in local_by_name:

                        local_referenced.add(
                            filename_only
                        )

                    else:

                        # Sometimes Roboflow paths contain
                        # split folders, so try suffix matching.
                        matches = [
                            p
                            for p in image_files
                            if p.name == filename_only
                        ]

                        if matches:
                            local_referenced.add(
                                filename_only
                            )


            else:

                total_orphan_annotations += 1


            # ------------------------------------------------------------------
            # Bounding box validation
            # ------------------------------------------------------------------

            bbox = ann.get("bbox")

            if bbox is not None:

                if (
                    not isinstance(bbox, list)
                    or len(bbox) != 4
                ):

                    invalid_bboxes += 1

                else:

                    try:

                        x, y, w, h = [
                            float(v)
                            for v in bbox
                        ]

                        if (
                            not math.isfinite(x)
                            or not math.isfinite(y)
                            or not math.isfinite(w)
                            or not math.isfinite(h)
                            or w < 0
                            or h < 0
                        ):

                            invalid_bboxes += 1

                    except Exception:

                        invalid_bboxes += 1


            # ------------------------------------------------------------------
            # Segmentation validation
            # ------------------------------------------------------------------

            segmentation = ann.get("segmentation")

            if segmentation is None:

                segmentation_types[
                    "missing"
                ] += 1

                invalid_segmentations += 1

            elif isinstance(segmentation, list):

                segmentation_types[
                    "polygon"
                ] += 1

                valid_polygon = False

                for polygon in segmentation:

                    if (
                        isinstance(polygon, list)
                        and len(polygon) >= 6
                        and len(polygon) % 2 == 0
                    ):

                        valid_polygon = True

                        area = polygon_area(
                            polygon
                        )

                        class_area[
                            category_name
                        ] += area

                if not valid_polygon:
                    invalid_segmentations += 1

            elif isinstance(segmentation, dict):

                segmentation_types[
                    "rle"
                ] += 1

                counts = segmentation.get("counts")
                size = segmentation.get("size")

                if counts is None or size is None:
                    invalid_segmentations += 1

            else:

                segmentation_types[
                    type(segmentation).__name__
                ] += 1

                invalid_segmentations += 1


            file_class_counts[
                category_name
            ] += 1


        # ----------------------------------------------------------------------
        # Empty images
        # ----------------------------------------------------------------------

        annotated_image_ids = {
            ann.get("image_id")
            for ann in annotations
            if ann.get("image_id") in image_ids
        }

        empty_images = (
            len(image_ids - annotated_image_ids)
        )

        total_empty_images += empty_images
        total_invalid_segmentations += invalid_segmentations
        total_invalid_bboxes += invalid_bboxes


        # ----------------------------------------------------------------------
        # Print file report
        # ----------------------------------------------------------------------

        print()
        print("Images:", len(images))
        print("Annotations:", len(annotations))
        print("Empty images:", empty_images)
        print("Invalid segmentations:", invalid_segmentations)
        print("Invalid bounding boxes:", invalid_bboxes)

        print()
        print("Segmentation types:")

        for seg_type, count in segmentation_types.items():
            print(f"  {seg_type}: {count}")


        print()
        print("Annotations per class:")

        for class_name, count in file_class_counts.items():
            print(
                f"  {class_name}: {count}"
            )


        coco_reports.append({
            "file": str(coco_path),

            "images": len(images),
            "annotations": len(annotations),

            "categories": categories,

            "empty_images": empty_images,

            "invalid_segmentations":
                invalid_segmentations,

            "invalid_bounding_boxes":
                invalid_bboxes,

            "orphan_annotations":
                total_orphan_annotations,

            "segmentation_types":
                dict(segmentation_types),

            "annotations_per_class":
                dict(file_class_counts),

            "images_per_class": {
                cls: len(ids)
                for cls, ids
                in file_class_images.items()
            },

            "annotation_area_by_class":
                dict(class_area),
        })


    # =============================================================================
    # DATASET SUMMARY
    # =============================================================================

    print()
    print("-" * 80)
    print("DATASET SUMMARY")
    print("-" * 80)

    print()
    print("Image inventory")
    print("  Total:", len(image_files))
    print("  Valid:", valid_images)
    print("  Corrupt:", corrupt_images)

    print()
    print("Image dimensions")

    for (w, h), count in image_dimensions.most_common():
        print(
            f"  {w} x {h}: {count}"
        )


    print()
    print("Raw classes")

    for class_name, count in (
        raw_class_annotation_counts.most_common()
    ):

        image_count = len(
            raw_class_image_counts[class_name]
        )

        print(
            f"  {class_name}: "
            f"{count} annotations | "
            f"{image_count} images"
        )


    print()
    print("Segmentation types")

    all_segmentation_types = Counter()

    for report in coco_reports:

        all_segmentation_types.update(
            report.get(
                "segmentation_types",
                {}
            )
        )

    for seg_type, count in (
        all_segmentation_types.items()
    ):

        print(
            f"  {seg_type}: {count}"
        )


    print()
    print("QC")

    print(
        "  Invalid segmentations:",
        total_invalid_segmentations
    )

    print(
        "  Invalid bounding boxes:",
        total_invalid_bboxes
    )

    print(
        "  Orphan annotations:",
        total_orphan_annotations
    )

    print(
        "  Empty COCO images:",
        total_empty_images
    )


    # =============================================================================
    # SPLIT SUMMARY
    # =============================================================================

    split_summary = {}

    for split_dir in sorted(
        p for p in dataset_path.iterdir()
        if p.is_dir()
    ):

        split_images = find_image_files(
            split_dir
        )

        split_coco = find_coco_files(
            split_dir
        )

        split_summary[
            split_dir.name
        ] = {
            "images": len(split_images),
            "coco_files": len(split_coco),
        }


    if split_summary:

        print()
        print("Directory split structure:")

        for split, info in split_summary.items():

            print(
                f"  {split}: "
                f"{info['images']} images | "
                f"{info['coco_files']} COCO files"
            )


    # =============================================================================
    # RETURN REPORT
    # =============================================================================

    return {
        "status": "OK",

        "path": str(dataset_path),

        "images": {
            "total": len(image_files),
            "valid": valid_images,
            "corrupt": corrupt_images,
        },

        "dimensions": {
            f"{w}x{h}": count
            for (w, h), count
            in image_dimensions.items()
        },

        "coco_files": len(coco_files),

        "raw_classes": {
            class_name: {
                "annotations": count,
                "images": len(
                    raw_class_image_counts[
                        class_name
                    ]
                ),
            }
            for class_name, count
            in raw_class_annotation_counts.items()
        },

        "segmentation_types": dict(
            all_segmentation_types
        ),

        "quality": {
            "invalid_segmentations":
                total_invalid_segmentations,

            "invalid_bounding_boxes":
                total_invalid_bboxes,

            "orphan_annotations":
                total_orphan_annotations,

            "empty_images":
                total_empty_images,
        },

        "splits": split_summary,

        "coco_reports": coco_reports,
    }


# ================================================================================
# RUN AUDIT
# ================================================================================

reports = {}

for dataset_name, dataset_path in DATASETS.items():

    reports[dataset_name] = audit_dataset(
        dataset_name,
        dataset_path
    )


# ================================================================================
# SAVE REPORT
# ================================================================================

final_report = {
    "stage": "8B",

    "description":
        "Comprehensive multi-source dataset audit",

    "datasets": reports,

    "target_classes": [
        "PIH",
        "PIE",
        "SCAR",
        "DARK_SPOT",
        "MELASMA",
    ],

    "semantic_mapping": None,

    "augmentation": False,

    "normalization": False,

    "resplitting": False,

    "training": False,

    "data_modified": False,
}


AUDIT.mkdir(parents=True, exist_ok=True)

with open(
    REPORT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        final_report,
        f,
        indent=2
    )


# ================================================================================
# FINAL SUMMARY
# ================================================================================

print()
print("=" * 80)
print("STAGE 8B COMPLETE")
print("=" * 80)

print()
print("Dataset summary:")

for dataset_name, report in reports.items():

    if report.get("status") != "OK":

        print(
            f"  {dataset_name}: "
            f"{report.get('status')}"
        )

        continue

    print()
    print(
        f"  {dataset_name}"
    )

    print(
        f"    Images       : "
        f"{report['images']['total']}"
    )

    print(
        f"    Valid images : "
        f"{report['images']['valid']}"
    )

    print(
        f"    Corrupt      : "
        f"{report['images']['corrupt']}"
    )

    print(
        f"    COCO files   : "
        f"{report['coco_files']}"
    )

    print(
        f"    Classes      : "
        f"{len(report['raw_classes'])}"
    )

print(f"    Annotations  : {sum(v['annotations'] for v in report['raw_classes'].values())}")


print()
print("Audit report:")
print(REPORT_PATH)

print()
print("NO DATA WAS MODIFIED.")
print("NO CLASS MAPPING WAS PERFORMED.")
print("NO AUGMENTATION WAS PERFORMED.")
print("NO NORMALIZATION WAS PERFORMED.")
print("NO RESPLITTING WAS PERFORMED.")
print("NO TRAINING WAS PERFORMED.")
```

```python
# ================================================================================
# CLINDERMA STAGE 8D
# UNIFIED MULTI-SOURCE SEGMENTATION DATASET CONSTRUCTION
# ================================================================================
#
# TARGET TAXONOMY
#   0 = background
#   1 = PIH
#   2 = PIE
#   3 = SCAR
#   4 = DARK_SPOT
#   5 = MELASMA
#
# APPROVED MAPPINGS ONLY
#
# PIE:
#   PIE-Annotation -> PIE
#
# MELASMA:
#   melasma -> MELASMA
#
# FACE_SEGMENT:
#   Dark spots -> DARK_SPOT
#   Scar       -> SCAR
#
# SKIN_PROBLEM:
#   No mappings
#
# IMPORTANT
#   Raw data is NOT modified.
#   No augmentation.
#   No normalization.
#   No resplitting.
#   No training.
#
# ================================================================================

import json
import shutil
from pathlib import Path
from collections import defaultdict, Counter

import numpy as np
from PIL import Image


# ================================================================================
# PATHS
# ================================================================================

WORKSPACE = Path("/content/clinderma_workspace")

RAW = WORKSPACE / "raw_multisource"

OUTPUT = (
    WORKSPACE
    / "processed"
    / "pigmentation_multisource"
)

REPORT_DIR = (
    WORKSPACE
    / "audit_multisource"
)

REPORT_PATH = (
    REPORT_DIR
    / "stage8d_unified_dataset_report.json"
)

MANIFEST_PATH = (
    OUTPUT
    / "dataset_manifest.json"
)

METADATA_PATH = (
    OUTPUT
    / "metadata.json"
)


# ================================================================================
# TARGET TAXONOMY
# ================================================================================

TARGET_CLASSES = {
    0: "background",
    1: "PIH",
    2: "PIE",
    3: "SCAR",
    4: "DARK_SPOT",
    5: "MELASMA",
}

TARGET_IDS = {
    name: idx
    for idx, name in TARGET_CLASSES.items()
}


# ================================================================================
# APPROVED SEMANTIC MAPPINGS
# ================================================================================

MAPPINGS = {

    "pie": {
        "PIE-Annotation": "PIE",
    },

    "melasma": {
        "melasma": "MELASMA",
    },

    "face_segment": {
        "Dark spots": "DARK_SPOT",
        "Scar": "SCAR",
    },

    "skin_problem": {
        # No approved mappings.
    },
}


# ================================================================================
# SOURCE DATASETS
# ================================================================================

DATASETS = {

    "pie": RAW / "pie" / "version_1",

    "melasma": RAW / "melasma" / "version_1",

    "face_segment": RAW / "face_segment" / "version_1",

    "skin_problem": RAW / "skin_problem" / "version_1",
}


# ================================================================================
# REMOVE PREVIOUS INCOMPLETE OUTPUT
# ================================================================================

if OUTPUT.exists():

    print(
        "Removing previous incomplete Stage 8D output:"
    )

    print(OUTPUT)

    shutil.rmtree(OUTPUT)


# ================================================================================
# CREATE OUTPUT DIRECTORIES
# ================================================================================

for split in [
    "train",
    "valid",
    "test"
]:

    (
        OUTPUT
        / split
        / "images"
    ).mkdir(
        parents=True,
        exist_ok=True
    )

    (
        OUTPUT
        / split
        / "masks"
    ).mkdir(
        parents=True,
        exist_ok=True
    )


REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================================
# RLE SUPPORT
# ================================================================================

try:

    from pycocotools import mask as mask_utils

    print("pycocotools: OK")

except ImportError:

    import subprocess
    import sys

    print(
        "Installing pycocotools..."
    )

    subprocess.check_call(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            "pycocotools",
        ]
    )

    from pycocotools import mask as mask_utils


# ================================================================================
# HELPERS
# ================================================================================

def load_json(path):

    with open(
        path,
        "r",
        encoding="utf-8"
    ) as f:

        return json.load(f)


def find_coco_files(dataset_path):

    return sorted(
        dataset_path.rglob(
            "_annotations.coco.json"
        )
    )


def find_image(dataset_path, split, file_name):

    candidates = [

        dataset_path / split / file_name,

        dataset_path / file_name,

    ]

    for path in candidates:

        if path.exists():

            return path


    basename = Path(file_name).name

    matches = list(
        dataset_path.rglob(
            basename
        )
    )

    if matches:

        return matches[0]


    return None


def polygon_mask(
    segmentation,
    width,
    height
):

    mask = Image.new(
        "L",
        (width, height),
        0
    )

    from PIL import ImageDraw

    draw = ImageDraw.Draw(mask)

    if isinstance(
        segmentation,
        list
    ):

        for polygon in segmentation:

            if not polygon:

                continue

            if len(polygon) < 6:

                continue

            points = []

            for i in range(
                0,
                len(polygon) - 1,
                2
            ):

                points.append(
                    (
                        polygon[i],
                        polygon[i + 1]
                    )
                )

            if len(points) >= 3:

                draw.polygon(
                    points,
                    fill=1
                )

    return np.asarray(
        mask,
        dtype=np.uint8
    )


def rle_mask(
    segmentation,
    width,
    height
):

    decoded = mask_utils.decode(
        segmentation
    )

    if decoded.ndim == 3:

        decoded = decoded[:, :, 0]

    return decoded.astype(
        np.uint8
    )


def segmentation_mask(
    segmentation,
    width,
    height
):

    if isinstance(
        segmentation,
        list
    ):

        return polygon_mask(
            segmentation,
            width,
            height
        )

    if isinstance(
        segmentation,
        dict
    ):

        return rle_mask(
            segmentation,
            width,
            height
        )

    raise ValueError(
        "Unsupported segmentation type: "
        + str(type(segmentation))
    )


# ================================================================================
# GLOBAL MANIFEST
# ================================================================================

manifest = {

    "stage":
        "8D",

    "target_classes":
        TARGET_CLASSES,

    "approved_mappings":
        MAPPINGS,

    "sources":
        [],

    "images":
        [],

    "statistics": {

        "total_images":
            0,

        "images_with_targets":
            0,

        "negative_images":
            0,

        "annotations_used":
            0,

        "annotations_ignored":
            0,

        "class_annotations":
            Counter(),

        "class_images":
            Counter(),
    },

    "policy": {

        "raw_data_modified":
            False,

        "augmentation":
            False,

        "normalization":
            False,

        "resplitting":
            False,

        "training":
            False,
    }
}


# ================================================================================
# PROCESS DATASETS
# ================================================================================

for dataset_name, dataset_path in DATASETS.items():

    print()
    print("=" * 80)
    print(dataset_name.upper())
    print("=" * 80)

    if not dataset_path.exists():

        print(
            "WARNING: dataset directory not found:"
        )

        print(dataset_path)

        continue


    coco_files = find_coco_files(
        dataset_path
    )

    if not coco_files:

        print(
            "WARNING: no COCO files found."
        )

        continue


    source_report = {

        "dataset":
            dataset_name,

        "source_images":
            0,

        "output_images":
            0,

        "target_annotations":
            0,

        "ignored_annotations":
            0,

        "target_classes":
            Counter(),

        "target_images":
            Counter(),

        "splits":
            {},
    }


    # ============================================================================
    # PROCESS EACH SPLIT
    # ============================================================================

    for coco_path in coco_files:

        split = coco_path.parent.name

        if split not in [
            "train",
            "valid",
            "test"
        ]:

            print(
                f"Skipping unknown split: {split}"
            )

            continue


        print()
        print(
            f"--- {dataset_name.upper()} / {split} ---"
        )


        coco = load_json(
            coco_path
        )


        images = {
            item["id"]: item
            for item in coco.get(
                "images",
                []
            )
        }


        categories = {
            item["id"]: item["name"]
            for item in coco.get(
                "categories",
                []
            )
        }


        annotations_by_image = defaultdict(list)


        for annotation in coco.get(
            "annotations",
            []
        ):

            annotations_by_image[
                annotation["image_id"]
            ].append(
                annotation
            )


        mapping = MAPPINGS.get(
            dataset_name,
            {}
        )


        split_report = {

            "source_images":
                len(images),

            "output_images":
                0,

            "images_with_targets":
                0,

            "negative_images":
                0,

            "target_annotations":
                0,

            "ignored_annotations":
                0,

            "target_classes":
                Counter(),

            "target_images":
                Counter(),
        }


        # ========================================================================
        # IMAGE LOOP
        # ========================================================================

        for image_id, image_info in images.items():

            source_report[
                "source_images"
            ] += 1


            file_name = image_info[
                "file_name"
            ]


            image_path = find_image(
                dataset_path,
                split,
                file_name
            )


            if image_path is None:

                print(
                    "WARNING: image not found:",
                    file_name
                )

                continue


            # ====================================================================
            # LOAD IMAGE
            # ====================================================================

            try:

                image = Image.open(
                    image_path
                ).convert("RGB")

            except Exception as e:

                print(
                    "WARNING: image failed:",
                    image_path
                )

                print(e)

                continue


            width, height = image.size


            # ====================================================================
            # CREATE UNIFIED MASK
            # ====================================================================

            unified_mask = np.zeros(
                (
                    height,
                    width
                ),
                dtype=np.uint8
            )


            used_annotations = []

            ignored_annotations = []

            image_classes = set()


            # ====================================================================
            # PROCESS ANNOTATIONS
            # ====================================================================

            for annotation in annotations_by_image.get(
                image_id,
                []
            ):

                category_id = annotation.get(
                    "category_id"
                )

                raw_class = categories.get(
                    category_id
                )


                target_class = mapping.get(
                    raw_class
                )


                # ----------------------------------------------------------------
                # Unmapped class
                # ----------------------------------------------------------------

                if target_class is None:

                    ignored_annotations.append(
                        annotation["id"]
                    )

                    continue


                target_id = TARGET_IDS[
                    target_class
                ]


                segmentation = annotation.get(
                    "segmentation"
                )


                if not segmentation:

                    ignored_annotations.append(
                        annotation["id"]
                    )

                    continue


                try:

                    ann_mask = segmentation_mask(
                        segmentation,
                        width,
                        height
                    )

                except Exception as e:

                    print(
                        "WARNING: segmentation failed"
                    )

                    print(
                        dataset_name,
                        split,
                        file_name,
                        e
                    )

                    ignored_annotations.append(
                        annotation["id"]
                    )

                    continue


                # ----------------------------------------------------------------
                # Dimension safety
                # ----------------------------------------------------------------

                if ann_mask.shape != (
                    height,
                    width
                ):

                    print(
                        "WARNING: mask dimension mismatch"
                    )

                    print(
                        dataset_name,
                        split,
                        file_name
                    )

                    ignored_annotations.append(
                        annotation["id"]
                    )

                    continue


                # ----------------------------------------------------------------
                # Apply target class
                #
                # Existing non-background pixels are preserved.
                # ----------------------------------------------------------------

                available = (
                    ann_mask > 0
                ) & (
                    unified_mask == 0
                )


                unified_mask[
                    available
                ] = target_id


                used_annotations.append(
                    annotation["id"]
                )


                image_classes.add(
                    target_class
                )


                split_report[
                    "target_annotations"
                ] += 1


                split_report[
                    "target_classes"
                ][
                    target_class
                ] += 1


                source_report[
                    "target_annotations"
                ] += 1


                source_report[
                    "target_classes"
                ][
                    target_class
                ] += 1


                manifest[
                    "statistics"
                ][
                    "annotations_used"
                ] += 1


            # ====================================================================
            # TARGET / NEGATIVE STATUS
            # ====================================================================

            has_target = (
                len(used_annotations) > 0
            )


            if has_target:

                split_report[
                    "images_with_targets"
                ] += 1


                source_report[
                    "target_images"
                ]


                for class_name in image_classes:

                    split_report[
                        "target_images"
                    ][
                        class_name
                    ] += 1


                    source_report[
                        "target_images"
                    ][
                        class_name
                    ] += 1


                    manifest[
                        "statistics"
                    ][
                        "class_images"
                    ][
                        class_name
                    ] += 1


                manifest[
                    "statistics"
                ][
                    "images_with_targets"
                ] += 1


            else:

                split_report[
                    "negative_images"
                ] += 1


                manifest[
                    "statistics"
                ][
                    "negative_images"
                ] += 1


            # ====================================================================
            # SAVE IMAGE
            # ====================================================================

            output_index = manifest[
                "statistics"
            ][
                "total_images"
            ]


            output_name = (
                f"{dataset_name}__"
                f"{split}__"
                f"{output_index:06d}"
            )


            output_image = (
                OUTPUT
                / split
                / "images"
                / f"{output_name}.jpg"
            )


            output_mask = (
                OUTPUT
                / split
                / "masks"
                / f"{output_name}.png"
            )


            image.save(
                output_image,
                "JPEG",
                quality=95
            )


            Image.fromarray(
                unified_mask
            ).save(
                output_mask
            )


            # ====================================================================
            # SAVE IMAGE METADATA
            # ====================================================================

            manifest[
                "images"
            ].append({

                "id":
                    output_name,

                "dataset":
                    dataset_name,

                "source_split":
                    split,

                "source_image_id":
                    image_id,

                "source_file":
                    str(image_path),

                "output_image":
                    str(output_image),

                "output_mask":
                    str(output_mask),

                "width":
                    width,

                "height":
                    height,

                "has_target":
                    has_target,

                "approved_classes":
                    sorted(
                        image_classes
                    ),

                "used_annotations":
                    used_annotations,

                "ignored_annotations":
                    ignored_annotations,
            })


            manifest[
                "statistics"
            ][
                "total_images"
            ] += 1


            manifest[
                "statistics"
            ][
                "annotations_ignored"
            ] += len(
                ignored_annotations
            )


            split_report[
                "output_images"
            ] += 1


            source_report[
                "output_images"
            ] += 1


        # ========================================================================
        # FINALIZE SPLIT REPORT
        # ========================================================================

        split_report[
            "target_classes"
        ] = dict(
            split_report[
                "target_classes"
            ]
        )


        split_report[
            "target_images"
        ] = dict(
            split_report[
                "target_images"
            ]
        )


        source_report[
            "splits"
        ][split] = split_report


        print(
            f"Images processed       : "
            f"{split_report['output_images']}"
        )

        print(
            f"Images with targets    : "
            f"{split_report['images_with_targets']}"
        )

        print(
            f"Negative images        : "
            f"{split_report['negative_images']}"
        )

        print(
            f"Target annotations     : "
            f"{split_report['target_annotations']}"
        )

        print(
            f"Ignored annotations    : "
            f"{split_report['ignored_annotations']}"
        )


    # ============================================================================
    # FINALIZE SOURCE REPORT
    # ============================================================================

    source_report[
        "target_classes"
    ] = dict(
        source_report[
            "target_classes"
        ]
    )


    source_report[
        "target_images"
    ] = dict(
        source_report[
            "target_images"
        ]
    )


    manifest[
        "sources"
    ].append(
        source_report
    )


# ================================================================================
# FINALIZE GLOBAL COUNTERS
# ================================================================================

manifest[
    "statistics"
][
    "class_annotations"
] = dict(
    manifest[
        "statistics"
    ][
        "class_annotations"
    ]
)


manifest[
    "statistics"
][
    "class_images"
] = dict(
    manifest[
        "statistics"
    ][
        "class_images"
    ]
)


# ================================================================================
# SAVE MANIFEST
# ================================================================================

with open(
    MANIFEST_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ================================================================================
# SAVE METADATA
# ================================================================================

metadata = {

    "stage":
        "8D",

    "target_classes":
        TARGET_CLASSES,

    "approved_mappings":
        MAPPINGS,

    "statistics":
        manifest["statistics"],

    "sources":
        manifest["sources"],

    "policy":
        manifest["policy"],
}


with open(
    METADATA_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=2
    )


# ================================================================================
# SAVE REPORT
# ================================================================================

with open(
    REPORT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ================================================================================
# FINAL OUTPUT
# ================================================================================

stats = manifest[
    "statistics"
]


print()
print("=" * 80)
print("CLINDERMA STAGE 8D COMPLETE")
print("=" * 80)


print()
print("TARGET CLASSES")

for class_id, class_name in TARGET_CLASSES.items():

    print(
        f"  {class_id}: {class_name}"
    )


print()
print("GLOBAL DATASET")

print(
    f"Total images        : "
    f"{stats['total_images']}"
)

print(
    f"Images with targets : "
    f"{stats['images_with_targets']}"
)

print(
    f"Negative images     : "
    f"{stats['negative_images']}"
)

print(
    f"Annotations used    : "
    f"{stats['annotations_used']}"
)

print(
    f"Annotations ignored : "
    f"{stats['annotations_ignored']}"
)


print()
print("CLASS ANNOTATIONS")

for class_id, class_name in TARGET_CLASSES.items():

    if class_id == 0:

        continue

    print(
        f"  {class_name:12s}: "
        f"{stats['class_annotations'].get(class_name, 0)}"
    )


print()
print("CLASS IMAGE COUNTS")

for class_id, class_name in TARGET_CLASSES.items():

    if class_id == 0:

        continue

    print(
        f"  {class_name:12s}: "
        f"{stats['class_images'].get(class_name, 0)}"
    )


print()
print("SOURCE SUMMARY")

for source in manifest[
    "sources"
]:

    print()
    print(
        source["dataset"].upper()
    )

    print(
        f"  Source images       : "
        f"{source['source_images']}"
    )

    print(
        f"  Output images       : "
        f"{source['output_images']}"
    )

    print(
        f"  Target annotations  : "
        f"{source['target_annotations']}"
    )

    print(
        f"  Ignored annotations : "
        f"{source['ignored_annotations']}"
    )

    print(
        "  Target classes:"
    )

    for class_name, count in source[
        "target_classes"
    ].items():

        print(
            f"    {class_name}: {count}"
        )


print()
print("=" * 80)
print("STAGE 8D COMPLETE")
print("=" * 80)


print()
print("Output:")
print(OUTPUT)


print()
print("Manifest:")
print(MANIFEST_PATH)


print()
print("Metadata:")
print(METADATA_PATH)


print()
print("Report:")
print(REPORT_PATH)


print()
print("RAW DATA WAS NOT MODIFIED.")
print("NO AUGMENTATION WAS PERFORMED.")
print("NO NORMALIZATION WAS PERFORMED.")
print("NO RESPLITTING WAS PERFORMED.")
print("NO TRAINING WAS PERFORMED.")
```

```python
# ================================================================================
# CLINDERMA STAGE 8E
# HYPERPIGMENTATION DATASET ACQUISITION + OBJECT-DETECTION AUDIT
# ================================================================================

from pathlib import Path
import json
import shutil
import time
from collections import Counter, defaultdict
from PIL import Image

from roboflow import Roboflow


# ================================================================================
# CONFIGURATION
# ================================================================================

WORKSPACE = "hanil-shah-s-workspace"

PROJECT_ID = "hyperpigmentation-ctmqm-j85yd"
VERSION_NUMBER = 1

DATASET_NAME = "hyperpigmentation"

RAW_ROOT = Path("/content/clinderma_workspace/raw_multisource")
AUDIT_ROOT = Path("/content/clinderma_workspace/audit_multisource")

DESTINATION = RAW_ROOT / DATASET_NAME / f"version_{VERSION_NUMBER}"

MANIFEST_PATH = (
    AUDIT_ROOT /
    "stage8e_hyperpigmentation_acquisition_manifest.json"
)

RAW_ROOT.mkdir(parents=True, exist_ok=True)
AUDIT_ROOT.mkdir(parents=True, exist_ok=True)


# ================================================================================
# HEADER
# ================================================================================

print("=" * 80)
print("CLINDERMA STAGE 8E")
print("HYPERPIGMENTATION DATASET ACQUISITION + OBJECT-DETECTION AUDIT")
print("=" * 80)

print()
print("Roboflow workspace:")
print(f"  {WORKSPACE}")

print()
print("Dataset:")
print(f"  {PROJECT_ID}")
print(f"  Version: {VERSION_NUMBER}")

print()
print("Destination:")
print(f"  {DESTINATION}")

print()
print("IMPORTANT:")
print("  This dataset is OBJECT DETECTION / BOUNDING BOX.")
print("  Bounding boxes will NOT be converted into segmentation masks.")

print()
print("No class mapping.")
print("No augmentation.")
print("No normalization.")
print("No resplitting.")
print("No training.")
print("Existing datasets will NOT be modified.")


# ================================================================================
# LOAD ROBOFLOW API KEY FROM COLAB SECRET
# ================================================================================

print()
print("=" * 80)
print("LOADING ROBOFLOW API KEY FROM COLAB SECRET")
print("=" * 80)

try:
    from google.colab import userdata

    ROBOFLOW_API_KEY = userdata.get("ROBOFLOW_API_KEY")

except Exception as e:

    raise RuntimeError(
        "\nCould not access the Colab Secret.\n"
        "Create a Colab secret named:\n"
        "  ROBOFLOW_API_KEY\n"
        "and enable notebook access for the secret."
    ) from e


if not ROBOFLOW_API_KEY:
    raise RuntimeError(
        "\nROBOFLOW_API_KEY was not found.\n"
        "Add it under Colab → Secrets."
    )

print()
print("✓ Roboflow API key loaded from Colab Secret")
print("  Key value is not displayed.")


# ================================================================================
# CONNECT TO ROBOFLOW
# ================================================================================

print()
print("=" * 80)
print("CONNECTING TO ROBOFLOW")
print("=" * 80)

rf = Roboflow(api_key=ROBOFLOW_API_KEY)

workspace = rf.workspace(WORKSPACE)

print()
print("✓ Roboflow workspace accessible")
print(f"Workspace: {workspace.name}")


# ================================================================================
# LOAD PROJECT
# ================================================================================

print()
print("=" * 80)
print("LOADING HYPERPIGMENTATION PROJECT")
print("=" * 80)

project = workspace.project(PROJECT_ID)

print()
print("Project:")
print(f"  Name : {project.name}")
print(f"  Type : {project.type}")
print(f"  ID   : {PROJECT_ID}")

version = project.version(VERSION_NUMBER)

print()
print("Version:")
print(f"  Name   : {version.name}")
print(f"  Version: {version.version}")
print(f"  Images : {getattr(version, 'images', 'unknown')}")
print(f"  Splits : {version.splits}")
print(f"  Exports: {version.exports}")


# ================================================================================
# PREPARE DESTINATION
# ================================================================================

print()
print("=" * 80)
print("PREPARING DOWNLOAD LOCATION")
print("=" * 80)

if DESTINATION.exists():

    existing_files = [
        p for p in DESTINATION.rglob("*")
        if p.is_file()
    ]

    if existing_files:

        print()
        print("Existing files found in destination:")
        print(f"  {len(existing_files)} files")

        print()
        print("Removing previous incomplete download...")

        shutil.rmtree(DESTINATION)

DESTINATION.mkdir(parents=True, exist_ok=True)

print()
print(f"✓ Destination ready:")
print(f"  {DESTINATION}")


# ================================================================================
# DIRECT ROBOFLOW DOWNLOAD
# ================================================================================

print()
print("=" * 80)
print("DIRECT ROBOFLOW OBJECT-DETECTION DOWNLOAD")
print("=" * 80)

print()
print("Roboflow reports:")
print(f"  Available exports: {version.exports}")

print()
print("Because no generated export is currently exposed,")
print("the pipeline will attempt the native Version.download()")
print("mechanism directly using object-detection formats.")


downloaded = False
download_error = None
dataset_object = None

# Try the formats most commonly used by Roboflow
candidate_formats = [
    "coco",
    "coco-detection",
    "yolov8",
    "yolov11",
    "yolov5",
]

for candidate_format in candidate_formats:

    if downloaded:
        break

    print()
    print("-" * 80)
    print(f"TRYING FORMAT: {candidate_format}")
    print("-" * 80)

    try:

        print("Calling version.download()...")

        dataset_object = version.download(
            model_format=candidate_format,
            location=str(DESTINATION),
            overwrite=True,
        )

        files_now = [
            p for p in DESTINATION.rglob("*")
            if p.is_file()
        ]

        if files_now:

            downloaded = True

            print()
            print("✓ DOWNLOAD SUCCESSFUL")
            print(f"Format: {candidate_format}")
            print(f"Files found: {len(files_now)}")

            selected_format = candidate_format

        else:

            print()
            print("Download returned but no files were found.")

    except Exception as e:

        download_error = repr(e)

        print()
        print("Format failed:")
        print(download_error)


# ================================================================================
# FINAL DOWNLOAD FAILURE
# ================================================================================

if not downloaded:

    print()
    print("=" * 80)
    print("DOWNLOAD FAILED")
    print("=" * 80)

    print()
    print("Roboflow project is accessible, but the API did not")
    print("provide a downloadable object-detection export.")

    manifest = {
        "stage": "8E",
        "dataset": DATASET_NAME,
        "workspace": WORKSPACE,
        "project_id": PROJECT_ID,
        "version": VERSION_NUMBER,
        "project_type": str(project.type),
        "available_exports": list(
            getattr(version, "exports", []) or []
        ),
        "status": "DOWNLOAD_FAILED",
        "last_error": download_error,
        "destination": str(DESTINATION),
        "data_modified": False,
        "training_performed": False,
    }

    with open(MANIFEST_PATH, "w") as f:
        json.dump(manifest, f, indent=2)

    raise RuntimeError(
        "\nHyperpigmentation dataset could not be downloaded.\n"
        f"Project: {PROJECT_ID}\n"
        f"Version: {VERSION_NUMBER}\n"
        f"Available exports: {version.exports}\n"
        f"Last error: {download_error}"
    )


# ================================================================================
# FILE INVENTORY
# ================================================================================

print()
print("=" * 80)
print("VERIFYING DOWNLOADED DATA")
print("=" * 80)

all_files = [
    p for p in DESTINATION.rglob("*")
    if p.is_file()
]

image_extensions = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp",
}

image_files = [
    p for p in all_files
    if p.suffix.lower() in image_extensions
]

json_files = [
    p for p in all_files
    if p.suffix.lower() == ".json"
]

txt_files = [
    p for p in all_files
    if p.suffix.lower() == ".txt"
]

print()
print(f"Total files      : {len(all_files)}")
print(f"Image files      : {len(image_files)}")
print(f"JSON files       : {len(json_files)}")
print(f"TXT files        : {len(txt_files)}")


# ================================================================================
# IMAGE AUDIT
# ================================================================================

print()
print("=" * 80)
print("IMAGE INVENTORY AUDIT")
print("=" * 80)

valid_images = []
corrupt_images = []

dimension_counts = Counter()

for image_path in image_files:

    try:

        with Image.open(image_path) as img:
            img.verify()

        with Image.open(image_path) as img:
            width, height = img.size

        valid_images.append(image_path)
        dimension_counts[(width, height)] += 1

    except Exception:

        corrupt_images.append(image_path)


print()
print(f"Total images   : {len(image_files)}")
print(f"Valid images   : {len(valid_images)}")
print(f"Corrupt images : {len(corrupt_images)}")

print()
print("Most common image dimensions:")

for (width, height), count in dimension_counts.most_common(20):

    print(
        f"  {width} x {height}: {count}"
    )


# ================================================================================
# COCO DETECTION AUDIT
# ================================================================================

coco_files = [
    p for p in json_files
    if (
        "_annotations.coco" in p.name.lower()
        or "coco" in p.name.lower()
    )
]

total_annotations = 0
invalid_boxes = 0
orphan_annotations = 0
empty_images = 0

category_counts = Counter()
category_images = defaultdict(set)

coco_reports = []

if coco_files:

    annotation_type = "COCO_OBJECT_DETECTION"

    print()
    print("=" * 80)
    print("COCO OBJECT-DETECTION AUDIT")
    print("=" * 80)

    for coco_path in sorted(coco_files):

        print()
        print("-" * 80)
        print("COCO FILE")
        print(coco_path.relative_to(DESTINATION))
        print("-" * 80)

        with open(coco_path, "r") as f:
            coco = json.load(f)

        categories = {
            c["id"]: c["name"]
            for c in coco.get("categories", [])
        }

        images = {
            img["id"]: img
            for img in coco.get("images", [])
        }

        annotations = coco.get("annotations", [])

        print()
        print("Categories:")

        for category_id, category_name in categories.items():

            print(
                f"  {category_id}: {category_name}"
            )

        annotated_image_ids = set()

        file_invalid_boxes = 0
        file_orphans = 0

        for annotation in annotations:

            total_annotations += 1

            image_id = annotation.get("image_id")
            category_id = annotation.get("category_id")

            if image_id not in images:

                orphan_annotations += 1
                file_orphans += 1
                continue

            if category_id not in categories:

                orphan_annotations += 1
                file_orphans += 1
                continue

            bbox = annotation.get("bbox")

            if not bbox or len(bbox) != 4:

                invalid_boxes += 1
                file_invalid_boxes += 1
                continue

            x, y, width, height = bbox

            image_width = images[image_id]["width"]
            image_height = images[image_id]["height"]

            valid_box = True

            if width <= 0 or height <= 0:
                valid_box = False

            if x < 0 or y < 0:
                valid_box = False

            if x + width > image_width:
                valid_box = False

            if y + height > image_height:
                valid_box = False

            if not valid_box:

                invalid_boxes += 1
                file_invalid_boxes += 1
                continue

            annotated_image_ids.add(image_id)

            category_name = categories[category_id]

            category_counts[category_name] += 1
            category_images[category_name].add(image_id)

        file_empty_images = (
            len(images) - len(annotated_image_ids)
        )

        empty_images += file_empty_images

        print()
        print(f"Images                : {len(images)}")
        print(f"Annotations           : {len(annotations)}")
        print(f"Empty images          : {file_empty_images}")
        print(f"Invalid bounding boxes: {file_invalid_boxes}")
        print(f"Orphan annotations    : {file_orphans}")

        print()
        print("Annotations per class:")

        for category_id, category_name in categories.items():

            print(
                f"  {category_name}: "
                f"{category_counts[category_name]} annotations | "
                f"{len(category_images[category_name])} images"
            )

        coco_reports.append({
            "file": str(
                coco_path.relative_to(DESTINATION)
            ),
            "categories": categories,
            "images": len(images),
            "annotations": len(annotations),
            "empty_images": file_empty_images,
            "invalid_bounding_boxes": file_invalid_boxes,
            "orphan_annotations": file_orphans,
        })


# ================================================================================
# YOLO DETECTION AUDIT
# ================================================================================

else:

    yolo_label_files = [
        p for p in txt_files
        if "readme" not in p.name.lower()
    ]

    if yolo_label_files:

        annotation_type = "YOLO_OBJECT_DETECTION"

        print()
        print("=" * 80)
        print("YOLO OBJECT-DETECTION AUDIT")
        print("=" * 80)

        print()
        print(f"Label files: {len(yolo_label_files)}")

        yolo_invalid = 0
        yolo_annotations = 0

        for label_path in yolo_label_files:

            try:

                with open(label_path, "r") as f:

                    lines = [
                        line.strip()
                        for line in f
                        if line.strip()
                    ]

                image_candidates = [
                    label_path.with_suffix(ext)
                    for ext in image_extensions
                ]

                image_path = next(
                    (
                        p for p in image_candidates
                        if p.exists()
                    ),
                    None
                )

                if image_path is None:
                    continue

                for line in lines:

                    parts = line.split()

                    if len(parts) != 5:

                        yolo_invalid += 1
                        continue

                    class_id = int(parts[0])

                    x_center = float(parts[1])
                    y_center = float(parts[2])
                    width = float(parts[3])
                    height = float(parts[4])

                    yolo_annotations += 1

                    if not (
                        0 <= x_center <= 1
                        and 0 <= y_center <= 1
                        and 0 < width <= 1
                        and 0 < height <= 1
                    ):

                        yolo_invalid += 1
                        continue

                    category_counts[str(class_id)] += 1
                    category_images[str(class_id)].add(
                        image_path.name
                    )

            except Exception:

                yolo_invalid += 1

        total_annotations = yolo_annotations
        invalid_boxes = yolo_invalid

        print()
        print(f"Annotations           : {total_annotations}")
        print(f"Invalid bounding boxes: {invalid_boxes}")

        print()
        print("Annotations per class:")

        for class_name, count in category_counts.items():

            print(
                f"  {class_name}: "
                f"{count} annotations | "
                f"{len(category_images[class_name])} images"
            )

    else:

        annotation_type = "UNKNOWN"

        print()
        print("=" * 80)
        print("NO RECOGNIZED OBJECT-DETECTION ANNOTATIONS")
        print("=" * 80)

        print()
        print("Images were downloaded, but no COCO or YOLO")
        print("object-detection labels were detected.")


# ================================================================================
# FINAL SUMMARY
# ================================================================================

print()
print("=" * 80)
print("STAGE 8E — FINAL SUMMARY")
print("=" * 80)

print()
print("Dataset:")
print("  Hyperpigmentation")

print()
print("Project:")
print(f"  {PROJECT_ID}")

print()
print("Version:")
print(f"  {VERSION_NUMBER}")

print()
print("Export:")
print(f"  {selected_format}")

print()
print("Annotation type:")
print(f"  {annotation_type}")

print()
print("Images:")
print(f"  Total downloaded : {len(image_files)}")
print(f"  Valid            : {len(valid_images)}")
print(f"  Corrupt          : {len(corrupt_images)}")

print()
print("Annotations:")
print(f"  Total            : {total_annotations}")

print()
print("QC:")
print(f"  Invalid boxes    : {invalid_boxes}")
print(f"  Orphan annotations: {orphan_annotations}")
print(f"  Empty images     : {empty_images}")

print()
print("Classes:")

for class_name, count in category_counts.most_common():

    print(
        f"  {class_name}: "
        f"{count} annotations | "
        f"{len(category_images[class_name])} images"
    )


# ================================================================================
# SAVE MANIFEST
# ================================================================================

manifest = {
    "stage": "8E",
    "dataset": DATASET_NAME,
    "workspace": WORKSPACE,
    "project_id": PROJECT_ID,
    "version": VERSION_NUMBER,
    "project_type": str(project.type),
    "selected_export": str(selected_format),
    "available_exports": list(
        getattr(version, "exports", []) or []
    ),
    "annotation_type": annotation_type,

    "images": {
        "total": len(image_files),
        "valid": len(valid_images),
        "corrupt": len(corrupt_images),
        "empty": empty_images,
    },

    "annotations": {
        "total": total_annotations,
        "invalid_bounding_boxes": invalid_boxes,
        "orphan_annotations": orphan_annotations,
    },

    "classes": {
        class_name: {
            "annotations": count,
            "images": len(category_images[class_name]),
        }
        for class_name, count in category_counts.items()
    },

    "coco_reports": coco_reports,

    "destination": str(DESTINATION),

    "data_modified": False,
    "class_mapping_performed": False,
    "mask_conversion_performed": False,
    "augmentation_performed": False,
    "normalization_performed": False,
    "resplitting_performed": False,
    "training_performed": False,
}

with open(MANIFEST_PATH, "w") as f:
    json.dump(manifest, f, indent=2)


# ================================================================================
# COMPLETE
# ================================================================================

print()
print("=" * 80)
print("STAGE 8E COMPLETE")
print("=" * 80)

print()
print("Download:")
print(f"  {DESTINATION}")

print()
print("Manifest:")
print(f"  {MANIFEST_PATH}")

print()
print("NO CLASS MAPPING.")
print("NO MASK CONVERSION.")
print("NO AUGMENTATION.")
print("NO NORMALIZATION.")
print("NO RESPLITTING.")
print("NO TRAINING.")
print("EXISTING DATA WAS NOT MODIFIED.")

print("=" * 80)
```

```python
# =============================================================================
# CLINDERMA STAGE 8F
# HYPERPIGMENTATION DATASET — FINAL QC + INTEGRATION ANALYSIS
# CORRECTED VERSION
# =============================================================================

from pathlib import Path
import json
from collections import Counter, defaultdict
from PIL import Image
import numpy as np

print("=" * 80)
print("CLINDERMA STAGE 8F")
print("HYPERPIGMENTATION DATASET — FINAL QC + INTEGRATION ANALYSIS")
print("=" * 80)

# -----------------------------------------------------------------------------
# PATHS
# -----------------------------------------------------------------------------

ROOT = Path("/content/clinderma_workspace")

HYPER_DIR = ROOT / "raw_multisource" / "hyperpigmentation" / "version_1"

REPORT_DIR = ROOT / "audit_multisource"
REPORT_DIR.mkdir(parents=True, exist_ok=True)

REPORT_PATH = REPORT_DIR / "stage8f_hyperpigmentation_integration_report.json"

if not HYPER_DIR.exists():
    raise RuntimeError(
        f"Hyperpigmentation dataset not found:\n{HYPER_DIR}"
    )

print("\nDataset:")
print(f"  {HYPER_DIR}")

# -----------------------------------------------------------------------------
# FIND COCO FILES
# -----------------------------------------------------------------------------

coco_files = sorted(HYPER_DIR.rglob("_annotations.coco.json"))

print("\n" + "=" * 80)
print("COCO FILE DISCOVERY")
print("=" * 80)

print(f"COCO annotation files found: {len(coco_files)}")

if not coco_files:
    raise RuntimeError("No COCO annotation files found.")

for path in coco_files:
    print(f"  {path.relative_to(HYPER_DIR)}")

# -----------------------------------------------------------------------------
# IMAGE VALIDATION
# -----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("IMAGE VALIDATION")
print("=" * 80)

image_extensions = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

all_images = [
    p for p in HYPER_DIR.rglob("*")
    if p.is_file() and p.suffix.lower() in image_extensions
]

valid_images = 0
corrupt_images = 0
image_sizes = Counter()

for image_path in all_images:
    try:
        with Image.open(image_path) as img:
            img.verify()

        with Image.open(image_path) as img:
            image_sizes[(img.width, img.height)] += 1

        valid_images += 1

    except Exception:
        corrupt_images += 1

print(f"Images found   : {len(all_images)}")
print(f"Valid images   : {valid_images}")
print(f"Corrupt images : {corrupt_images}")

# -----------------------------------------------------------------------------
# GLOBAL STATISTICS
# -----------------------------------------------------------------------------

total_images = 0
total_annotations = 0
total_invalid_boxes = 0
total_orphan_annotations = 0
total_empty_images = 0

class_annotation_counts = Counter()

# IMPORTANT:
# This MUST be a defaultdict(set), because we store image IDs.
class_image_counts = defaultdict(set)

split_summary = {}

invalid_box_records = []
empty_image_records = []
orphan_records = []

# -----------------------------------------------------------------------------
# COCO AUDIT
# -----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("COCO ANNOTATION AUDIT")
print("=" * 80)

for coco_path in coco_files:

    split_name = coco_path.parent.name

    print("\n" + "-" * 80)
    print(f"SPLIT: {split_name}")
    print("-" * 80)

    with open(coco_path, "r") as f:
        coco = json.load(f)

    categories = {
        int(c["id"]): c["name"]
        for c in coco.get("categories", [])
    }

    print("\nCategories:")
    for cid, cname in categories.items():
        print(f"  {cid}: {cname}")

    images = {
        int(img["id"]): img
        for img in coco.get("images", [])
    }

    annotations = coco.get("annotations", [])

    annotation_by_image = defaultdict(list)

    for ann in annotations:
        annotation_by_image[int(ann["image_id"])].append(ann)

    split_invalid_boxes = 0
    split_orphans = 0
    split_empty = 0

    split_class_annotations = Counter()
    split_class_images = defaultdict(set)

    # -------------------------------------------------------------------------
    # ANNOTATION CHECK
    # -------------------------------------------------------------------------

    for ann in annotations:

        ann_id = ann.get("id")
        image_id = ann.get("image_id")
        category_id = ann.get("category_id")

        # Check orphan annotation
        if image_id not in images:

            split_orphans += 1

            orphan_records.append({
                "split": split_name,
                "annotation_id": ann_id,
                "image_id": image_id,
            })

            continue

        class_name = categories.get(
            int(category_id),
            f"UNKNOWN_{category_id}"
        )

        split_class_annotations[class_name] += 1
        split_class_images[class_name].add(image_id)

        # ---------------------------------------------------------------------
        # BOUNDING BOX VALIDATION
        # ---------------------------------------------------------------------

        bbox = ann.get("bbox")

        valid_bbox = True

        if not isinstance(bbox, list) or len(bbox) != 4:
            valid_bbox = False

        else:
            try:
                x, y, w, h = map(float, bbox)

                if (
                    not np.isfinite(x)
                    or not np.isfinite(y)
                    or not np.isfinite(w)
                    or not np.isfinite(h)
                    or w <= 0
                    or h <= 0
                ):
                    valid_bbox = False

            except Exception:
                valid_bbox = False

        if not valid_bbox:

            split_invalid_boxes += 1

            invalid_box_records.append({
                "split": split_name,
                "annotation_id": ann_id,
                "image_id": image_id,
                "category_id": category_id,
                "bbox": bbox,
            })

    # -------------------------------------------------------------------------
    # EMPTY IMAGES
    # -------------------------------------------------------------------------

    for image_id, image_info in images.items():

        anns = annotation_by_image.get(image_id, [])

        if len(anns) == 0:

            split_empty += 1

            empty_image_records.append({
                "split": split_name,
                "image_id": image_id,
                "file_name": image_info.get("file_name"),
            })

    # -------------------------------------------------------------------------
    # GLOBAL COUNTS
    # -------------------------------------------------------------------------

    total_images += len(images)
    total_annotations += len(annotations)

    total_invalid_boxes += split_invalid_boxes
    total_orphan_annotations += split_orphans
    total_empty_images += split_empty

    for cname, count in split_class_annotations.items():

        class_annotation_counts[cname] += count

        # CORRECT:
        # class_image_counts[cname] is a SET
        class_image_counts[cname].update(
            split_class_images[cname]
        )

    # -------------------------------------------------------------------------
    # PRINT SPLIT SUMMARY
    # -------------------------------------------------------------------------

    print(f"\nImages                : {len(images)}")
    print(f"Annotations           : {len(annotations)}")
    print(f"Empty images          : {split_empty}")
    print(f"Invalid bounding boxes: {split_invalid_boxes}")
    print(f"Orphan annotations    : {split_orphans}")

    print("\nAnnotations per class:")

    for cname, count in split_class_annotations.items():

        print(
            f"  {cname}: "
            f"{count} annotations | "
            f"{len(split_class_images[cname])} images"
        )

    split_summary[split_name] = {
        "images": len(images),
        "annotations": len(annotations),
        "empty_images": split_empty,
        "invalid_bounding_boxes": split_invalid_boxes,
        "orphan_annotations": split_orphans,
        "class_annotations": dict(split_class_annotations),
        "class_images": {
            cname: len(ids)
            for cname, ids in split_class_images.items()
        },
    }

# -----------------------------------------------------------------------------
# SEMANTIC CLASS REVIEW
# -----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("SEMANTIC CLASS REVIEW")
print("=" * 80)

raw_classes = sorted(class_annotation_counts.keys())

print("\nRaw classes discovered:")

for cname in raw_classes:

    print(
        f"  {cname}: "
        f"{class_annotation_counts[cname]} annotations | "
        f"{len(class_image_counts[cname])} images"
    )

semantic_review = {}

for cname in raw_classes:

    normalized = cname.strip().lower()

    if normalized == "hyperpigmentation":

        semantic_review[cname] = {
            "proposed_semantic_class": "HYPERPIGMENTATION",
            "decision": "SAFE_NORMALIZATION",
            "reason": (
                "Capitalization-only duplicate of the same semantic label."
            ),
        }

    else:

        semantic_review[cname] = {
            "proposed_semantic_class": None,
            "decision": "REVIEW",
            "reason": (
                "No automatic semantic equivalence should be assumed."
            ),
        }

# -----------------------------------------------------------------------------
# CLINDERMA TARGET CLASSES
# -----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("RELATION TO CLINDERMA TARGET CLASSES")
print("=" * 80)

target_classes = {
    0: "background",
    1: "PIH",
    2: "PIE",
    3: "SCAR",
    4: "DARK_SPOT",
    5: "MELASMA",
}

print("\nCurrent segmentation targets:")

for cid, cname in target_classes.items():
    print(f"  {cid}: {cname}")

print("\nHyperpigmentation interpretation:")

print("  HYPERPIGMENTATION -> NOT automatically mapped to PIH")
print("  HYPERPIGMENTATION -> NOT automatically mapped to DARK_SPOT")
print("  HYPERPIGMENTATION -> NOT automatically mapped to MELASMA")

print("\nReason:")
print(
    "  The annotation label is broader than the specific "
    "Clinderma target classes."
)

# -----------------------------------------------------------------------------
# DATASET USABILITY
# -----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("DATASET USABILITY DECISION")
print("=" * 80)

usable_annotations = (
    total_annotations
    - total_invalid_boxes
    - total_orphan_annotations
)

print(f"\nTotal images        : {total_images}")
print(f"Total annotations   : {total_annotations}")
print(f"Usable annotations  : {usable_annotations}")
print(f"Invalid boxes       : {total_invalid_boxes}")
print(f"Orphan annotations : {total_orphan_annotations}")
print(f"Empty images        : {total_empty_images}")

print("\nDecision:")

print(
    "  ✓ Structurally usable as an OBJECT-DETECTION dataset."
)

print(
    "  ✓ Invalid bounding boxes should be excluded during future use."
)

print(
    "  ✓ Empty images may be retained as negatives if appropriate."
)

print(
    "  ✗ Bounding boxes will NOT be converted into segmentation masks."
)

print(
    "  ✗ Dataset will NOT yet be merged into the segmentation dataset."
)

print(
    "  ✗ Hyperpigmentation will NOT automatically become PIH, "
    "DARK_SPOT or MELASMA."
)

# -----------------------------------------------------------------------------
# FUTURE USE
# -----------------------------------------------------------------------------

future_use = {

    "primary_recommendation":
        "Keep the dataset as an auxiliary object-detection dataset "
        "until semantic correspondence with PIH/DARK_SPOT/MELASMA "
        "is established.",

    "segformer_training":
        "Do not include bounding-box annotations as segmentation "
        "ground truth.",

    "possible_future_use": [
        "auxiliary object-detection model",
        "pretraining for localization",
        "candidate-region proposal",
        "external validation of hyperpigmentation localization",
    ],

    "not_recommended": [
        "bbox-to-mask conversion",
        "automatic mapping to PIH",
        "automatic mapping to DARK_SPOT",
        "automatic mapping to MELASMA",
    ],
}

# -----------------------------------------------------------------------------
# REPORT
# -----------------------------------------------------------------------------

report = {

    "stage": "8F",

    "dataset": "Hyperpigmentation",

    "project_id":
        "hyperpigmentation-ctmqm-j85yd",

    "version": "1",

    "dataset_path": str(HYPER_DIR),

    "annotation_type":
        "COCO_OBJECT_DETECTION",

    "images": {
        "total": total_images,
        "valid_local_images": valid_images,
        "corrupt_local_images": corrupt_images,
        "empty_coco_images": total_empty_images,
    },

    "annotations": {
        "total": total_annotations,
        "usable": usable_annotations,
        "invalid_bounding_boxes": total_invalid_boxes,
        "orphan_annotations": total_orphan_annotations,
    },

    "raw_classes": {
        cname: {
            "annotations": class_annotation_counts[cname],
            "images": len(class_image_counts[cname]),
        }
        for cname in raw_classes
    },

    "semantic_review": semantic_review,

    "clinderma_target_classes": target_classes,

    "split_summary": split_summary,

    "invalid_box_records": invalid_box_records,

    "empty_image_records": empty_image_records,

    "orphan_records": orphan_records,

    "future_use": future_use,

    "data_modification": {
        "modified": False,
        "class_mapping": False,
        "mask_conversion": False,
        "augmentation": False,
        "normalization": False,
        "resplitting": False,
        "training": False,
    },
}

with open(REPORT_PATH, "w") as f:
    json.dump(report, f, indent=2)

# -----------------------------------------------------------------------------
# FINAL OUTPUT
# -----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("STAGE 8F COMPLETE")
print("=" * 80)

print("\nFINAL DATASET STATUS")

print(f"  Images          : {total_images}")
print(f"  Annotations     : {total_annotations}")
print(f"  Usable boxes    : {usable_annotations}")
print(f"  Invalid boxes   : {total_invalid_boxes}")
print(f"  Empty images    : {total_empty_images}")

print("\nCLASS STATUS")

for cname in raw_classes:

    print(
        f"  {cname}: "
        f"{class_annotation_counts[cname]} annotations | "
        f"{len(class_image_counts[cname])} images"
    )

print("\nSEMANTIC STATUS")
print("  Hyperpigmentation: BROAD CLASS — NOT MAPPED")

print("\nINTEGRATION STATUS")
print("  Segmentation training      : NOT INTEGRATED")
print("  Bounding boxes → masks     : NO")
print("  Existing datasets modified : NO")

print("\nReport:")
print(f"  {REPORT_PATH}")

print("\n" + "=" * 80)
print("NO DATA WAS MODIFIED.")
print("NO MASKS WERE CREATED FROM BOUNDING BOXES.")
print("NO CLASS MAPPING WAS PERFORMED.")
print("NO AUGMENTATION WAS PERFORMED.")
print("NO NORMALIZATION WAS PERFORMED.")
print("NO RESPLITTING WAS PERFORMED.")
print("NO TRAINING WAS PERFORMED.")
print("=" * 80)
```

```python
# =============================================================================
# CLINDERMA STAGE 8G
# BALANCED MULTI-SOURCE SEGMENTATION DATASET PREPARATION
# + SEPARATE HYPERPIGMENTATION DETECTION DATASET
# =============================================================================

from pathlib import Path
import json
import shutil
import random
from collections import defaultdict, Counter
from PIL import Image
import numpy as np

print("=" * 80)
print("CLINDERMA STAGE 8G")
print("BALANCED MULTI-SOURCE SEGMENTATION PREPARATION")
print("=" * 80)

# =============================================================================
# CONFIGURATION
# =============================================================================

ROOT = Path("/content/clinderma_workspace")

RAW_ROOT = ROOT / "raw_multisource"

OUTPUT_ROOT = ROOT / "processed" / "pigmentation_multisource_balanced"

DETECTION_OUTPUT = (
    ROOT
    / "processed"
    / "hyperpigmentation_detection"
)

REPORT_ROOT = ROOT / "audit_multisource"

REPORT_PATH = (
    REPORT_ROOT
    / "stage8g_balancing_report.json"
)

SEED = 42
random.seed(SEED)

# -------------------------------------------------------------------------
# TARGET CLASSES
# -------------------------------------------------------------------------

TARGET_CLASSES = {
    0: "background",
    1: "PIH",
    2: "PIE",
    3: "SCAR",
    4: "DARK_SPOT",
    5: "MELASMA",
}

print("\nTARGET CLASSES")

for cid, name in TARGET_CLASSES.items():
    print(f"  {cid}: {name}")

# =============================================================================
# SOURCE CONFIGURATION
# =============================================================================

SEGMENTATION_SOURCES = {

    "PIE": {
        "path": RAW_ROOT / "pie" / "version_1",
        "mapping": {
            "PIE-Annotation": "PIE"
        }
    },

    "MELASMA": {
        "path": RAW_ROOT / "melasma" / "version_1",
        "mapping": {
            "melasma": "MELASMA"
        }
    },

    "FACE_SEGMENT": {
        "path": RAW_ROOT / "face_segment" / "version_1",
        "mapping": {
            "Scar": "SCAR",
            "Dark spots": "DARK_SPOT"
        }
    },
}

# Skin problem is intentionally excluded because Stage 8D
# found no currently approved target-class mappings.

# Hyperpigmentation is intentionally excluded from segmentation.

# =============================================================================
# OUTPUT DIRECTORIES
# =============================================================================

if OUTPUT_ROOT.exists():
    print("\nRemoving previous incomplete Stage 8G output...")
    shutil.rmtree(OUTPUT_ROOT)

if DETECTION_OUTPUT.exists():
    print("Removing previous incomplete detection output...")
    shutil.rmtree(DETECTION_OUTPUT)

OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
DETECTION_OUTPUT.mkdir(parents=True, exist_ok=True)

for split in ["train", "valid", "test"]:
    (OUTPUT_ROOT / split / "images").mkdir(
        parents=True,
        exist_ok=True
    )

    (OUTPUT_ROOT / split / "masks").mkdir(
        parents=True,
        exist_ok=True
    )

# Detection structure
for split in ["train", "valid"]:
    (DETECTION_OUTPUT / split / "images").mkdir(
        parents=True,
        exist_ok=True
    )

# =============================================================================
# DISCOVER COCO FILES
# =============================================================================

def find_coco_files(dataset_path):

    return sorted(
        dataset_path.rglob("_annotations.coco.json")
    )

# =============================================================================
# LOAD SEGMENTATION ANNOTATIONS
# =============================================================================

records = []

print("\n" + "=" * 80)
print("COLLECTING APPROVED SEGMENTATION DATA")
print("=" * 80)

for source_name, cfg in SEGMENTATION_SOURCES.items():

    dataset_path = cfg["path"]

    if not dataset_path.exists():
        raise RuntimeError(
            f"Dataset missing:\n{dataset_path}"
        )

    print(f"\nSOURCE: {source_name}")

    coco_files = find_coco_files(dataset_path)

    if not coco_files:
        raise RuntimeError(
            f"No COCO annotations found:\n{dataset_path}"
        )

    for coco_path in coco_files:

        split = coco_path.parent.name

        if split not in ["train", "valid", "test"]:
            continue

        with open(coco_path, "r") as f:
            coco = json.load(f)

        categories = {
            int(c["id"]): c["name"]
            for c in coco.get("categories", [])
        }

        images = {
            int(img["id"]): img
            for img in coco.get("images", [])
        }

        annotations_by_image = defaultdict(list)

        for ann in coco.get("annotations", []):

            image_id = int(ann["image_id"])

            if image_id not in images:
                continue

            category_name = categories.get(
                int(ann["category_id"])
            )

            target_name = cfg["mapping"].get(
                category_name
            )

            if target_name is None:
                continue

            annotations_by_image[image_id].append(
                ann
            )

        for image_id, anns in annotations_by_image.items():

            image_info = images[image_id]

            records.append({
                "source": source_name,
                "split": split,
                "image_id": image_id,
                "file_name": image_info["file_name"],
                "dataset_path": str(dataset_path),
                "target_class": cfg["mapping"].get(
                    categories.get(
                        int(anns[0]["category_id"])
                    )
                ),
                "annotation_count": len(anns),
                "annotations": anns,
            })

# =============================================================================
# IMAGE-LEVEL CLASS COUNTS
# =============================================================================

print("\n" + "=" * 80)
print("IMAGE-LEVEL CLASS DISTRIBUTION")
print("=" * 80)

class_records = defaultdict(list)

for record in records:

    class_name = record["target_class"]

    if class_name in TARGET_CLASSES.values():
        class_records[class_name].append(record)

for class_name in [
    "PIH",
    "PIE",
    "SCAR",
    "DARK_SPOT",
    "MELASMA"
]:

    print(
        f"  {class_name:10s}: "
        f"{len(class_records[class_name])} image-records"
    )

# =============================================================================
# IMPORTANT:
# DO NOT FORCE BALANCE YET
# =============================================================================

print("\n" + "=" * 80)
print("BALANCING POLICY")
print("=" * 80)

print("""
Stage 8G does NOT blindly discard data.

The dataset is first organized by class and source.

Final balancing will be performed through:
  1. class-aware sampling
  2. training-only augmentation
  3. source-aware sampling
  4. minority-class oversampling through augmentation
  5. majority-class sampling caps where necessary

Validation and test data remain untouched.

No synthetic images are created in this stage.
""")

# =============================================================================
# SOURCE DISTRIBUTION
# =============================================================================

source_class_counts = defaultdict(Counter)

for record in records:

    source = record["source"]
    cls = record["target_class"]

    source_class_counts[source][cls] += 1

print("\nSOURCE / CLASS DISTRIBUTION")

for source, counts in source_class_counts.items():

    print(f"\n{source}")

    for cls, count in counts.items():
        print(f"  {cls:10s}: {count}")

# =============================================================================
# CHECK FOR DUPLICATE IMAGE PATHS
# =============================================================================

print("\n" + "=" * 80)
print("DUPLICATE IMAGE CHECK")
print("=" * 80)

image_keys = defaultdict(list)

for record in records:

    key = (
        record["source"],
        record["split"],
        record["file_name"]
    )

    image_keys[key].append(record)

duplicate_groups = {
    key: value
    for key, value in image_keys.items()
    if len(value) > 1
}

print(
    f"Duplicate annotation groups: "
    f"{len(duplicate_groups)}"
)

# =============================================================================
# COPY ONLY APPROVED SEGMENTATION IMAGES
# =============================================================================

print("\n" + "=" * 80)
print("BUILDING SEGMENTATION WORKING SET")
print("=" * 80)

copied = Counter()

for index, record in enumerate(records):

    source = record["source"]
    split = record["split"]
    class_name = record["target_class"]

    source_path = Path(record["dataset_path"])

    image_path = source_path / record["file_name"]

    # Robust fallback in case file_name contains a split directory.
    if not image_path.exists():
        candidates = list(
            source_path.rglob(
                Path(record["file_name"]).name
            )
        )

        if not candidates:
            print(
                f"WARNING: image not found: "
                f"{record['file_name']}"
            )
            continue

        image_path = candidates[0]

    output_name = (
        f"{source}_{split}_{index:06d}"
        f"{image_path.suffix.lower()}"
    )

    destination = (
        OUTPUT_ROOT
        / split
        / "images"
        / output_name
    )

    shutil.copy2(
        image_path,
        destination
    )

    copied[split] += 1

print("\nImages copied:")

for split in ["train", "valid", "test"]:
    print(
        f"  {split:5s}: "
        f"{copied[split]}"
    )

# =============================================================================
# HYPERPIGMENTATION DETECTION DATASET
# =============================================================================

print("\n" + "=" * 80)
print("HYPERPIGMENTATION DETECTION DATASET")
print("=" * 80)

HYPER_DIR = (
    RAW_ROOT
    / "hyperpigmentation"
    / "version_1"
)

if not HYPER_DIR.exists():

    print(
        "Hyperpigmentation dataset not found."
    )

else:

    hyper_coco_files = find_coco_files(
        HYPER_DIR
    )

    hyper_records = []

    for coco_path in hyper_coco_files:

        split = coco_path.parent.name

        if split not in ["train", "valid"]:
            continue

        with open(coco_path, "r") as f:
            coco = json.load(f)

        categories = {
            int(c["id"]): c["name"]
            for c in coco.get("categories", [])
        }

        images = {
            int(img["id"]): img
            for img in coco.get("images", [])
        }

        valid_annotations = defaultdict(list)

        for ann in coco.get("annotations", []):

            image_id = int(
                ann["image_id"]
            )

            if image_id not in images:
                continue

            bbox = ann.get("bbox")

            if (
                not isinstance(bbox, list)
                or len(bbox) != 4
            ):
                continue

            try:
                x, y, w, h = map(
                    float,
                    bbox
                )

                if (
                    not np.isfinite(x)
                    or not np.isfinite(y)
                    or not np.isfinite(w)
                    or not np.isfinite(h)
                    or w <= 0
                    or h <= 0
                ):
                    continue

            except Exception:
                continue

            valid_annotations[image_id].append(
                ann
            )

        for image_id, anns in valid_annotations.items():

            image_info = images[image_id]

            hyper_records.append({
                "split": split,
                "image_id": image_id,
                "file_name": image_info["file_name"],
                "annotations": anns,
                "categories": categories,
            })

    # Copy detection images only.
    detection_counts = Counter()

    for index, record in enumerate(hyper_records):

        split = record["split"]

        image_path = (
            HYPER_DIR
            / record["file_name"]
        )

        if not image_path.exists():

            candidates = list(
                HYPER_DIR.rglob(
                    Path(
                        record["file_name"]
                    ).name
                )
            )

            if not candidates:
                continue

            image_path = candidates[0]

        output_name = (
            f"hyper_{split}_{index:06d}"
            f"{image_path.suffix.lower()}"
        )

        destination = (
            DETECTION_OUTPUT
            / split
            / "images"
            / output_name
        )

        shutil.copy2(
            image_path,
            destination
        )

        detection_counts[split] += 1

    print(
        f"Detection train images : "
        f"{detection_counts['train']}"
    )

    print(
        f"Detection valid images : "
        f"{detection_counts['valid']}"
    )

# =============================================================================
# SAVE MANIFEST
# =============================================================================

manifest = {

    "stage": "8G",

    "target_classes": TARGET_CLASSES,

    "segmentation_sources": {
        name: {
            "path": str(cfg["path"]),
            "mapping": cfg["mapping"],
        }
        for name, cfg in SEGMENTATION_SOURCES.items()
    },

    "segmentation_records": len(records),

    "class_distribution": {
        cls: len(class_records[cls])
        for cls in [
            "PIH",
            "PIE",
            "SCAR",
            "DARK_SPOT",
            "MELASMA"
        ]
    },

    "source_class_distribution": {
        source: dict(counts)
        for source, counts
        in source_class_counts.items()
    },

    "hyperpigmentation": {
        "type": "object_detection",
        "integrated_into_segmentation": False,
        "mask_conversion": False,
    },

    "balancing": {
        "completed": False,
        "strategy": [
            "class-aware sampling",
            "source-aware sampling",
            "training-only augmentation",
            "minority oversampling",
            "majority sampling cap",
        ]
    },

    "data_modification": {
        "raw_data_modified": False,
        "raw_annotations_modified": False,
        "augmentation": False,
        "normalization": False,
        "resplitting": False,
        "training": False,
    }
}

manifest_path = (
    OUTPUT_ROOT
    / "stage8g_manifest.json"
)

with open(manifest_path, "w") as f:
    json.dump(
        manifest,
        f,
        indent=2
    )

# =============================================================================
# REPORT
# =============================================================================

report = {

    "stage": "8G",

    "segmentation_output":
        str(OUTPUT_ROOT),

    "detection_output":
        str(DETECTION_OUTPUT),

    "class_distribution":
        manifest["class_distribution"],

    "source_class_distribution":
        manifest["source_class_distribution"],

    "duplicate_groups":
        len(duplicate_groups),

    "policy": {
        "blind_equal_image_subsampling": False,
        "training_only_augmentation": True,
        "class_aware_sampling": True,
        "source_aware_sampling": True,
        "validation_untouched": True,
        "test_untouched": True,
    },

    "hyperpigmentation": {
        "kept_separate": True,
        "annotation_type":
            "COCO_OBJECT_DETECTION",
        "bbox_to_mask": False,
    },

    "outputs": {
        "segmentation":
            str(OUTPUT_ROOT),
        "detection":
            str(DETECTION_OUTPUT),
        "manifest":
            str(manifest_path),
    }
}

with open(REPORT_PATH, "w") as f:
    json.dump(
        report,
        f,
        indent=2
    )

# =============================================================================
# FINAL
# =============================================================================

print("\n" + "=" * 80)
print("STAGE 8G COMPLETE")
print("=" * 80)

print("\nSEGMENTATION WORKING SET")
print(
    f"  Train images : {copied['train']}"
)
print(
    f"  Valid images : {copied['valid']}"
)
print(
    f"  Test images  : {copied['test']}"
)

print("\nCLASS DISTRIBUTION")

for cls in [
    "PIH",
    "PIE",
    "SCAR",
    "DARK_SPOT",
    "MELASMA"
]:

    print(
        f"  {cls:10s}: "
        f"{len(class_records[cls])}"
    )

print("\nHYPERPIGMENTATION")
print(
    "  Kept as separate object-detection dataset"
)
print(
    "  Bounding boxes NOT converted to masks"
)

print("\nIMPORTANT")
print(
    "  Final balancing has NOT yet been applied."
)
print(
    "  Augmentation has NOT yet been applied."
)
print(
    "  Training has NOT been performed."
)

print("\nManifest:")
print(f"  {manifest_path}")

print("\nReport:")
print(f"  {REPORT_PATH}")

print("\n" + "=" * 80)
print("RAW DATA WAS NOT MODIFIED.")
print("NO CLASS REMAPPING WAS PERFORMED.")
print("NO BOUNDING BOXES WERE CONVERTED TO MASKS.")
print("NO AUGMENTATION WAS PERFORMED.")
print("NO TRAINING WAS PERFORMED.")
print("=" * 80)
```

```python
# =============================================================================
# CLINDERMA STAGE 8H
# TRAINING DATA BALANCING + AUGMENTATION PREPARATION
# =============================================================================

from pathlib import Path
import json
import shutil
import random
from collections import defaultdict, Counter

import numpy as np
from PIL import Image, ImageEnhance, ImageOps, ImageFilter

print("=" * 80)
print("CLINDERMA STAGE 8H")
print("TRAINING DATA BALANCING + AUGMENTATION PREPARATION")
print("=" * 80)

# =============================================================================
# PATHS
# =============================================================================

ROOT = Path("/content/clinderma_workspace")

INPUT_ROOT = (
    ROOT / "processed" /
    "pigmentation_multisource_balanced"
)

OUTPUT_ROOT = (
    ROOT / "processed" /
    "pigmentation_multisource_balanced_v2"
)

REPORT_ROOT = ROOT / "audit_multisource"

REPORT_PATH = (
    REPORT_ROOT /
    "stage8h_balanced_training_report.json"
)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

# =============================================================================
# TARGET CLASSES
# =============================================================================

TARGET_CLASSES = {
    0: "background",
    1: "PIH",
    2: "PIE",
    3: "SCAR",
    4: "DARK_SPOT",
    5: "MELASMA",
}

print("\nTARGET CLASSES")

for cid, name in TARGET_CLASSES.items():
    print(f"  {cid}: {name}")

# =============================================================================
# IMPORTANT POLICY
# =============================================================================

print("""
================================================================================
BALANCING POLICY
================================================================================

PIH:
  No approved training images currently exist.
  PIH will NOT be fabricated or mapped from broad pigmentation labels.

PIE:
  Retain the available real images.

MELASMA:
  Retain the available real images.

SCAR:
  Minority class.
  Additional training exposure will be created through augmentation.

DARK_SPOT:
  Severe minority class.
  Additional training exposure will be created through augmentation.

VALIDATION / TEST:
  NO augmentation.
  NO oversampling.
  NO balancing.
  NO modification.

RAW DATA:
  Never modified.
""")

# =============================================================================
# VERIFY INPUT
# =============================================================================

if not INPUT_ROOT.exists():
    raise RuntimeError(
        f"Stage 8G output was not found:\n{INPUT_ROOT}"
    )

INPUT_IMAGES = INPUT_ROOT / "train" / "images"

if not INPUT_IMAGES.exists():
    raise RuntimeError(
        f"Training images not found:\n{INPUT_IMAGES}"
    )

# =============================================================================
# CLEAN PREVIOUS OUTPUT
# =============================================================================

if OUTPUT_ROOT.exists():
    print(
        "\nRemoving previous incomplete Stage 8H output:"
    )
    print(OUTPUT_ROOT)
    shutil.rmtree(OUTPUT_ROOT)

for split in ["train", "valid", "test"]:

    (
        OUTPUT_ROOT /
        split /
        "images"
    ).mkdir(
        parents=True,
        exist_ok=True
    )

# =============================================================================
# READ STAGE 8G MANIFEST
# =============================================================================

MANIFEST_PATH = (
    INPUT_ROOT /
    "stage8g_manifest.json"
)

if not MANIFEST_PATH.exists():
    raise RuntimeError(
        f"Stage 8G manifest not found:\n{MANIFEST_PATH}"
    )

with open(MANIFEST_PATH, "r") as f:
    stage8g_manifest = json.load(f)

class_distribution = (
    stage8g_manifest.get(
        "class_distribution",
        {}
    )
)

print("\nStage 8G class distribution:")

for cls in [
    "PIH",
    "PIE",
    "SCAR",
    "DARK_SPOT",
    "MELASMA"
]:

    print(
        f"  {cls:10s}: "
        f"{class_distribution.get(cls, 0)}"
    )

# =============================================================================
# IMPORTANT DESIGN DECISION
# =============================================================================
#
# We do NOT create a mathematically equal dataset.
#
# Instead:
#
# PIE / MELASMA -> retain real images
# SCAR          -> moderate augmentation
# DARK_SPOT     -> stronger augmentation
#
# This avoids:
#   - throwing away hundreds of useful images
#   - excessive duplication
#   - extreme oversampling of 48 DARK_SPOT images
#
# Target exposure levels are deliberately moderate.
# =============================================================================

REAL_COUNTS = {
    "PIH": class_distribution.get("PIH", 0),
    "PIE": class_distribution.get("PIE", 0),
    "SCAR": class_distribution.get("SCAR", 0),
    "DARK_SPOT": class_distribution.get("DARK_SPOT", 0),
    "MELASMA": class_distribution.get("MELASMA", 0),
}

# Maximum number of training examples we want to expose
# for each minority class after augmentation.
#
# These are NOT fake labels.
# They are augmented views of existing labeled images.

AUGMENTED_TARGETS = {
    "SCAR": 350,
    "DARK_SPOT": 250,
}

# =============================================================================
# CLASSIFICATION OF STAGE 8G FILES
# =============================================================================

def infer_class_from_filename(filename):

    upper = filename.upper()

    if "PIE" in upper:
        return "PIE"

    if "MELASMA" in upper:
        return "MELASMA"

    if "FACE_SEGMENT" in upper:
        # FACE_SEGMENT files need class determination from
        # their source annotation metadata. Since Stage 8G
        # filenames alone may contain mixed target classes,
        # this function intentionally does not guess.
        return None

    return None


# =============================================================================
# READ RECORDS FROM STAGE 8G MANIFEST
# =============================================================================

# Stage 8G stores the aggregate information but not a complete
# per-output-image annotation table. Therefore, we reconstruct
# class membership from the original datasets and the manifest.

RAW_ROOT = ROOT / "raw_multisource"

SOURCE_CONFIG = {

    "PIE": {
        "path": RAW_ROOT / "pie" / "version_1",
        "class": "PIE",
    },

    "MELASMA": {
        "path": RAW_ROOT / "melasma" / "version_1",
        "class": "MELASMA",
    },

    "FACE_SEGMENT": {
        "path": RAW_ROOT / "face_segment" / "version_1",
        "classes": {
            "Scar": "SCAR",
            "Dark spots": "DARK_SPOT",
        },
    },
}

# =============================================================================
# COCO READER
# =============================================================================

def find_coco_files(path):
    return sorted(
        path.rglob("_annotations.coco.json")
    )


class_image_files = defaultdict(list)

print("\n" + "=" * 80)
print("RECONSTRUCTING CLASS-SPECIFIC TRAINING FILES")
print("=" * 80)

for source_name, cfg in SOURCE_CONFIG.items():

    dataset_path = cfg["path"]

    if not dataset_path.exists():
        raise RuntimeError(
            f"Missing source dataset:\n{dataset_path}"
        )

    coco_files = find_coco_files(dataset_path)

    for coco_path in coco_files:

        split = coco_path.parent.name

        # Stage 8H balances TRAIN ONLY.
        if split != "train":
            continue

        with open(coco_path, "r") as f:
            coco = json.load(f)

        categories = {
            int(c["id"]): c["name"]
            for c in coco.get("categories", [])
        }

        images = {
            int(img["id"]): img
            for img in coco.get("images", [])
        }

        image_classes = defaultdict(set)

        for ann in coco.get("annotations", []):

            image_id = int(
                ann["image_id"]
            )

            category_id = int(
                ann["category_id"]
            )

            raw_class = categories.get(
                category_id
            )

            if source_name == "PIE":

                target = (
                    "PIE"
                    if raw_class == "PIE-Annotation"
                    else None
                )

            elif source_name == "MELASMA":

                target = (
                    "MELASMA"
                    if raw_class == "melasma"
                    else None
                )

            else:

                target = cfg["classes"].get(
                    raw_class
                )

            if target:
                image_classes[
                    image_id
                ].add(target)

        for image_id, classes in image_classes.items():

            image_info = images.get(
                image_id
            )

            if image_info is None:
                continue

            filename = image_info[
                "file_name"
            ]

            candidates = list(
                dataset_path.rglob(
                    Path(filename).name
                )
            )

            if not candidates:
                continue

            image_path = candidates[0]

            # If an image contains multiple target classes,
            # retain it for each relevant class.
            for cls in classes:

                class_image_files[
                    cls
                ].append(
                    image_path
                )

# Remove duplicate paths per class
for cls in class_image_files:

    class_image_files[cls] = list(
        dict.fromkeys(
            class_image_files[cls]
        )
    )

# =============================================================================
# PRINT REAL DISTRIBUTION
# =============================================================================

print("\nREAL TRAINING IMAGES")

for cls in [
    "PIH",
    "PIE",
    "SCAR",
    "DARK_SPOT",
    "MELASMA"
]:

    print(
        f"  {cls:10s}: "
        f"{len(class_image_files[cls])}"
    )

# =============================================================================
# AUGMENTATION FUNCTIONS
# =============================================================================

def augment_image(image, seed):

    rng = random.Random(seed)

    # Horizontal flip
    if rng.random() < 0.5:
        image = ImageOps.mirror(image)

    # Small rotation
    angle = rng.uniform(
        -10,
        10
    )

    image = image.rotate(
        angle,
        resample=Image.Resampling.BILINEAR,
        expand=False
    )

    # Mild brightness variation
    brightness = rng.uniform(
        0.90,
        1.10
    )

    image = ImageEnhance.Brightness(
        image
    ).enhance(
        brightness
    )

    # Mild contrast variation
    contrast = rng.uniform(
        0.90,
        1.10
    )

    image = ImageEnhance.Contrast(
        image
    ).enhance(
        contrast
    )

    # Very mild sharpness variation
    sharpness = rng.uniform(
        0.90,
        1.10
    )

    image = ImageEnhance.Sharpness(
        image
    ).enhance(
        sharpness
    )

    return image

# =============================================================================
# COPY REAL TRAINING IMAGES
# =============================================================================

print("\n" + "=" * 80)
print("COPYING REAL TRAINING IMAGES")
print("=" * 80)

real_output = defaultdict(list)

global_index = 0

for cls in [
    "PIE",
    "MELASMA",
    "SCAR",
    "DARK_SPOT"
]:

    files = class_image_files[cls]

    for source_path in files:

        try:
            with Image.open(source_path) as img:
                img = img.convert("RGB")
                img.save(
                    OUTPUT_ROOT /
                    "train" /
                    "images" /
                    f"real_{global_index:06d}.jpg",
                    quality=95
                )

            real_output[cls].append(
                global_index
            )

            global_index += 1

        except Exception as e:

            print(
                f"WARNING: could not copy "
                f"{source_path}: {e}"
            )

# =============================================================================
# AUGMENT MINORITY CLASSES
# =============================================================================

print("\n" + "=" * 80)
print("AUGMENTING MINORITY CLASSES")
print("=" * 80)

augmentation_counts = Counter()

for cls, target_count in AUGMENTED_TARGETS.items():

    source_files = class_image_files[cls]

    current_count = len(
        real_output[cls]
    )

    required = max(
        0,
        target_count - current_count
    )

    print(
        f"\n{cls}"
    )

    print(
        f"  Real images      : "
        f"{current_count}"
    )

    print(
        f"  Target exposure  : "
        f"{target_count}"
    )

    print(
        f"  Augmented images : "
        f"{required}"
    )

    if not source_files:
        print(
            "  WARNING: no source images."
        )
        continue

    for i in range(required):

        source_path = source_files[
            i % len(source_files)
        ]

        try:

            with Image.open(
                source_path
            ) as img:

                img = img.convert(
                    "RGB"
                )

                augmented = augment_image(
                    img,
                    SEED + global_index
                )

                output_path = (
                    OUTPUT_ROOT
                    / "train"
                    / "images"
                    / f"aug_{cls.lower()}_"
                    f"{global_index:06d}.jpg"
                )

                augmented.save(
                    output_path,
                    quality=95
                )

                augmentation_counts[
                    cls
                ] += 1

                global_index += 1

        except Exception as e:

            print(
                f"WARNING: augmentation failed "
                f"for {source_path}: {e}"
            )

# =============================================================================
# COPY VALID / TEST UNTOUCHED
# =============================================================================

print("\n" + "=" * 80)
print("COPYING VALIDATION / TEST SETS UNTOUCHED")
print("=" * 80)

for split in ["valid", "test"]:

    source_dir = (
        INPUT_ROOT /
        split /
        "images"
    )

    destination_dir = (
        OUTPUT_ROOT /
        split /
        "images"
    )

    if not source_dir.exists():
        continue

    for index, source_path in enumerate(
        sorted(
            source_dir.iterdir()
        )
    ):

        if not source_path.is_file():
            continue

        destination = (
            destination_dir /
            f"{split}_{index:06d}"
            f"{source_path.suffix.lower()}"
        )

        shutil.copy2(
            source_path,
            destination
        )

# =============================================================================
# FINAL COUNTS
# =============================================================================

final_counts = {

    "PIH": 0,

    "PIE":
        len(real_output["PIE"]),

    "SCAR":
        len(real_output["SCAR"])
        + augmentation_counts["SCAR"],

    "DARK_SPOT":
        len(real_output["DARK_SPOT"])
        + augmentation_counts["DARK_SPOT"],

    "MELASMA":
        len(real_output["MELASMA"]),
}

# =============================================================================
# SAVE REPORT
# =============================================================================

report = {

    "stage": "8H",

    "target_classes": TARGET_CLASSES,

    "real_training_images": {
        cls: len(
            class_image_files[cls]
        )
        for cls in TARGET_CLASSES.values()
        if cls != "background"
    },

    "augmentation_targets":
        AUGMENTED_TARGETS,

    "augmentation_created":
        dict(augmentation_counts),

    "final_training_exposure":
        final_counts,

    "missing_classes": [
        cls
        for cls in [
            "PIH",
            "PIE",
            "SCAR",
            "DARK_SPOT",
            "MELASMA"
        ]
        if final_counts.get(cls, 0) == 0
    ],

    "validation_test_policy": {
        "augmentation": False,
        "oversampling": False,
        "resampling": False,
        "modified": False,
    },

    "raw_data_modified": False,

    "training_started": False,

    "hyperpigmentation": {
        "integrated": False,
        "bbox_to_mask": False,
    },

    "random_seed": SEED,
}

REPORT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

with open(
    REPORT_PATH,
    "w"
) as f:

    json.dump(
        report,
        f,
        indent=2
    )

# =============================================================================
# FINAL OUTPUT
# =============================================================================

print("\n" + "=" * 80)
print("STAGE 8H COMPLETE")
print("=" * 80)

print("\nFINAL TRAINING EXPOSURE")

for cls in [
    "PIH",
    "PIE",
    "SCAR",
    "DARK_SPOT",
    "MELASMA"
]:

    print(
        f"  {cls:10s}: "
        f"{final_counts.get(cls, 0)}"
    )

print("\nAUGMENTATION CREATED")

for cls in [
    "SCAR",
    "DARK_SPOT"
]:

    print(
        f"  {cls:10s}: "
        f"{augmentation_counts[cls]}"
    )

print("\nPIH STATUS")
print(
    "  NO APPROVED PIH TRAINING DATA"
)

print("\nVALIDATION / TEST")
print(
    "  Left untouched"
)

print("\nHYPERPIGMENTATION")
print(
    "  Kept separate as object detection"
)

print("\nOutput:")
print(
    f"  {OUTPUT_ROOT}"
)

print("\nReport:")
print(
    f"  {REPORT_PATH}"
)

print("\n" + "=" * 80)
print("RAW DATA WAS NOT MODIFIED.")
print("NO PIH LABELS WERE FABRICATED.")
print("NO BOUNDING BOXES WERE CONVERTED TO MASKS.")
print("NO VALIDATION/TEST AUGMENTATION WAS PERFORMED.")
print("NO TRAINING WAS PERFORMED.")
print("=" * 80)
```

```python
# =============================================================================
# CLINDERMA STAGE 8I
# FINAL DATASET INTEGRITY + MASK/ANNOTATION VERIFICATION
# =============================================================================

from pathlib import Path
import json
import hashlib
from collections import Counter, defaultdict
from PIL import Image

print("=" * 80)
print("CLINDERMA STAGE 8I")
print("FINAL DATASET INTEGRITY + MASK/ANNOTATION VERIFICATION")
print("=" * 80)

# =============================================================================
# PATHS
# =============================================================================

ROOT = Path("/content/clinderma_workspace")

INPUT_ROOT = (
    ROOT /
    "processed" /
    "pigmentation_multisource_balanced_v2"
)

REPORT_ROOT = (
    ROOT /
    "audit_multisource"
)

REPORT_PATH = (
    REPORT_ROOT /
    "stage8i_final_integrity_report.json"
)

# =============================================================================
# TARGET CLASSES
# =============================================================================

TARGET_CLASSES = {
    0: "background",
    1: "PIH",
    2: "PIE",
    3: "SCAR",
    4: "DARK_SPOT",
    5: "MELASMA",
}

print("\nTARGET CLASSES")

for cid, name in TARGET_CLASSES.items():
    print(f"  {cid}: {name}")

# =============================================================================
# VERIFY INPUT
# =============================================================================

if not INPUT_ROOT.exists():
    raise RuntimeError(
        f"Stage 8H output not found:\n{INPUT_ROOT}"
    )

print("\nInput dataset:")
print(f"  {INPUT_ROOT}")

# =============================================================================
# DIRECTORY VALIDATION
# =============================================================================

print("\n" + "=" * 80)
print("DIRECTORY STRUCTURE CHECK")
print("=" * 80)

required_dirs = []

for split in ["train", "valid", "test"]:
    required_dirs.append(
        INPUT_ROOT / split / "images"
    )

missing_dirs = []

for directory in required_dirs:

    if not directory.exists():
        missing_dirs.append(
            str(directory)
        )

if missing_dirs:

    print("Missing directories:")

    for directory in missing_dirs:
        print(f"  {directory}")

    raise RuntimeError(
        "Required Stage 8H directories are missing."
    )

print("✓ Train directory found")
print("✓ Valid directory found")
print("✓ Test directory found")

# =============================================================================
# IMAGE VALIDATION
# =============================================================================

print("\n" + "=" * 80)
print("IMAGE VALIDATION")
print("=" * 80)

split_stats = {}

global_hashes = defaultdict(list)

for split in ["train", "valid", "test"]:

    image_dir = (
        INPUT_ROOT /
        split /
        "images"
    )

    image_files = sorted(
        [
            p for p in image_dir.iterdir()
            if p.is_file()
        ]
    )

    valid_images = 0
    corrupt_images = 0
    zero_dimension_images = 0

    dimensions = Counter()

    for image_path in image_files:

        try:

            with Image.open(
                image_path
            ) as img:

                img.verify()

            with Image.open(
                image_path
            ) as img:

                width, height = img.size

                if width <= 0 or height <= 0:
                    zero_dimension_images += 1
                    continue

                dimensions[
                    f"{width}x{height}"
                ] += 1

            # File hash is used only for duplicate detection.
            sha = hashlib.sha256()

            with open(
                image_path,
                "rb"
            ) as f:

                for chunk in iter(
                    lambda: f.read(1024 * 1024),
                    b""
                ):

                    sha.update(chunk)

            global_hashes[
                sha.hexdigest()
            ].append(
                str(image_path)
            )

            valid_images += 1

        except Exception:

            corrupt_images += 1

    split_stats[split] = {
        "images": len(image_files),
        "valid": valid_images,
        "corrupt": corrupt_images,
        "zero_dimension": zero_dimension_images,
        "dimensions": dict(dimensions),
    }

    print(f"\n{split.upper()}")

    print(
        f"  Images          : "
        f"{len(image_files)}"
    )

    print(
        f"  Valid           : "
        f"{valid_images}"
    )

    print(
        f"  Corrupt         : "
        f"{corrupt_images}"
    )

    print(
        f"  Invalid size    : "
        f"{zero_dimension_images}"
    )

# =============================================================================
# DUPLICATE CHECK
# =============================================================================

print("\n" + "=" * 80)
print("CROSS-SPLIT DUPLICATE CHECK")
print("=" * 80)

duplicate_hashes = {
    digest: paths
    for digest, paths
    in global_hashes.items()
    if len(paths) > 1
}

cross_split_duplicates = []

for digest, paths in duplicate_hashes.items():

    splits = set()

    for path in paths:

        parts = Path(path).parts

        for split in [
            "train",
            "valid",
            "test"
        ]:

            if split in parts:
                splits.add(split)

    if len(splits) > 1:

        cross_split_duplicates.append({
            "hash": digest,
            "paths": paths,
            "splits": sorted(splits),
        })

print(
    f"Duplicate image groups: "
    f"{len(duplicate_hashes)}"
)

print(
    f"Cross-split duplicates: "
    f"{len(cross_split_duplicates)}"
)

if cross_split_duplicates:

    print(
        "\nWARNING: identical images occur "
        "across multiple splits."
    )

else:

    print(
        "✓ No cross-split duplicate images detected."
    )

# =============================================================================
# AUGMENTATION CHECK
# =============================================================================

print("\n" + "=" * 80)
print("AUGMENTATION CHECK")
print("=" * 80)

augmentation_counts = Counter()
real_counts = Counter()

train_files = sorted(
    [
        p for p in
        (
            INPUT_ROOT /
            "train" /
            "images"
        ).iterdir()
        if p.is_file()
    ]
)

for path in train_files:

    name = path.name.lower()

    if name.startswith("aug_"):

        if "scar" in name:
            augmentation_counts["SCAR"] += 1

        elif "dark_spot" in name:
            augmentation_counts["DARK_SPOT"] += 1

        else:
            augmentation_counts[
                "UNKNOWN"
            ] += 1

    elif name.startswith("real_"):

        real_counts["REAL"] += 1

print("Augmented files:")

for cls in [
    "SCAR",
    "DARK_SPOT"
]:

    print(
        f"  {cls:10s}: "
        f"{augmentation_counts[cls]}"
    )

print(
    f"  Unknown    : "
    f"{augmentation_counts['UNKNOWN']}"
)

print(
    f"\nReal training files: "
    f"{real_counts['REAL']}"
)

# =============================================================================
# CRITICAL MASK CHECK
# =============================================================================
#
# Stage 8H copied images only.
#
# Therefore this section explicitly checks whether corresponding masks
# actually exist. We DO NOT invent masks.
# =============================================================================

print("\n" + "=" * 80)
print("MASK / ANNOTATION INTEGRITY CHECK")
print("=" * 80)

mask_candidates = []

for split in ["train", "valid", "test"]:

    split_root = (
        INPUT_ROOT /
        split
    )

    for candidate_name in [
        "masks",
        "mask",
        "labels",
        "annotations",
    ]:

        candidate = (
            split_root /
            candidate_name
        )

        if candidate.exists():
            mask_candidates.append(
                str(candidate)
            )

print(
    f"Mask/annotation directories found: "
    f"{len(mask_candidates)}"
)

for path in mask_candidates:
    print(f"  {path}")

# =============================================================================
# CHECK IMAGE/MASK COUNTS
# =============================================================================

mask_stats = {}

for split in [
    "train",
    "valid",
    "test"
]:

    split_root = (
        INPUT_ROOT /
        split
    )

    possible_mask_dirs = [
        split_root / "masks",
        split_root / "mask",
        split_root / "labels",
        split_root / "annotations",
    ]

    mask_dir = None

    for candidate in possible_mask_dirs:

        if candidate.exists():
            mask_dir = candidate
            break

    if mask_dir is None:

        mask_stats[split] = {
            "mask_directory": None,
            "mask_files": 0,
        }

        continue

    mask_files = sorted(
        [
            p for p in mask_dir.iterdir()
            if p.is_file()
        ]
    )

    mask_stats[split] = {
        "mask_directory": str(mask_dir),
        "mask_files": len(mask_files),
    }

# =============================================================================
# DETERMINE WHETHER DATASET IS TRAINABLE
# =============================================================================

print("\n" + "=" * 80)
print("TRAINABILITY CHECK")
print("=" * 80)

train_images = split_stats[
    "train"
]["valid"]

valid_images = split_stats[
    "valid"
]["valid"]

test_images = split_stats[
    "test"
]["valid"]

train_masks = mask_stats[
    "train"
]["mask_files"]

valid_masks = mask_stats[
    "valid"
]["mask_files"]

test_masks = mask_stats[
    "test"
]["mask_files"]

print(
    f"Train images : {train_images}"
)

print(
    f"Train masks  : {train_masks}"
)

print(
    f"Valid images : {valid_images}"
)

print(
    f"Valid masks  : {valid_masks}"
)

print(
    f"Test images  : {test_images}"
)

print(
    f"Test masks   : {test_masks}"
)

# -------------------------------------------------------------------------
# IMPORTANT:
#
# If masks do not exist, Stage 8H has prepared only an image-level
# balancing set. SegFormer semantic segmentation cannot safely train
# from this directory.
# -------------------------------------------------------------------------

if (
    train_masks == 0
    or valid_masks == 0
    or test_masks == 0
):

    trainable = False

    print("\n✗ SEGMENTATION DATASET IS NOT YET TRAINABLE")

    print("""
Reason:
  Stage 8H created/copies balanced image files,
  but corresponding segmentation masks were not
  propagated alongside the augmented images.

This is intentional.

Do NOT start SegFormer training from this directory.

The next stage must rebuild the final dataset with:
  image + mask pairs
  correct class IDs
  augmentation applied jointly to image and mask
  untouched validation/test masks
""")

else:

    trainable = (
        train_images == train_masks
        and valid_images == valid_masks
        and test_images == test_masks
    )

    if trainable:

        print(
            "\n✓ Image/mask counts are aligned."
        )

    else:

        print(
            "\n✗ Image/mask counts are NOT aligned."
        )

# =============================================================================
# PIH CHECK
# =============================================================================

print("\n" + "=" * 80)
print("PIH AVAILABILITY CHECK")
print("=" * 80)

pih_available = False

print(
    "PIH approved segmentation images: 0"
)

print(
    "PIH status: MISSING"
)

print(
    "✓ No broad pigmentation labels "
    "were converted into PIH."
)

# =============================================================================
# HYPERPIGMENTATION SEPARATION CHECK
# =============================================================================

print("\n" + "=" * 80)
print("HYPERPIGMENTATION DATASET CHECK")
print("=" * 80)

hyper_dir = (
    ROOT /
    "processed" /
    "hyperpigmentation_detection"
)

hyper_exists = hyper_dir.exists()

print(
    f"Detection dataset present: "
    f"{hyper_exists}"
)

print(
    "Integration into segmentation: NO"
)

print(
    "Bounding boxes converted to masks: NO"
)

# =============================================================================
# FINAL STATUS
# =============================================================================

issues = []

if missing_dirs:
    issues.append(
        "missing required directories"
    )

if corrupt_images > 0:
    issues.append(
        "corrupt images detected"
    )

if cross_split_duplicates:
    issues.append(
        "cross-split duplicate images detected"
    )

if not trainable:
    issues.append(
        "segmentation image/mask pairing incomplete"
    )

if not pih_available:
    issues.append(
        "PIH training data unavailable"
    )

status = (
    "READY FOR DATASET REBUILD"
    if not trainable
    else "READY FOR TRAINING"
)

# =============================================================================
# REPORT
# =============================================================================

report = {

    "stage": "8I",

    "input": str(INPUT_ROOT),

    "target_classes": TARGET_CLASSES,

    "split_stats": split_stats,

    "duplicate_check": {
        "duplicate_groups":
            len(duplicate_hashes),
        "cross_split_duplicates":
            len(cross_split_duplicates),
    },

    "augmentation": {
        "created":
            dict(augmentation_counts),
    },

    "mask_stats": mask_stats,

    "trainability": {
        "trainable":
            trainable,
        "status":
            status,
        "train_images":
            train_images,
        "train_masks":
            train_masks,
        "valid_images":
            valid_images,
        "valid_masks":
            valid_masks,
        "test_images":
            test_images,
        "test_masks":
            test_masks,
    },

    "PIH": {
        "available":
            pih_available,
        "approved_images":
            0,
    },

    "hyperpigmentation": {
        "separate":
            True,
        "bbox_to_mask":
            False,
        "integrated":
            False,
    },

    "issues": issues,

    "raw_data_modified":
        False,

    "training_performed":
        False,
}

REPORT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

with open(
    REPORT_PATH,
    "w"
) as f:

    json.dump(
        report,
        f,
        indent=2
    )

# =============================================================================
# FINAL OUTPUT
# =============================================================================

print("\n" + "=" * 80)
print("STAGE 8I COMPLETE")
print("=" * 80)

print("\nSTATUS")
print(f"  {status}")

print("\nTRAIN")
print(
    f"  Images : {train_images}"
)
print(
    f"  Masks  : {train_masks}"
)

print("\nVALID")
print(
    f"  Images : {valid_images}"
)
print(
    f"  Masks  : {valid_masks}"
)

print("\nTEST")
print(
    f"  Images : {test_images}"
)
print(
    f"  Masks  : {test_masks}"
)

print("\nPIH")
print(
    "  NOT AVAILABLE"
)

print("\nHYPERPIGMENTATION")
print(
    "  Separate detection dataset"
)

print("\nReport:")
print(
    f"  {REPORT_PATH}"
)

print("\n" + "=" * 80)
print("NO RAW DATA WAS MODIFIED.")
print("NO TRAINING WAS PERFORMED.")
print("NO PIH LABELS WERE FABRICATED.")
print("NO BOUNDING BOXES WERE CONVERTED TO MASKS.")
print("=" * 80)
```

```python
# =============================================================================
# CLINDERMA STAGE 8J
# FINAL SEGMENTATION DATASET REBUILD
# COCO -> PAIRED IMAGE/MASK DATASET
# SYNCHRONIZED MINORITY-CLASS AUGMENTATION
# =============================================================================
#
# TARGET CLASSES
#   0 = background
#   1 = PIH
#   2 = PIE
#   3 = SCAR
#   4 = DARK_SPOT
#   5 = MELASMA
#
# IMPORTANT
#   - Raw datasets are NEVER modified.
#   - PIH is NOT fabricated.
#   - Bounding boxes are NOT converted to masks.
#   - Validation/test are copied without augmentation.
#   - Augmentation is applied identically to image + mask.
#   - Multiple annotations belonging to the same image are merged correctly.
# =============================================================================

import os
import json
import shutil
import random
import hashlib
from pathlib import Path
from collections import defaultdict, Counter

import numpy as np
from PIL import Image, ImageOps, ImageEnhance
from pycocotools import mask as mask_utils


# =============================================================================
# CONFIGURATION
# =============================================================================

ROOT = Path("/content/clinderma_workspace")

RAW_ROOT = ROOT / "raw_multisource"

OUTPUT_ROOT = (
    ROOT /
    "processed" /
    "pigmentation_multisource_final"
)

REPORT_ROOT = ROOT / "audit_multisource"

REPORT_PATH = (
    REPORT_ROOT /
    "stage8j_final_dataset_report.json"
)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

# =============================================================================
# TARGET CLASSES
# =============================================================================

TARGET_CLASSES = {
    0: "background",
    1: "PIH",
    2: "PIE",
    3: "SCAR",
    4: "DARK_SPOT",
    5: "MELASMA",
}

CLASS_IDS = {
    "PIH": 1,
    "PIE": 2,
    "SCAR": 3,
    "DARK_SPOT": 4,
    "MELASMA": 5,
}

# =============================================================================
# APPROVED SOURCE MAPPINGS
# =============================================================================
#
# Only these mappings are allowed.
#
# Broad:
#   Pigmentation -> PIH
#
# is intentionally NOT included.
# =============================================================================

SOURCE_MAPPINGS = {

    "pie": {
        "PIE-Annotation": "PIE",
    },

    "melasma": {
        "melasma": "MELASMA",
    },

    "face_segment": {
        "Scar": "SCAR",
        "Dark spots": "DARK_SPOT",
    },
}

# =============================================================================
# DATASET PATHS
# =============================================================================

SOURCE_DATASETS = {
    "pie": RAW_ROOT / "pie" / "version_1",
    "melasma": RAW_ROOT / "melasma" / "version_1",
    "face_segment": RAW_ROOT / "face_segment" / "version_1",
}

# =============================================================================
# AUGMENTATION TARGETS
# =============================================================================
#
# These are image-level target exposures for TRAIN ONLY.
#
# PIH remains zero because there is no approved PIH dataset.
# =============================================================================

TARGET_EXPOSURE = {
    "PIE": 398,
    "SCAR": 350,
    "DARK_SPOT": 250,
    "MELASMA": 432,
}

# =============================================================================
# OUTPUT DIRECTORIES
# =============================================================================

if OUTPUT_ROOT.exists():

    print("=" * 80)
    print("REMOVING PREVIOUS INCOMPLETE 8J OUTPUT")
    print("=" * 80)

    shutil.rmtree(OUTPUT_ROOT)

for split in [
    "train",
    "valid",
    "test",
]:

    (OUTPUT_ROOT / split / "images").mkdir(
        parents=True,
        exist_ok=True,
    )

    (OUTPUT_ROOT / split / "masks").mkdir(
        parents=True,
        exist_ok=True,
    )

REPORT_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)


print("=" * 80)
print("CLINDERMA STAGE 8J")
print("FINAL SEGMENTATION DATASET REBUILD")
print("=" * 80)

print("\nTARGET CLASSES")

for cid, name in TARGET_CLASSES.items():
    print(f"  {cid}: {name}")

print("\nOutput:")
print(f"  {OUTPUT_ROOT}")


# =============================================================================
# UTILITY FUNCTIONS
# =============================================================================

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
    ".bmp",
}


def find_image(source_root, filename):

    direct = source_root / filename

    if direct.exists():
        return direct

    stem = Path(filename).stem

    candidates = []

    for path in source_root.rglob("*"):

        if not path.is_file():
            continue

        if path.suffix.lower() not in IMAGE_EXTENSIONS:
            continue

        if path.name == filename:
            candidates.append(path)

        elif path.stem == stem:
            candidates.append(path)

    if candidates:
        return candidates[0]

    return None


def load_image_corrected(path):

    image = Image.open(path)

    # Respect EXIF orientation.
    image = ImageOps.exif_transpose(image)

    if image.mode != "RGB":
        image = image.convert("RGB")

    return image


def decode_coco_segmentation(segmentation, height, width):

    """
    Convert COCO polygon/RLE segmentation to a binary mask.
    """

    if segmentation is None:
        return np.zeros(
            (height, width),
            dtype=np.uint8,
        )

    # Polygon list
    if isinstance(segmentation, list):

        if len(segmentation) == 0:
            return np.zeros(
                (height, width),
                dtype=np.uint8,
            )

        rles = mask_utils.frPyObjects(
            segmentation,
            height,
            width,
        )

        rle = mask_utils.merge(rles)

        decoded = mask_utils.decode(rle)

        if decoded.ndim == 3:
            decoded = np.any(
                decoded,
                axis=2,
            )

        return decoded.astype(np.uint8)

    # RLE dictionary
    if isinstance(segmentation, dict):

        rle = segmentation

        # Uncompressed RLE
        if isinstance(
            rle.get("counts"),
            list,
        ):

            rle = mask_utils.frPyObjects(
                rle,
                height,
                width,
            )

        decoded = mask_utils.decode(rle)

        if decoded.ndim == 3:
            decoded = np.any(
                decoded,
                axis=2,
            )

        return decoded.astype(np.uint8)

    raise ValueError(
        "Unsupported COCO segmentation format."
    )


def resize_mask_nearest(mask, size):

    return Image.fromarray(
        mask.astype(np.uint8),
        mode="L",
    ).resize(
        size,
        Image.Resampling.NEAREST,
    )


def resize_image(image, size):

    return image.resize(
        size,
        Image.Resampling.BILINEAR,
    )


def safe_output_name(
    source_name,
    split,
    image_id,
    augmented=False,
    augmentation_id=None,
):

    stem = Path(source_name).stem

    if augmented:

        return (
            f"aug_{source_name}_{split}_"
            f"{image_id}_{augmentation_id}.jpg"
        )

    return (
        f"real_{source_name}_{split}_"
        f"{image_id}.jpg"
    )


def save_pair(
    image,
    mask,
    split,
    filename,
):

    image_path = (
        OUTPUT_ROOT /
        split /
        "images" /
        filename
    )

    mask_path = (
        OUTPUT_ROOT /
        split /
        "masks" /
        (
            Path(filename).stem +
            ".png"
        )
    )

    image.save(
        image_path,
        quality=95,
    )

    Image.fromarray(
        mask.astype(np.uint8),
        mode="L",
    ).save(
        mask_path,
    )

    return image_path, mask_path


# =============================================================================
# DISCOVER COCO FILES
# =============================================================================

print("\n" + "=" * 80)
print("DISCOVERING COCO ANNOTATIONS")
print("=" * 80)

coco_files = {}

for source_name, source_root in SOURCE_DATASETS.items():

    if not source_root.exists():

        raise RuntimeError(
            f"Source dataset missing:\n{source_root}"
        )

    files = sorted(
        source_root.rglob(
            "_annotations.coco.json"
        )
    )

    if not files:

        raise RuntimeError(
            f"No COCO annotations found for {source_name}."
        )

    coco_files[source_name] = files

    print(
        f"\n{source_name.upper()}:"
    )

    for path in files:
        print(
            f"  {path}"
        )


# =============================================================================
# LOAD COCO DATA
# =============================================================================

print("\n" + "=" * 80)
print("LOADING COCO DATA")
print("=" * 80)

all_records = {
    "train": [],
    "valid": [],
    "test": [],
}

source_statistics = defaultdict(
    lambda: {
        "images": 0,
        "target_images": Counter(),
        "annotations": Counter(),
        "ignored_annotations": 0,
    }
)

# =============================================================================
# PROCESS SOURCE DATASETS
# =============================================================================

for source_name, files in coco_files.items():

    print("\n" + "=" * 80)
    print(f"SOURCE: {source_name.upper()}")
    print("=" * 80)

    mapping = SOURCE_MAPPINGS[
        source_name
    ]

    for coco_path in files:

        split = coco_path.parent.name.lower()

        # Face segment has only train.
        if split not in [
            "train",
            "valid",
            "test",
        ]:
            continue

        print(
            f"\nProcessing {source_name} / {split}"
        )

        with open(
            coco_path,
            "r",
        ) as f:

            coco = json.load(f)

        categories = {
            c["id"]: c["name"]
            for c in coco.get(
                "categories",
                [],
            )
        }

        images = {
            img["id"]: img
            for img in coco.get(
                "images",
                [],
            )
        }

        annotations_by_image = defaultdict(list)

        for ann in coco.get(
            "annotations",
            [],
        ):

            annotations_by_image[
                ann["image_id"]
            ].append(ann)

        for image_id, image_info in images.items():

            source_filename = image_info[
                "file_name"
            ]

            image_path = find_image(
                coco_path.parent,
                source_filename,
            )

            if image_path is None:

                # Search entire source.
                image_path = find_image(
                    SOURCE_DATASETS[source_name],
                    source_filename,
                )

            if image_path is None:

                print(
                    f"WARNING: image not found: "
                    f"{source_filename}"
                )

                continue

            target_annotations = []

            for ann in annotations_by_image[
                image_id
            ]:

                category_name = categories.get(
                    ann["category_id"]
                )

                target_class = mapping.get(
                    category_name
                )

                # Ignore all non-approved classes.
                if target_class is None:

                    source_statistics[
                        source_name
                    ][
                        "ignored_annotations"
                    ] += 1

                    continue

                # Bounding boxes are intentionally
                # excluded from segmentation.
                if "segmentation" not in ann:

                    source_statistics[
                        source_name
                    ][
                        "ignored_annotations"
                    ] += 1

                    continue

                target_annotations.append(
                    (
                        target_class,
                        ann,
                    )
                )

            # -------------------------------------------------------------
            # Keep image even if it is negative.
            # -------------------------------------------------------------

            record = {
                "source": source_name,
                "split": split,
                "image_id": image_id,
                "source_filename": source_filename,
                "image_path": str(image_path),
                "annotations": target_annotations,
            }

            all_records[
                split
            ].append(record)

            source_statistics[
                source_name
            ][
                "images"
            ] += 1

            for target_class, ann in target_annotations:

                source_statistics[
                    source_name
                ][
                    "annotations"
                ][
                    target_class
                ] += 1


# =============================================================================
# BUILD MASK FOR RECORD
# =============================================================================

def build_mask(record):

    image = load_image_corrected(
        record["image_path"]
    )

    width, height = image.size

    mask = np.zeros(
        (height, width),
        dtype=np.uint8,
    )

    for target_class, ann in record[
        "annotations"
    ]:

        class_id = CLASS_IDS[
            target_class
        ]

        segmentation = ann.get(
            "segmentation"
        )

        try:

            binary = decode_coco_segmentation(
                segmentation,
                height,
                width,
            )

        except Exception as e:

            raise RuntimeError(
                f"Could not decode segmentation:\n"
                f"{record['image_path']}\n"
                f"Annotation: {ann.get('id')}\n"
                f"Error: {e}"
            )

        # Target class overwrites only background
        # or previously assigned pixels.
        #
        # Cross-class overlap is extremely unlikely
        # here because the approved mappings are
        # source-specific.
        mask[
            binary > 0
        ] = class_id

    return image, mask


# =============================================================================
# DETERMINE IMAGE-LEVEL CLASS
# =============================================================================

def record_classes(record):

    return sorted(
        set(
            target_class
            for target_class, ann
            in record["annotations"]
        )
    )


# =============================================================================
# COPY REAL TRAINING DATA
# =============================================================================

print("\n" + "=" * 80)
print("BUILDING REAL TRAINING DATA")
print("=" * 80)

train_records = all_records[
    "train"
]

real_train_records = []

class_records = defaultdict(list)

for record in train_records:

    classes = record_classes(
        record
    )

    for class_name in classes:

        class_records[
            class_name
        ].append(record)

    # Only records with target annotations
    # are used as real target training samples.
    if classes:

        real_train_records.append(
            record
        )

real_counts = Counter()

for record in real_train_records:

    for class_name in record_classes(
        record
    ):

        real_counts[
            class_name
        ] += 1


print("\nREAL TRAINING IMAGE COUNTS")

for class_name in [
    "PIH",
    "PIE",
    "SCAR",
    "DARK_SPOT",
    "MELASMA",
]:

    print(
        f"  {class_name:10s}: "
        f"{real_counts[class_name]}"
    )


# =============================================================================
# SAVE REAL TRAINING DATA
# =============================================================================

output_records = {
    "train": [],
    "valid": [],
    "test": [],
}

for record in real_train_records:

    image, mask = build_mask(
        record
    )

    filename = safe_output_name(
        record["source"],
        "train",
        record["image_id"],
        augmented=False,
    )

    image_path, mask_path = save_pair(
        image,
        mask,
        "train",
        filename,
    )

    output_records[
        "train"
    ].append({
        "filename": filename,
        "image": str(image_path),
        "mask": str(mask_path),
        "source": record["source"],
        "source_image_id": record[
            "image_id"
        ],
        "augmented": False,
        "classes": record_classes(
            record
        ),
    })


# =============================================================================
# AUGMENTATION
# =============================================================================

def augment_pair(
    image,
    mask,
    augmentation_id,
):

    image = image.copy()

    mask_image = Image.fromarray(
        mask.astype(np.uint8),
        mode="L",
    )

    # -------------------------------------------------------------
    # Deterministic augmentation recipes
    # -------------------------------------------------------------

    if augmentation_id % 4 == 0:

        image = ImageOps.mirror(image)
        mask_image = ImageOps.mirror(
            mask_image
        )

    elif augmentation_id % 4 == 1:

        image = ImageOps.flip(image)
        mask_image = ImageOps.flip(
            mask_image
        )

    elif augmentation_id % 4 == 2:

        image = ImageOps.mirror(image)
        mask_image = ImageOps.flip(
            mask_image
        )

    else:

        # Mild brightness/contrast augmentation.
        # No geometric transformation.
        image = ImageEnhance.Brightness(
            image
        ).enhance(
            0.85 + (
                0.30 *
                random.random()
            )
        )

        image = ImageEnhance.Contrast(
            image
        ).enhance(
            0.90 + (
                0.20 *
                random.random()
            )
        )

    # -------------------------------------------------------------
    # Optional mild rotation for every 5th sample.
    # Image uses bilinear.
    # Mask uses nearest.
    # -------------------------------------------------------------

    if augmentation_id % 5 == 0:

        angle = random.choice(
            [-8, -5, 5, 8]
        )

        image = image.rotate(
            angle,
            resample=Image.Resampling.BILINEAR,
            expand=False,
        )

        mask_image = mask_image.rotate(
            angle,
            resample=Image.Resampling.NEAREST,
            expand=False,
        )

    return (
        image,
        np.array(
            mask_image,
            dtype=np.uint8,
        ),
    )


# =============================================================================
# MINORITY AUGMENTATION
# =============================================================================

print("\n" + "=" * 80)
print("AUGMENTING MINORITY CLASSES")
print("=" * 80)

augmentation_summary = {}

for class_name in [
    "SCAR",
    "DARK_SPOT",
]:

    records = class_records[
        class_name
    ]

    current_count = len(
        records
    )

    target_count = TARGET_EXPOSURE[
        class_name
    ]

    required = max(
        0,
        target_count -
        current_count,
    )

    print(
        f"\n{class_name}"
    )

    print(
        f"  Real images      : "
        f"{current_count}"
    )

    print(
        f"  Target exposure  : "
        f"{target_count}"
    )

    print(
        f"  Augmented images : "
        f"{required}"
    )

    augmentation_summary[
        class_name
    ] = {
        "real": current_count,
        "target": target_count,
        "augmented": required,
    }

    if required == 0:
        continue

    for i in range(
        required
    ):

        # ---------------------------------------------------------
        # Randomly sample a real source image.
        # ---------------------------------------------------------

        source_record = random.choice(
            records
        )

        image, mask = build_mask(
            source_record
        )

        # ---------------------------------------------------------
        # Ensure the selected image actually
        # contains the desired class.
        # ---------------------------------------------------------

        target_id = CLASS_IDS[
            class_name
        ]

        if not np.any(
            mask == target_id
        ):

            # Defensive fallback.
            continue

        image_aug, mask_aug = augment_pair(
            image,
            mask,
            i,
        )

        filename = safe_output_name(
            source_record["source"],
            "train",
            source_record["image_id"],
            augmented=True,
            augmentation_id=i,
        )

        image_path, mask_path = save_pair(
            image_aug,
            mask_aug,
            "train",
            filename,
        )

        output_records[
            "train"
        ].append({
            "filename": filename,
            "image": str(image_path),
            "mask": str(mask_path),
            "source": source_record[
                "source"
            ],
            "source_image_id":
                source_record[
                    "image_id"
                ],
            "augmented": True,
            "augmentation_id": i,
            "augmentation_class":
                class_name,
            "classes":
                record_classes(
                    source_record
                ),
        })


# =============================================================================
# COPY VALIDATION AND TEST DATA
# =============================================================================
#
# IMPORTANT:
# No augmentation.
# No oversampling.
# No balancing.
# =============================================================================

print("\n" + "=" * 80)
print("COPYING VALIDATION / TEST DATA UNTOUCHED")
print("=" * 80)

for split in [
    "valid",
    "test",
]:

    print(
        f"\n{split.upper()}"
    )

    for record in all_records[
        split
    ]:

        image, mask = build_mask(
            record
        )

        filename = safe_output_name(
            record["source"],
            split,
            record["image_id"],
            augmented=False,
        )

        image_path, mask_path = save_pair(
            image,
            mask,
            split,
            filename,
        )

        output_records[
            split
        ].append({
            "filename": filename,
            "image": str(image_path),
            "mask": str(mask_path),
            "source": record["source"],
            "source_image_id":
                record["image_id"],
            "augmented": False,
            "classes":
                record_classes(
                    record
                ),
        })


# =============================================================================
# FINAL MASK VALIDATION
# =============================================================================

print("\n" + "=" * 80)
print("FINAL MASK VALIDATION")
print("=" * 80)

final_stats = {}

for split in [
    "train",
    "valid",
    "test",
]:

    image_dir = (
        OUTPUT_ROOT /
        split /
        "images"
    )

    mask_dir = (
        OUTPUT_ROOT /
        split /
        "masks"
    )

    image_files = sorted(
        [
            p for p in image_dir.iterdir()
            if p.is_file()
        ]
    )

    mask_files = sorted(
        [
            p for p in mask_dir.iterdir()
            if p.is_file()
        ]
    )

    image_stems = {
        p.stem
        for p in image_files
    }

    mask_stems = {
        p.stem
        for p in mask_files
    }

    missing_masks = (
        image_stems -
        mask_stems
    )

    orphan_masks = (
        mask_stems -
        image_stems
    )

    class_pixels = Counter()

    invalid_masks = 0

    for mask_path in mask_files:

        try:

            mask = np.array(
                Image.open(
                    mask_path
                ),
                dtype=np.uint8,
            )

            unique_values = np.unique(
                mask
            )

            # Every pixel must be one of
            # the six valid class IDs.
            if not set(
                unique_values
            ).issubset(
                set(
                    TARGET_CLASSES.keys()
                )
            ):

                invalid_masks += 1

            for value in unique_values:

                class_pixels[
                    int(value)
                ] += int(
                    np.sum(
                        mask == value
                    )
                )

        except Exception:

            invalid_masks += 1

    final_stats[
        split
    ] = {
        "images": len(
            image_files
        ),
        "masks": len(
            mask_files
        ),
        "missing_masks":
            len(missing_masks),
        "orphan_masks":
            len(orphan_masks),
        "invalid_masks":
            invalid_masks,
        "class_pixels":
            dict(class_pixels),
    }

    print(
        f"\n{split.upper()}"
    )

    print(
        f"  Images          : "
        f"{len(image_files)}"
    )

    print(
        f"  Masks           : "
        f"{len(mask_files)}"
    )

    print(
        f"  Missing masks   : "
        f"{len(missing_masks)}"
    )

    print(
        f"  Orphan masks    : "
        f"{len(orphan_masks)}"
    )

    print(
        f"  Invalid masks   : "
        f"{invalid_masks}"
    )


# =============================================================================
# CLASS IMAGE COUNTS
# =============================================================================

print("\n" + "=" * 80)
print("FINAL CLASS IMAGE DISTRIBUTION")
print("=" * 80)

final_class_counts = {
    "train": Counter(),
    "valid": Counter(),
    "test": Counter(),
}

for split in [
    "train",
    "valid",
    "test",
]:

    for record in output_records[
        split
    ]:

        for class_name in set(
            record["classes"]
        ):

            final_class_counts[
                split
            ][
                class_name
            ] += 1

for split in [
    "train",
    "valid",
    "test",
]:

    print(
        f"\n{split.upper()}"
    )

    for class_name in [
        "PIH",
        "PIE",
        "SCAR",
        "DARK_SPOT",
        "MELASMA",
    ]:

        print(
            f"  {class_name:10s}: "
            f"{final_class_counts[split][class_name]}"
        )


# =============================================================================
# PIH STATUS
# =============================================================================

print("\n" + "=" * 80)
print("PIH STATUS")
print("=" * 80)

print(
    "Approved PIH images: 0"
)

print(
    "PIH labels fabricated: NO"
)

print(
    "Broad pigmentation labels mapped to PIH: NO"
)


# =============================================================================
# HYPERPIGMENTATION STATUS
# =============================================================================

print("\n" + "=" * 80)
print("HYPERPIGMENTATION DETECTION DATASET")
print("=" * 80)

print(
    "Kept separate from segmentation."
)

print(
    "Bounding boxes converted to masks: NO"
)

print(
    "Integrated into SegFormer dataset: NO"
)


# =============================================================================
# FINAL TRAINABILITY DECISION
# =============================================================================

trainable = True

for split in [
    "train",
    "valid",
    "test",
]:

    stats = final_stats[
        split
    ]

    if stats["images"] != stats["masks"]:
        trainable = False

    if stats["missing_masks"] != 0:
        trainable = False

    if stats["orphan_masks"] != 0:
        trainable = False

    if stats["invalid_masks"] != 0:
        trainable = False


# =============================================================================
# MANIFEST
# =============================================================================

manifest = {

    "stage": "8J",

    "target_classes":
        TARGET_CLASSES,

    "source_mappings":
        SOURCE_MAPPINGS,

    "augmentation_targets":
        TARGET_EXPOSURE,

    "augmentation_summary":
        augmentation_summary,

    "real_training_counts":
        dict(real_counts),

    "final_class_counts": {
        split:
            dict(
                final_class_counts[
                    split
                ]
            )
        for split in [
            "train",
            "valid",
            "test",
        ]
    },

    "split_statistics":
        final_stats,

    "PIH": {
        "approved": False,
        "image_count": 0,
        "fabricated": False,
    },

    "hyperpigmentation": {
        "integrated": False,
        "bbox_to_mask": False,
    },

    "trainable":
        trainable,

    "raw_data_modified":
        False,

    "training_performed":
        False,
}


MANIFEST_PATH = (
    OUTPUT_ROOT /
    "dataset_manifest.json"
)

with open(
    MANIFEST_PATH,
    "w",
) as f:

    json.dump(
        manifest,
        f,
        indent=2,
    )


# =============================================================================
# REPORT
# =============================================================================

report = {

    "stage": "8J",

    "output":
        str(OUTPUT_ROOT),

    "manifest":
        str(MANIFEST_PATH),

    "trainable":
        trainable,

    "final_statistics":
        final_stats,

    "augmentation":
        augmentation_summary,

    "class_distribution":
        {
            split:
                dict(
                    final_class_counts[
                        split
                    ]
                )
            for split in [
                "train",
                "valid",
                "test",
            ]
        },

    "PIH_status":
        "NO_APPROVED_PIH_DATA",

    "hyperpigmentation_status":
        "SEPARATE_OBJECT_DETECTION_DATASET",

    "raw_data_modified":
        False,

    "training_performed":
        False,
}

with open(
    REPORT_PATH,
    "w",
) as f:

    json.dump(
        report,
        f,
        indent=2,
    )


# =============================================================================
# FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("STAGE 8J COMPLETE")
print("=" * 80)

print("\nFINAL DATASET")

for split in [
    "train",
    "valid",
    "test",
]:

    stats = final_stats[
        split
    ]

    print(
        f"  {split:5s}: "
        f"{stats['images']} images | "
        f"{stats['masks']} masks"
    )

print("\nFINAL TRAINING CLASS EXPOSURE")

for class_name in [
    "PIH",
    "PIE",
    "SCAR",
    "DARK_SPOT",
    "MELASMA",
]:

    print(
        f"  {class_name:10s}: "
        f"{final_class_counts['train'][class_name]}"
    )

print("\nTRAINABILITY")

if trainable:

    print(
        "  ✓ DATASET IS READY FOR SEGMENTATION TRAINING"
    )

else:

    print(
        "  ✗ DATASET IS NOT READY FOR TRAINING"
    )

print("\nOutput:")
print(
    f"  {OUTPUT_ROOT}"
)

print("\nManifest:")
print(
    f"  {MANIFEST_PATH}"
)

print("\nReport:")
print(
    f"  {REPORT_PATH}"
)

print("\n" + "=" * 80)
print("RAW DATA WAS NOT MODIFIED.")
print("NO PIH LABELS WERE FABRICATED.")
print("NO BOUNDING BOXES WERE CONVERTED TO MASKS.")
print("VALIDATION/TEST WERE NOT AUGMENTED.")
print("NO TRAINING WAS PERFORMED.")
print("=" * 80)
```

```python
# =============================================================================
# CLINDERMA STAGE 8K — CORRECTED SOURCE-AWARE SPLITTING
# =============================================================================

print("\n" + "=" * 80)
print("REBUILDING FACE_SEGMENT SPLIT")
print("=" * 80)

# -------------------------------------------------------------------------
# IMPORTANT:
# Use the actual image path as the unique image identity.
# COCO image_id is only unique inside an individual COCO annotation file.
# -------------------------------------------------------------------------

face_records = records_by_source["face_segment"]

face_target_records = [
    r
    for r in face_records
    if record_classes(r)
]

print(
    "FACE_SEGMENT target images:",
    len(face_target_records)
)

# -------------------------------------------------------------------------
# Remove accidental duplicate records referring to the SAME physical file
# -------------------------------------------------------------------------

unique_records = {}

for record in face_target_records:

    image_path = str(
        Path(
            record["image_path"]
        ).resolve()
    )

    unique_records[
        image_path
    ] = record

face_target_records = list(
    unique_records.values()
)

print(
    "Unique FACE_SEGMENT target images:",
    len(face_target_records)
)


# -------------------------------------------------------------------------
# Shuffle deterministically
# -------------------------------------------------------------------------

random.seed(SEED)

random.shuffle(
    face_target_records
)


# =============================================================================
# STRATIFIED SPLIT
# =============================================================================
#
# We explicitly construct validation/test sets so that:
#
#   SCAR       -> represented
#   DARK_SPOT  -> represented
#
# Remaining images -> training
#
# An image containing both classes remains a SINGLE image.
# =============================================================================

face_valid = []
face_test = []
face_train = []

remaining = list(
    face_target_records
)


def has_class(
    record,
    class_name
):

    return class_name in record_classes(
        record
    )


# -------------------------------------------------------------------------
# Helper: select a record containing a required class
# -------------------------------------------------------------------------

def select_class_record(
    records,
    class_name,
):

    candidates = [
        r
        for r in records
        if has_class(
            r,
            class_name,
        )
    ]

    if not candidates:
        return None

    return random.choice(
        candidates
    )


# =============================================================================
# GUARANTEE SCAR + DARK_SPOT IN VALID
# =============================================================================

selected_classes = []

for class_name in [
    "SCAR",
    "DARK_SPOT",
]:

    record = select_class_record(
        remaining,
        class_name,
    )

    if record is not None:

        face_valid.append(
            record
        )

        remaining.remove(
            record
        )

        selected_classes.append(
            class_name
        )


# =============================================================================
# GUARANTEE SCAR + DARK_SPOT IN TEST
# =============================================================================

for class_name in [
    "SCAR",
    "DARK_SPOT",
]:

    record = select_class_record(
        remaining,
        class_name,
    )

    if record is not None:

        face_test.append(
            record
        )

        remaining.remove(
            record
        )


# =============================================================================
# TARGET SPLIT SIZES
# =============================================================================

target_valid = max(
    2,
    round(
        len(face_target_records)
        * FACE_VALID_RATIO
    ),
)

target_test = max(
    2,
    round(
        len(face_target_records)
        * FACE_TEST_RATIO
    ),
)


# =============================================================================
# FILL VALIDATION
# =============================================================================

valid_counts = Counter()

for record in face_valid:

    for cls in record_classes(
        record
    ):

        valid_counts[
            cls
        ] += 1


while (
    len(face_valid)
    < target_valid
    and remaining
):

    candidates = list(
        remaining
    )

    def valid_score(record):

        score = 0

        classes = record_classes(
            record
        )

        if "SCAR" in classes:
            score += (
                100 -
                valid_counts["SCAR"]
            )

        if "DARK_SPOT" in classes:
            score += (
                100 -
                valid_counts["DARK_SPOT"]
            )

        return score

    best = max(
        candidates,
        key=valid_score,
    )

    face_valid.append(
        best
    )

    remaining.remove(
        best
    )

    for cls in record_classes(
        best
    ):

        valid_counts[
            cls
        ] += 1


# =============================================================================
# FILL TEST
# =============================================================================

test_counts = Counter()

for record in face_test:

    for cls in record_classes(
        record
    ):

        test_counts[
            cls
        ] += 1


while (
    len(face_test)
    < target_test
    and remaining
):

    candidates = list(
        remaining
    )

    def test_score(record):

        score = 0

        classes = record_classes(
            record
        )

        if "SCAR" in classes:
            score += (
                100 -
                test_counts["SCAR"]
            )

        if "DARK_SPOT" in classes:
            score += (
                100 -
                test_counts["DARK_SPOT"]
            )

        return score

    best = max(
        candidates,
        key=test_score,
    )

    face_test.append(
        best
    )

    remaining.remove(
        best
    )

    for cls in record_classes(
        best
    ):

        test_counts[
            cls
        ] += 1


# =============================================================================
# REMAINING = TRAIN
# =============================================================================

face_train = remaining


# =============================================================================
# RESET FINAL SOURCE RECORDS
# =============================================================================

final_source_records = {
    "train": [],
    "valid": [],
    "test": [],
}


# PIE — preserve official splits
for record in records_by_source[
    "pie"
]:

    split = record[
        "original_split"
    ]

    final_source_records[
        split
    ].append(
        record
    )


# MELASMA — preserve official splits
for record in records_by_source[
    "melasma"
]:

    split = record[
        "original_split"
    ]

    final_source_records[
        split
    ].append(
        record
    )


# FACE_SEGMENT — new stratified split
final_source_records[
    "train"
].extend(
    face_train
)

final_source_records[
    "valid"
].extend(
    face_valid
)

final_source_records[
    "test"
].extend(
    face_test
)


# =============================================================================
# REPORT FACE SPLIT
# =============================================================================

print("\nFACE_SEGMENT SPLIT")

for split_name, records in [
    ("TRAIN", face_train),
    ("VALID", face_valid),
    ("TEST", face_test),
]:

    counts = Counter()

    for record in records:

        for cls in record_classes(
            record
        ):

            counts[
                cls
            ] += 1

    print(
        f"\n{split_name}"
    )

    print(
        f"  Images       : {len(records)}"
    )

    print(
        f"  SCAR         : {counts['SCAR']}"
    )

    print(
        f"  DARK_SPOT    : {counts['DARK_SPOT']}"
    )


# =============================================================================
# CORRECT SOURCE-PATH LEAKAGE CHECK
# =============================================================================

print("\n" + "=" * 80)
print("CORRECTED SOURCE IMAGE LEAKAGE CHECK")
print("=" * 80)


split_identity = defaultdict(
    set
)

for split in [
    "train",
    "valid",
    "test",
]:

    for record in final_source_records[
        split
    ]:

        # -------------------------------------------------------------
        # CRITICAL:
        # Physical image path is the true unique identity.
        # -------------------------------------------------------------

        identity = str(
            Path(
                record[
                    "image_path"
                ]
            ).resolve()
        )

        split_identity[
            identity
        ].add(
            split
        )


leakage = {

    identity: splits

    for identity, splits
    in split_identity.items()

    if len(splits) > 1
}


print(
    "Cross-split physical-image leakage:",
    len(leakage)
)


if leakage:

    print(
        "\nFirst leakage examples:"
    )

    for identity, splits in list(
        leakage.items()
    )[:10]:

        print(
            " ",
            identity,
            "->",
            sorted(splits),
        )

    raise RuntimeError(
        "REAL CROSS-SPLIT IMAGE LEAKAGE DETECTED."
    )


print(
    "✓ No physical image occurs in multiple splits."
)
```

```python
# =============================================================================
# CLINDERMA STAGE 8K
# FINAL SEGMENTATION DATASET REBUILD — ORIENTATION-SAFE VERSION
# =============================================================================
#
# PURPOSE
# -------
# Rebuild the final multi-source segmentation dataset from the RAW COCO data.
#
# TARGET CLASSES
#   0: background
#   1: PIH
#   2: PIE
#   3: SCAR
#   4: DARK_SPOT
#   5: MELASMA
#
# IMPORTANT
# ---------
# - PIH remains unavailable and is NOT fabricated.
# - Hyperpigmentation detection remains separate.
# - Bounding boxes are NOT converted into masks.
# - Validation/test are NOT augmented.
# - RAW data is never modified.
# - Image dimensions are taken from the ACTUAL image file.
# - COCO/image orientation mismatches are handled before mask creation.
# =============================================================================

import os
import json
import shutil
import hashlib
from pathlib import Path
from collections import defaultdict

import numpy as np
from PIL import Image, ImageDraw, ImageOps

try:
    from pycocotools import mask as mask_utils
except ImportError:
    raise RuntimeError(
        "pycocotools is required. Install it before running Stage 8K."
    )


# =============================================================================
# CONFIGURATION
# =============================================================================

RAW_ROOT = Path(
    "/content/clinderma_workspace/raw_multisource"
)

OUTPUT_ROOT = Path(
    "/content/clinderma_workspace/processed/pigmentation_multisource_final"
)

REPORT_ROOT = Path(
    "/content/clinderma_workspace/audit_multisource"
)

REPORT_PATH = (
    REPORT_ROOT /
    "stage8k_final_dataset_rebuild_report.json"
)

MANIFEST_PATH = (
    OUTPUT_ROOT /
    "dataset_manifest.json"
)


# =============================================================================
# TARGET CLASS DEFINITIONS
# =============================================================================

TARGET_CLASSES = {
    0: "background",
    1: "PIH",
    2: "PIE",
    3: "SCAR",
    4: "DARK_SPOT",
    5: "MELASMA",
}

CLASS_NAME_TO_ID = {
    name: class_id
    for class_id, name in TARGET_CLASSES.items()
}


# =============================================================================
# APPROVED SOURCE MAPPINGS
# =============================================================================

SOURCE_MAPPINGS = {

    "pie": {
        "PIE-Annotation": 2,
    },

    "melasma": {
        "melasma": 5,
    },

    "face_segment": {
        "Scar": 3,
        "Dark spots": 4,
    },
}


# =============================================================================
# SOURCE DATASETS
# =============================================================================

SOURCE_DATASETS = {
    "pie": {
        "root": RAW_ROOT / "pie" / "version_1",
    },

    "melasma": {
        "root": RAW_ROOT / "melasma" / "version_1",
    },

    "face_segment": {
        "root": RAW_ROOT / "face_segment" / "version_1",
    },
}


# =============================================================================
# TRAINING BALANCING
# =============================================================================
#
# These are the final target exposures established in Stage 8H.
#
# PIH = 0 because there is currently no approved PIH segmentation dataset.
# =============================================================================

TRAIN_TARGET_EXPOSURE = {
    1: 0,      # PIH
    2: 398,    # PIE
    3: 350,    # SCAR
    4: 250,    # DARK_SPOT
    5: 432,    # MELASMA
}


# =============================================================================
# CREATE / CLEAN OUTPUT
# =============================================================================

print("=" * 80)
print("CLINDERMA STAGE 8K")
print("FINAL SEGMENTATION DATASET REBUILD — ORIENTATION-SAFE")
print("=" * 80)

print()
print("TARGET CLASSES")

for class_id, class_name in TARGET_CLASSES.items():
    print(f"  {class_id}: {class_name}")

print()
print("Output:")
print(f"  {OUTPUT_ROOT}")

print()

# -------------------------------------------------------------------------
# IMPORTANT:
# The final dataset is rebuilt from scratch.
# RAW data is never touched.
# -------------------------------------------------------------------------

if OUTPUT_ROOT.exists():

    print("Removing previous incomplete final dataset...")

    shutil.rmtree(
        OUTPUT_ROOT
    )

    print("✓ Previous final dataset removed")

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# =============================================================================
# CREATE SPLIT DIRECTORIES
# =============================================================================

for split in [
    "train",
    "valid",
    "test",
]:

    (OUTPUT_ROOT / split / "images").mkdir(
        parents=True,
        exist_ok=True
    )

    (OUTPUT_ROOT / split / "masks").mkdir(
        parents=True,
        exist_ok=True
    )


# =============================================================================
# HELPERS
# =============================================================================

def normalize_source_name(name):
    """
    Normalize source names used internally.
    """

    name = name.lower().strip()

    aliases = {
        "face-segment": "face_segment",
        "face_segment": "face_segment",
        "pie": "pie",
        "melasma": "melasma",
    }

    return aliases.get(
        name,
        name
    )


def find_coco_files(dataset_root):
    """
    Find all COCO annotation files.
    """

    return sorted(
        dataset_root.rglob(
            "_annotations.coco.json"
        )
    )


def find_image_path(
    dataset_root,
    split,
    file_name
):
    """
    Resolve an image path safely.
    """

    candidate_paths = [

        dataset_root /
        split /
        file_name,

        dataset_root /
        file_name,
    ]

    for candidate in candidate_paths:

        if candidate.exists():
            return candidate

    # Last-resort filename search.
    matches = list(
        dataset_root.rglob(
            Path(file_name).name
        )
    )

    if matches:
        return matches[0]

    return None


def get_file_hash(path):
    """
    SHA256 hash used for physical-image leakage checking.
    """

    sha = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(
                1024 * 1024
            )

            if not chunk:
                break

            sha.update(chunk)

    return sha.hexdigest()


def load_image_orientation_safe(path):
    """
    Load image while applying EXIF orientation.

    Returns:
        PIL RGB image
    """

    with Image.open(path) as img:

        img = ImageOps.exif_transpose(
            img
        )

        img = img.convert(
            "RGB"
        )

        return img.copy()


def actual_image_size(path):
    """
    Get actual physical dimensions after EXIF orientation.
    """

    with Image.open(path) as img:

        img = ImageOps.exif_transpose(
            img
        )

        return img.size


def polygon_to_points(
    polygon,
    coco_width,
    coco_height,
    actual_width,
    actual_height
):
    """
    Convert COCO polygon coordinates into the actual physical image
    coordinate system.

    Handles the common case where COCO metadata dimensions are transposed
    relative to the physical image dimensions.
    """

    points = []

    if not isinstance(
        polygon,
        list
    ):
        return points

    if len(polygon) < 6:
        return points

    # -------------------------------------------------------------------------
    # Determine whether COCO dimensions are transposed relative to the
    # physical image.
    # -------------------------------------------------------------------------

    transposed = (
        coco_width == actual_height
        and
        coco_height == actual_width
        and
        coco_width != actual_width
    )

    for i in range(
        0,
        len(polygon) - 1,
        2
    ):

        x = float(
            polygon[i]
        )

        y = float(
            polygon[i + 1]
        )

        # ---------------------------------------------------------------------
        # IMPORTANT ORIENTATION FIX
        #
        # If metadata says:
        #   COCO = 1080 x 1920
        #
        # but actual image is:
        #   1920 x 1080
        #
        # swap coordinate system appropriately.
        # ---------------------------------------------------------------------

        if transposed:

            new_x = y

            new_y = (
                coco_width - x
            )

            x = new_x
            y = new_y

        # ---------------------------------------------------------------------
        # Clamp coordinates.
        # ---------------------------------------------------------------------

        x = max(
            0.0,
            min(
                x,
                actual_width - 1
            )
        )

        y = max(
            0.0,
            min(
                y,
                actual_height - 1
            )
        )

        points.append(
            (x, y)
        )

    return points


def create_mask(
    image_path,
    image_info,
    annotations,
    output_mask
):
    """
    Create a semantic segmentation mask.

    Mask values:
      0 = background
      1 = PIH
      2 = PIE
      3 = SCAR
      4 = DARK_SPOT
      5 = MELASMA

    IMPORTANT:
    Actual image dimensions are used instead of blindly trusting COCO
    metadata dimensions.
    """

    # -------------------------------------------------------------------------
    # Get actual dimensions AFTER EXIF orientation.
    # -------------------------------------------------------------------------

    actual_width, actual_height = (
        actual_image_size(
            image_path
        )
    )

    # -------------------------------------------------------------------------
    # COCO metadata dimensions.
    # -------------------------------------------------------------------------

    coco_width = int(
        image_info.get(
            "width",
            actual_width
        )
    )

    coco_height = int(
        image_info.get(
            "height",
            actual_height
        )
    )

    # -------------------------------------------------------------------------
    # Create empty semantic mask.
    # -------------------------------------------------------------------------

    mask_array = np.zeros(
        (
            actual_height,
            actual_width
        ),
        dtype=np.uint8
    )

    # -------------------------------------------------------------------------
    # Draw polygons.
    # -------------------------------------------------------------------------

    polygon_mask = Image.fromarray(
        mask_array
    )

    draw = ImageDraw.Draw(
        polygon_mask
    )

    # -------------------------------------------------------------------------
    # Process annotations.
    # -------------------------------------------------------------------------

    for ann in annotations:

        class_id = ann.get(
            "_target_class_id"
        )

        if class_id is None:
            continue

        class_id = int(
            class_id
        )

        segmentation = ann.get(
            "segmentation"
        )

        if not segmentation:
            continue

        # =====================================================================
        # POLYGON SEGMENTATION
        # =====================================================================

        if isinstance(
            segmentation,
            list
        ):

            for polygon in segmentation:

                points = polygon_to_points(
                    polygon,
                    coco_width,
                    coco_height,
                    actual_width,
                    actual_height
                )

                if len(points) >= 3:

                    draw.polygon(
                        points,
                        fill=class_id
                    )

        # =====================================================================
        # RLE SEGMENTATION
        # =====================================================================

        elif isinstance(
            segmentation,
            dict
        ):

            try:

                decoded = mask_utils.decode(
                    segmentation
                )

                if decoded.ndim == 3:

                    decoded = decoded[:, :, 0]

                decoded = (
                    decoded.astype(bool)
                )

                rle_height, rle_width = (
                    decoded.shape
                )

                # -----------------------------------------------------------------
                # Exact match.
                # -----------------------------------------------------------------

                if (
                    rle_width == actual_width
                    and
                    rle_height == actual_height
                ):

                    decoded_final = decoded

                # -----------------------------------------------------------------
                # Transposed RLE.
                # -----------------------------------------------------------------

                elif (
                    rle_width == actual_height
                    and
                    rle_height == actual_width
                ):

                    decoded_final = decoded.T

                else:

                    print(
                        "WARNING: RLE dimensions incompatible."
                    )

                    print(
                        f"  Image: {image_path}"
                    )

                    print(
                        f"  RLE: {(rle_width, rle_height)}"
                    )

                    print(
                        f"  Actual: {(actual_width, actual_height)}"
                    )

                    continue

                mask_array = np.array(
                    polygon_mask,
                    dtype=np.uint8
                )

                mask_array[
                    decoded_final
                ] = class_id

                polygon_mask = Image.fromarray(
                    mask_array
                )

                draw = ImageDraw.Draw(
                    polygon_mask
                )

            except Exception as exc:

                print(
                    "WARNING: Failed to decode RLE:"
                )

                print(
                    f"  {image_path}"
                )

                print(
                    f"  {exc}"
                )

    # -------------------------------------------------------------------------
    # Save mask.
    # -------------------------------------------------------------------------

    polygon_mask.save(
        output_mask
    )

    return (
        actual_width,
        actual_height
    )


# =============================================================================
# DISCOVER COCO FILES
# =============================================================================

print()
print("=" * 80)
print("DISCOVERING COCO ANNOTATIONS")
print("=" * 80)

coco_files_by_source = {}

for source_name, config in SOURCE_DATASETS.items():

    files = find_coco_files(
        config["root"]
    )

    coco_files_by_source[
        source_name
    ] = files

    print()
    print(
        f"{source_name.upper()}:"
    )

    for file in files:

        print(
            f"  {file}"
        )


# =============================================================================
# LOAD ALL APPROVED DATA
# =============================================================================

records_by_split = {
    "train": [],
    "valid": [],
    "test": [],
}

global_hashes = {}

source_statistics = defaultdict(
    lambda: {
        "images": 0,
        "annotations": 0,
        "classes": defaultdict(int),
    }
)

print()
print("=" * 80)
print("LOADING COCO DATA")
print("=" * 80)


for source_name, coco_files in (
    coco_files_by_source.items()
):

    print()
    print(
        "=" * 80
    )

    print(
        f"SOURCE: {source_name.upper()}"
    )

    print(
        "=" * 80
    )

    mapping = SOURCE_MAPPINGS.get(
        source_name,
        {}
    )

    for coco_path in coco_files:

        # ---------------------------------------------------------------------
        # Infer split from parent directory.
        # ---------------------------------------------------------------------

        split = coco_path.parent.name.lower()

        if split not in {
            "train",
            "valid",
            "test",
        }:

            print(
                f"Skipping unknown split: {coco_path}"
            )

            continue

        print()
        print(
            f"Processing {source_name} / {split}"
        )

        with open(
            coco_path,
            "r",
            encoding="utf-8"
        ) as f:

            coco = json.load(
                f
            )

        categories = {
            int(cat["id"]): cat["name"]
            for cat in coco.get(
                "categories",
                []
            )
        }

        # ---------------------------------------------------------------------
        # Image records.
        # ---------------------------------------------------------------------

        image_records = {
            int(img["id"]): img
            for img in coco.get(
                "images",
                []
            )
        }

        # ---------------------------------------------------------------------
        # Group annotations by image.
        # ---------------------------------------------------------------------

        annotations_by_image = defaultdict(
            list
        )

        for ann in coco.get(
            "annotations",
            []
        ):

            category_id = int(
                ann["category_id"]
            )

            raw_class = categories.get(
                category_id
            )

            # -----------------------------------------------------------------
            # Only approved semantic mappings.
            # -----------------------------------------------------------------

            if raw_class not in mapping:
                continue

            target_class_id = mapping[
                raw_class
            ]

            ann_copy = dict(
                ann
            )

            ann_copy[
                "_target_class_id"
            ] = target_class_id

            annotations_by_image[
                int(
                    ann["image_id"]
                )
            ].append(
                ann_copy
            )

        # ---------------------------------------------------------------------
        # Build image-level records.
        # ---------------------------------------------------------------------

        for image_id, image_info in (
            image_records.items()
        ):

            file_name = image_info.get(
                "file_name"
            )

            image_path = find_image_path(
                SOURCE_DATASETS[
                    source_name
                ]["root"],
                split,
                file_name
            )

            if image_path is None:

                print(
                    "WARNING: image not found:"
                )

                print(
                    file_name
                )

                continue

            annotations = (
                annotations_by_image.get(
                    image_id,
                    []
                )
            )

            # -----------------------------------------------------------------
            # Only keep images containing approved targets.
            #
            # This prevents the huge number of unrelated negative images in
            # FACE_SEGMENT from overwhelming the segmentation training set.
            # -----------------------------------------------------------------

            if len(annotations) == 0:
                continue

            # -----------------------------------------------------------------
            # Physical image hash.
            # -----------------------------------------------------------------

            image_hash = get_file_hash(
                image_path
            )

            record = {
                "source": source_name,
                "split": split,
                "image_id": image_id,
                "file_name": file_name,
                "image_path": str(
                    image_path
                ),
                "image_hash": image_hash,
                "image_info": image_info,
                "annotations": annotations,
            }

            records_by_split[
                split
            ].append(
                record
            )

            global_hashes.setdefault(
                image_hash,
                []
            ).append(
                (
                    source_name,
                    split,
                    str(image_path)
                )
            )

            # -----------------------------------------------------------------
            # Statistics.
            # -----------------------------------------------------------------

            source_statistics[
                source_name
            ]["images"] += 1

            source_statistics[
                source_name
            ]["annotations"] += len(
                annotations
            )

            for ann in annotations:

                class_id = ann[
                    "_target_class_id"
                ]

                source_statistics[
                    source_name
                ]["classes"][
                    TARGET_CLASSES[
                        class_id
                    ]
                ] += 1


# =============================================================================
# SOURCE IMAGE LEAKAGE CHECK
# =============================================================================

print()
print("=" * 80)
print("SOURCE IMAGE LEAKAGE CHECK")
print("=" * 80)

leakage_groups = []

for image_hash, occurrences in (
    global_hashes.items()
):

    splits = set(
        occurrence[1]
        for occurrence in occurrences
    )

    if len(splits) > 1:

        leakage_groups.append(
            {
                "hash": image_hash,
                "occurrences": occurrences,
            }
        )

print(
    f"Cross-split physical-image leakage: "
    f"{len(leakage_groups)}"
)

if leakage_groups:

    print(
        "WARNING: Physical images occur in multiple splits."
    )

    for group in leakage_groups[:10]:

        print(
            group
        )

    raise RuntimeError(
        "SOURCE IMAGE LEAKAGE DETECTED. "
        "Do not train until the split assignment is corrected."
    )

else:

    print(
        "✓ No physical image occurs in multiple splits."
    )


# =============================================================================
# REMOVE CROSS-SOURCE / CROSS-SPLIT DUPLICATES FROM VALIDATION AND TEST
# =============================================================================
#
# The previous leakage checks ensure a physical image isn't already duplicated
# across splits. This section additionally ensures training records are unique.
# =============================================================================

def deduplicate_records(records):

    seen = set()
    unique = []

    for record in records:

        image_hash = record[
            "image_hash"
        ]

        if image_hash in seen:
            continue

        seen.add(
            image_hash
        )

        unique.append(
            record
        )

    return unique


for split in records_by_split:

    before = len(
        records_by_split[
            split
        ]
    )

    records_by_split[
        split
    ] = deduplicate_records(
        records_by_split[
            split
        ]
    )

    after = len(
        records_by_split[
            split
        ]
    )

    if before != after:

        print(
            f"{split}: removed "
            f"{before - after} duplicate records"
        )


# =============================================================================
# CLASS-SPECIFIC RECORD GROUPING
# =============================================================================

train_records_by_class = defaultdict(
    list
)

for record in records_by_split[
    "train"
]:

    classes_in_image = set()

    for ann in record[
        "annotations"
    ]:

        classes_in_image.add(
            int(
                ann[
                    "_target_class_id"
                ]
            )
        )

    for class_id in classes_in_image:

        train_records_by_class[
            class_id
        ].append(
            record
        )


print()
print("=" * 80)
print("REAL TRAINING IMAGE COUNTS")
print("=" * 80)

for class_id in range(
    1,
    6
):

    print(
        f"  {TARGET_CLASSES[class_id]:<10}: "
        f"{len(train_records_by_class[class_id])}"
    )


# =============================================================================
# COPY / CREATE TRAINING DATA
# =============================================================================

print()
print("=" * 80)
print("BUILDING REAL TRAINING DATA")
print("=" * 80)


final_train_records = []

# -------------------------------------------------------------------------
# We select records class-by-class according to Stage 8H exposure targets.
#
# IMPORTANT:
# If a class has fewer real images than its target, ALL real images are kept.
# Augmentation is then used to increase exposure.
# -------------------------------------------------------------------------

for class_id in range(
    1,
    6
):

    class_records = (
        train_records_by_class[
            class_id
        ]
    )

    target_count = TRAIN_TARGET_EXPOSURE[
        class_id
    ]

    class_name = TARGET_CLASSES[
        class_id
    ]

    if len(class_records) == 0:

        print(
            f"{class_name}: no approved training images"
        )

        continue

    # ---------------------------------------------------------------------
    # We don't discard real data merely because the target is lower.
    # ---------------------------------------------------------------------

    if (
        target_count <= 0
        or
        len(class_records) <= target_count
    ):

        selected = list(
            class_records
        )

    else:

        # -----------------------------------------------------------------
        # Deterministic sampling.
        # -----------------------------------------------------------------

        rng = np.random.default_rng(
            42 + class_id
        )

        indices = rng.choice(
            len(class_records),
            size=target_count,
            replace=False
        )

        selected = [
            class_records[
                int(i)
            ]
            for i in indices
        ]

    final_train_records.extend(
        selected
    )

    print(
        f"{class_name}: "
        f"{len(selected)} real training images selected"
    )


# =============================================================================
# DEDUPLICATE FINAL TRAINING RECORDS
# =============================================================================

unique_train_records = []

seen_hashes = set()

for record in final_train_records:

    image_hash = record[
        "image_hash"
    ]

    if image_hash in seen_hashes:
        continue

    seen_hashes.add(
        image_hash
    )

    unique_train_records.append(
        record
    )

final_train_records = (
    unique_train_records
)


# =============================================================================
# WRITE TRAINING IMAGE + MASK
# =============================================================================

train_output_dir = (
    OUTPUT_ROOT /
    "train"
)

train_image_dir = (
    train_output_dir /
    "images"
)

train_mask_dir = (
    train_output_dir /
    "masks"
)


print()
print("=" * 80)
print("CREATING TRAINING IMAGE + MASK PAIRS")
print("=" * 80)

train_written = []

for index, record in enumerate(
    final_train_records
):

    source = record[
        "source"
    ]

    original_name = Path(
        record["file_name"]
    ).stem

    extension = Path(
        record["file_name"]
    ).suffix.lower()

    if extension not in {
        ".jpg",
        ".jpeg",
        ".png",
        ".webp",
    }:

        extension = ".jpg"

    unique_name = (
        f"{source}_"
        f"{index:06d}_"
        f"{original_name}"
    )

    output_image = (
        train_image_dir /
        f"{unique_name}{extension}"
    )

    output_mask = (
        train_mask_dir /
        f"{unique_name}.png"
    )

    # -------------------------------------------------------------------------
    # Load image with EXIF orientation.
    # -------------------------------------------------------------------------

    image = load_image_orientation_safe(
        record["image_path"]
    )

    actual_size = image.size

    # -------------------------------------------------------------------------
    # Save normalized image.
    # -------------------------------------------------------------------------

    image.save(
        output_image,
        quality=95
    )

    # -------------------------------------------------------------------------
    # Build segmentation mask using ACTUAL dimensions.
    # -------------------------------------------------------------------------

    mask_size = create_mask(
        record["image_path"],
        record["image_info"],
        record["annotations"],
        output_mask
    )

    # -------------------------------------------------------------------------
    # Verify image/mask dimensions immediately.
    # -------------------------------------------------------------------------

    with Image.open(
        output_image
    ) as img_check:

        image_size = img_check.size

    with Image.open(
        output_mask
    ) as mask_check:

        mask_size_actual = mask_check.size

    if image_size != mask_size_actual:

        raise RuntimeError(
            "\nImage/mask dimension mismatch:\n"
            f"Image: {output_image}\n"
            f"Mask : {output_mask}\n"
            f"Image size: {image_size}\n"
            f"Mask size : {mask_size_actual}\n"
        )

    train_written.append(
        {
            "image": str(
                output_image
            ),
            "mask": str(
                output_mask
            ),
            "source": source,
            "source_image": record[
                "image_path"
            ],
            "image_hash": record[
                "image_hash"
            ],
        }
    )


print(
    f"Training image/mask pairs created: "
    f"{len(train_written)}"
)


# =============================================================================
# AUGMENTATION
# =============================================================================
#
# We now create additional training examples for:
#
#   SCAR       -> target exposure 350
#   DARK_SPOT  -> target exposure 250
#
# Image and mask are transformed JOINTLY.
#
# No augmentation is applied to validation/test.
# =============================================================================

import random


def joint_augmentation(
    image,
    mask,
    seed
):
    """
    Apply deterministic spatial augmentation jointly to image and mask.
    """

    rng = random.Random(
        seed
    )

    # -------------------------------------------------------------------------
    # Horizontal flip
    # -------------------------------------------------------------------------

    if rng.random() < 0.5:

        image = image.transpose(
            Image.Transpose.FLIP_LEFT_RIGHT
        )

        mask = mask.transpose(
            Image.Transpose.FLIP_LEFT_RIGHT
        )

    # -------------------------------------------------------------------------
    # Vertical flip
    # -------------------------------------------------------------------------

    if rng.random() < 0.15:

        image = image.transpose(
            Image.Transpose.FLIP_TOP_BOTTOM
        )

        mask = mask.transpose(
            Image.Transpose.FLIP_TOP_BOTTOM
        )

    # -------------------------------------------------------------------------
    # 90-degree rotations
    # -------------------------------------------------------------------------

    rotation_choice = rng.choice(
        [0, 90, 180, 270]
    )

    if rotation_choice == 90:

        image = image.transpose(
            Image.Transpose.ROTATE_90
        )

        mask = mask.transpose(
            Image.Transpose.ROTATE_90
        )

    elif rotation_choice == 180:

        image = image.transpose(
            Image.Transpose.ROTATE_180
        )

        mask = mask.transpose(
            Image.Transpose.ROTATE_180
        )

    elif rotation_choice == 270:

        image = image.transpose(
            Image.Transpose.ROTATE_270
        )

        mask = mask.transpose(
            Image.Transpose.ROTATE_270
        )

    # -------------------------------------------------------------------------
    # IMPORTANT:
    # No independent color transformation is applied to masks.
    # -------------------------------------------------------------------------

    return image, mask


# =============================================================================
# DETERMINE EXISTING CLASS EXPOSURE
# =============================================================================

def mask_contains_class(
    mask_path,
    class_id
):

    with Image.open(
        mask_path
    ) as mask:

        array = np.array(
            mask
        )

    return bool(
        np.any(
            array == class_id
        )
    )


current_class_files = defaultdict(
    list
)

for item in train_written:

    mask_path = Path(
        item["mask"]
    )

    for class_id in range(
        1,
        6
    ):

        if mask_contains_class(
            mask_path,
            class_id
        ):

            current_class_files[
                class_id
            ].append(
                item
            )


print()
print("=" * 80)
print("CURRENT TRAINING CLASS EXPOSURE")
print("=" * 80)

for class_id in range(
    1,
    6
):

    print(
        f"  {TARGET_CLASSES[class_id]:<10}: "
        f"{len(current_class_files[class_id])}"
    )


# =============================================================================
# AUGMENTATION TARGETS
# =============================================================================

augmentation_summary = defaultdict(
    int
)

augmentation_records = []


for class_id in [
    3,  # SCAR
    4,  # DARK_SPOT
]:

    class_name = TARGET_CLASSES[
        class_id
    ]

    target = TRAIN_TARGET_EXPOSURE[
        class_id
    ]

    existing = current_class_files[
        class_id
    ]

    if len(existing) == 0:

        print()
        print(
            f"{class_name}: no real data available; "
            "augmentation skipped."
        )

        continue

    required = max(
        0,
        target - len(existing)
    )

    print()
    print(
        f"{class_name}"
    )

    print(
        f"  Real images      : {len(existing)}"
    )

    print(
        f"  Target exposure  : {target}"
    )

    print(
        f"  Augmented images : {required}"
    )

    for aug_index in range(
        required
    ):

        # ---------------------------------------------------------------------
        # Deterministic selection of source.
        # ---------------------------------------------------------------------

        source_item = existing[
            aug_index % len(existing)
        ]

        source_image = Path(
            source_item["image"]
        )

        source_mask = Path(
            source_item["mask"]
        )

        with Image.open(
            source_image
        ) as img:

            image = img.convert(
                "RGB"
            )

        with Image.open(
            source_mask
        ) as mask:

            mask = mask.convert(
                "L"
            )

        # ---------------------------------------------------------------------
        # Joint augmentation.
        # ---------------------------------------------------------------------

        seed = (
            800000
            +
            class_id * 10000
            +
            aug_index
        )

        image_aug, mask_aug = (
            joint_augmentation(
                image,
                mask,
                seed
            )
        )

        aug_name = (
            f"aug_"
            f"{class_name.lower()}_"
            f"{aug_index:04d}"
        )

        output_image = (
            train_image_dir /
            f"{aug_name}.jpg"
        )

        output_mask = (
            train_mask_dir /
            f"{aug_name}.png"
        )

        image_aug.save(
            output_image,
            quality=95
        )

        mask_aug.save(
            output_mask
        )

        # ---------------------------------------------------------------------
        # Verify augmented pair.
        # ---------------------------------------------------------------------

        if (
            image_aug.size
            !=
            mask_aug.size
        ):

            raise RuntimeError(
                "Augmented image/mask mismatch:\n"
                f"Image: {output_image}\n"
                f"Mask: {output_mask}\n"
                f"Image size: {image_aug.size}\n"
                f"Mask size: {mask_aug.size}"
            )

        augmentation_summary[
            class_name
        ] += 1

        augmentation_records.append(
            {
                "image": str(
                    output_image
                ),
                "mask": str(
                    output_mask
                ),
                "source_image": str(
                    source_image
                ),
                "source_mask": str(
                    source_mask
                ),
                "class": class_name,
                "augmentation": True,
            }
        )


# =============================================================================
# VALIDATION / TEST
# =============================================================================
#
# These remain completely untouched.
#
# We use the ORIGINAL split assignments from the source datasets.
# =============================================================================

print()
print("=" * 80)
print("COPYING VALIDATION / TEST DATA UNTOUCHED")
print("=" * 80)


validation_test_written = {
    "valid": [],
    "test": [],
}


for split in [
    "valid",
    "test",
]:

    print()
    print(
        split.upper()
    )

    image_dir = (
        OUTPUT_ROOT /
        split /
        "images"
    )

    mask_dir = (
        OUTPUT_ROOT /
        split /
        "masks"
    )

    for index, record in enumerate(
        records_by_split[
            split
        ]
    ):

        source = record[
            "source"
        ]

        original_name = Path(
            record["file_name"]
        ).stem

        extension = Path(
            record["file_name"]
        ).suffix.lower()

        if extension not in {
            ".jpg",
            ".jpeg",
            ".png",
            ".webp",
        }:

            extension = ".jpg"

        unique_name = (
            f"{source}_"
            f"{index:06d}_"
            f"{original_name}"
        )

        output_image = (
            image_dir /
            f"{unique_name}{extension}"
        )

        output_mask = (
            mask_dir /
            f"{unique_name}.png"
        )

        # ---------------------------------------------------------------------
        # Load and apply EXIF orientation.
        #
        # This is NOT augmentation; it is orientation normalization.
        # ---------------------------------------------------------------------

        image = load_image_orientation_safe(
            record["image_path"]
        )

        image.save(
            output_image,
            quality=95
        )

        # ---------------------------------------------------------------------
        # Build mask using actual image dimensions.
        # ---------------------------------------------------------------------

        create_mask(
            record["image_path"],
            record["image_info"],
            record["annotations"],
            output_mask
        )

        # ---------------------------------------------------------------------
        # Verify pair.
        # ---------------------------------------------------------------------

        with Image.open(
            output_image
        ) as img_check:

            image_size = img_check.size

        with Image.open(
            output_mask
        ) as mask_check:

            mask_size = mask_check.size

        if image_size != mask_size:

            raise RuntimeError(
                "Validation/test image-mask mismatch:\n"
                f"Image: {output_image}\n"
                f"Mask: {output_mask}\n"
                f"Image size: {image_size}\n"
                f"Mask size: {mask_size}"
            )

        validation_test_written[
            split
        ].append(
            {
                "image": str(
                    output_image
                ),
                "mask": str(
                    output_mask
                ),
                "source": source,
                "source_image": record[
                    "image_path"
                ],
                "image_hash": record[
                    "image_hash"
                ],
            }
        )


# =============================================================================
# FINAL MASK VALIDATION
# =============================================================================

print()
print("=" * 80)
print("FINAL MASK VALIDATION")
print("=" * 80)


final_statistics = {}

for split in [
    "train",
    "valid",
    "test",
]:

    image_dir = (
        OUTPUT_ROOT /
        split /
        "images"
    )

    mask_dir = (
        OUTPUT_ROOT /
        split /
        "masks"
    )

    image_files = sorted(
        [
            p
            for p in image_dir.iterdir()
            if p.is_file()
        ]
    )

    mask_files = sorted(
        [
            p
            for p in mask_dir.iterdir()
            if p.is_file()
        ]
    )

    image_stems = {
        p.stem
        for p in image_files
    }

    mask_stems = {
        p.stem
        for p in mask_files
    }

    missing_masks = (
        image_stems -
        mask_stems
    )

    orphan_masks = (
        mask_stems -
        image_stems
    )

    invalid_masks = []

    class_pixel_counts = defaultdict(
        int
    )

    for mask_path in mask_files:

        try:

            with Image.open(
                mask_path
            ) as mask:

                array = np.array(
                    mask
                )

                if array.ndim != 2:

                    invalid_masks.append(
                        (
                            str(mask_path),
                            "not single-channel"
                        )
                    )

                    continue

                unique_values = set(
                    np.unique(
                        array
                    ).tolist()
                )

                invalid_values = (
                    unique_values
                    -
                    set(
                        TARGET_CLASSES.keys()
                    )
                )

                if invalid_values:

                    invalid_masks.append(
                        (
                            str(mask_path),
                            f"invalid class IDs: "
                            f"{sorted(invalid_values)}"
                        )
                    )

                for class_id in unique_values:

                    if class_id in TARGET_CLASSES:

                        class_pixel_counts[
                            class_id
                        ] += int(
                            np.sum(
                                array == class_id
                            )
                        )

        except Exception as exc:

            invalid_masks.append(
                (
                    str(mask_path),
                    str(exc)
                )
            )

    final_statistics[
        split
    ] = {
        "images": len(
            image_files
        ),
        "masks": len(
            mask_files
        ),
        "missing_masks": len(
            missing_masks
        ),
        "orphan_masks": len(
            orphan_masks
        ),
        "invalid_masks": len(
            invalid_masks
        ),
        "class_pixel_counts": {
            TARGET_CLASSES[
                class_id
            ]: count
            for class_id, count
            in class_pixel_counts.items()
        },
    }

    print()
    print(
        split.upper()
    )

    print(
        f"  Images          : {len(image_files)}"
    )

    print(
        f"  Masks           : {len(mask_files)}"
    )

    print(
        f"  Missing masks   : {len(missing_masks)}"
    )

    print(
        f"  Orphan masks    : {len(orphan_masks)}"
    )

    print(
        f"  Invalid masks   : {len(invalid_masks)}"
    )

    if missing_masks:

        print(
            "  Missing examples:"
        )

        for item in list(
            missing_masks
        )[:5]:

            print(
                f"    {item}"
            )

    if orphan_masks:

        print(
            "  Orphan examples:"
        )

        for item in list(
            orphan_masks
        )[:5]:

            print(
                f"    {item}"
            )

    if invalid_masks:

        print(
            "  Invalid examples:"
        )

        for item in invalid_masks[:5]:

            print(
                f"    {item}"
            )


# =============================================================================
# FINAL CLASS IMAGE DISTRIBUTION
# =============================================================================

print()
print("=" * 80)
print("FINAL CLASS IMAGE DISTRIBUTION")
print("=" * 80)


def calculate_class_image_counts(
    image_dir,
    mask_dir
):

    counts = {
        class_id: 0
        for class_id in range(
            1,
            6
        )
    }

    for mask_path in mask_dir.glob(
        "*.png"
    ):

        with Image.open(
            mask_path
        ) as mask:

            array = np.array(
                mask
            )

        for class_id in counts:

            if np.any(
                array == class_id
            ):

                counts[
                    class_id
                ] += 1

    return counts


class_distribution = {}

for split in [
    "train",
    "valid",
    "test",
]:

    image_dir = (
        OUTPUT_ROOT /
        split /
        "images"
    )

    mask_dir = (
        OUTPUT_ROOT /
        split /
        "masks"
    )

    counts = calculate_class_image_counts(
        image_dir,
        mask_dir
    )

    class_distribution[
        split
    ] = counts

    print()
    print(
        split.upper()
    )

    for class_id, count in counts.items():

        print(
            f"  {TARGET_CLASSES[class_id]:<10}: "
            f"{count}"
        )


# =============================================================================
# VERIFY NO PIH FABRICATION
# =============================================================================

pih_train_count = (
    class_distribution[
        "train"
    ][1]
)

pih_valid_count = (
    class_distribution[
        "valid"
    ][1]
)

pih_test_count = (
    class_distribution[
        "test"
    ][1]
)

if (
    pih_train_count > 0
    or
    pih_valid_count > 0
    or
    pih_test_count > 0
):

    raise RuntimeError(
        "PIH data unexpectedly exists. "
        "PIH must remain unavailable until an approved PIH dataset "
        "is added."
    )

print()
print(
    "PIH verification:"
)

print(
    "  ✓ No PIH labels detected."
)


# =============================================================================
# VERIFY EXPECTED CLASS IDS
# =============================================================================

print()
print("=" * 80)
print("CLASS ID VERIFICATION")
print("=" * 80)

all_invalid_class_ids = []

for split in [
    "train",
    "valid",
    "test",
]:

    mask_dir = (
        OUTPUT_ROOT /
        split /
        "masks"
    )

    for mask_path in mask_dir.glob(
        "*.png"
    ):

        with Image.open(
            mask_path
        ) as mask:

            values = set(
                np.unique(
                    np.array(mask)
                ).tolist()
            )

        invalid = (
            values -
            set(
                TARGET_CLASSES.keys()
            )
        )

        if invalid:

            all_invalid_class_ids.append(
                (
                    str(mask_path),
                    sorted(invalid)
                )
            )

if all_invalid_class_ids:

    raise RuntimeError(
        "Invalid class IDs found in masks."
    )

print(
    "✓ All masks contain only class IDs 0–5."
)


# =============================================================================
# FINAL TRAINABILITY CHECK
# =============================================================================

print()
print("=" * 80)
print("TRAINABILITY CHECK")
print("=" * 80)


train_stats = final_statistics[
    "train"
]

valid_stats = final_statistics[
    "valid"
]

test_stats = final_statistics[
    "test"
]


trainable = True

for split, stats in [
    ("train", train_stats),
    ("valid", valid_stats),
    ("test", test_stats),
]:

    if stats[
        "images"
    ] != stats[
        "masks"
    ]:

        trainable = False

    if stats[
        "missing_masks"
    ] != 0:

        trainable = False

    if stats[
        "orphan_masks"
    ] != 0:

        trainable = False

    if stats[
        "invalid_masks"
    ] != 0:

        trainable = False


if train_stats[
    "images"
] == 0:

    trainable = False


if valid_stats[
    "images"
] == 0:

    trainable = False


if test_stats[
    "images"
] == 0:

    trainable = False


if trainable:

    print(
        "✓ SEGMENTATION DATASET IS TRAINABLE"
    )

else:

    print(
        "✗ SEGMENTATION DATASET IS NOT TRAINABLE"
    )

    raise RuntimeError(
        "Final dataset integrity checks failed. "
        "Do NOT begin SegFormer training."
    )


# =============================================================================
# FINAL EXPOSURE REPORT
# =============================================================================

print()
print("=" * 80)
print("FINAL TRAINING EXPOSURE")
print("=" * 80)

for class_id in range(
    1,
    6
):

    class_name = TARGET_CLASSES[
        class_id
    ]

    count = class_distribution[
        "train"
    ][class_id]

    print(
        f"  {class_name:<10}: {count}"
    )


print()
print("=" * 80)
print("AUGMENTATION CREATED")
print("=" * 80)

if augmentation_summary:

    for class_name, count in (
        augmentation_summary.items()
    ):

        print(
            f"  {class_name:<10}: {count}"
        )

else:

    print(
        "  None"
    )


# =============================================================================
# HYPERPIGMENTATION DETECTION DATASET
# =============================================================================

hyperpigmentation_path = (
    RAW_ROOT /
    "hyperpigmentation" /
    "version_1"
)

hyperpigmentation_present = (
    hyperpigmentation_path.exists()
)

print()
print("=" * 80)
print("HYPERPIGMENTATION DETECTION DATASET")
print("=" * 80)

print(
    f"Present: {hyperpigmentation_present}"
)

print(
    "Integration into segmentation: NO"
)

print(
    "Bounding boxes converted to masks: NO"
)


# =============================================================================
# MANIFEST
# =============================================================================

manifest = {

    "stage": "8K",

    "description":
        "Final orientation-safe multi-source segmentation dataset rebuild.",

    "target_classes":
        TARGET_CLASSES,

    "source_mappings":
        SOURCE_MAPPINGS,

    "output_root":
        str(OUTPUT_ROOT),

    "train": {
        "images":
            train_stats["images"],
        "masks":
            train_stats["masks"],
        "class_image_distribution": {
            TARGET_CLASSES[
                class_id
            ]: count
            for class_id, count
            in class_distribution[
                "train"
            ].items()
        },
    },

    "valid": {
        "images":
            valid_stats["images"],
        "masks":
            valid_stats["masks"],
        "class_image_distribution": {
            TARGET_CLASSES[
                class_id
            ]: count
            for class_id, count
            in class_distribution[
                "valid"
            ].items()
        },
    },

    "test": {
        "images":
            test_stats["images"],
        "masks":
            test_stats["masks"],
        "class_image_distribution": {
            TARGET_CLASSES[
                class_id
            ]: count
            for class_id, count
            in class_distribution[
                "test"
            ].items()
        },
    },

    "augmentation": dict(
        augmentation_summary
    ),

    "pih_status": {
        "approved_training_images":
            pih_train_count,
        "approved_valid_images":
            pih_valid_count,
        "approved_test_images":
            pih_test_count,
        "fabricated":
            False,
    },

    "hyperpigmentation": {
        "present":
            hyperpigmentation_present,
        "integrated":
            False,
        "boxes_converted_to_masks":
            False,
    },

    "integrity": final_statistics,

    "trainable":
        trainable,

    "raw_data_modified":
        False,

    "validation_test_augmented":
        False,

    "cross_split_physical_image_leakage":
        len(leakage_groups),

}


with open(
    MANIFEST_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# =============================================================================
# REPORT
# =============================================================================

report = {

    "stage": "8K",

    "output":
        str(OUTPUT_ROOT),

    "manifest":
        str(MANIFEST_PATH),

    "target_classes":
        TARGET_CLASSES,

    "source_mappings":
        SOURCE_MAPPINGS,

    "source_statistics":
        {
            source: {
                "images":
                    stats["images"],
                "annotations":
                    stats["annotations"],
                "classes":
                    dict(
                        stats["classes"]
                    ),
            }
            for source, stats
            in source_statistics.items()
        },

    "train_target_exposure":
        {
            TARGET_CLASSES[
                class_id
            ]: count
            for class_id, count
            in TRAIN_TARGET_EXPOSURE.items()
        },

    "augmentation_created":
        dict(
            augmentation_summary
        ),

    "class_distribution":
        {
            split: {
                TARGET_CLASSES[
                    class_id
                ]: count
                for class_id, count
                in counts.items()
            }
            for split, counts
            in class_distribution.items()
        },

    "integrity":
        final_statistics,

    "cross_split_leakage":
        len(leakage_groups),

    "pih":
        {
            "available":
                False,
            "fabricated":
                False,
        },

    "hyperpigmentation":
        {
            "present":
                hyperpigmentation_present,
            "integrated":
                False,
            "box_to_mask_conversion":
                False,
        },

    "trainable":
        trainable,

    "raw_data_modified":
        False,

    "validation_test_modified":
        False,

}


with open(
    REPORT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        report,
        f,
        indent=2
    )


# =============================================================================
# FINAL OUTPUT
# =============================================================================

print()
print("=" * 80)
print("STAGE 8K COMPLETE")
print("=" * 80)

print()

print("FINAL DATASET")

print(
    f"  train: "
    f"{train_stats['images']} images | "
    f"{train_stats['masks']} masks"
)

print(
    f"  valid: "
    f"{valid_stats['images']} images | "
    f"{valid_stats['masks']} masks"
)

print(
    f"  test : "
    f"{test_stats['images']} images | "
    f"{test_stats['masks']} masks"
)

print()

print("FINAL TRAINING CLASS EXPOSURE")

for class_id in range(
    1,
    6
):

    print(
        f"  {TARGET_CLASSES[class_id]:<10}: "
        f"{class_distribution['train'][class_id]}"
    )

print()

print("PIH STATUS")

print(
    f"  Approved PIH images: "
    f"{pih_train_count}"
)

print(
    "  PIH labels fabricated: NO"
)

print(
    "  Broad pigmentation labels mapped to PIH: NO"
)

print()

print("HYPERPIGMENTATION")

print(
    "  Kept separate as object detection"
)

print(
    "  Bounding boxes converted to masks: NO"
)

print(
    "  Integrated into SegFormer dataset: NO"
)

print()

print("TRAINABILITY")

print(
    "  ✓ DATASET IS READY FOR SEGMENTATION TRAINING"
)

print()

print("Output:")
print(
    f"  {OUTPUT_ROOT}"
)

print()

print("Manifest:")
print(
    f"  {MANIFEST_PATH}"
)

print()

print("Report:")
print(
    f"  {REPORT_PATH}"
)

print()

print("=" * 80)
print("RAW DATA WAS NOT MODIFIED.")
print("NO PIH LABELS WERE FABRICATED.")
print("NO BOUNDING BOXES WERE CONVERTED TO MASKS.")
print("VALIDATION/TEST WERE NOT AUGMENTED.")
print("NO TRAINING WAS PERFORMED.")
print("=" * 80)
```

```python
# =============================================================================
# CLINDERMA STAGE 8L
# SEGFORMER-B2 — MULTI-SOURCE SEGMENTATION TRAINING
# ORIENTATION-SAFE + RESOLUTION-SAFE METRICS
# =============================================================================

import os
import json
import time
import math
import random
import shutil
from pathlib import Path

import numpy as np
from PIL import Image

import torch
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from transformers import (
    SegformerImageProcessor,
    SegformerForSemanticSegmentation,
)

# =============================================================================
# CONFIGURATION
# =============================================================================

SEED = 42

NUM_CLASSES = 6

CLASS_NAMES = {
    0: "background",
    1: "PIH",
    2: "PIE",
    3: "SCAR",
    4: "DARK_SPOT",
    5: "MELASMA",
}

IMAGE_SIZE = 512

BATCH_SIZE = 2
GRADIENT_ACCUMULATION = 4
EFFECTIVE_BATCH_SIZE = BATCH_SIZE * GRADIENT_ACCUMULATION

MAX_EPOCHS = 30
WARMUP_EPOCHS = 2

LEARNING_RATE = 6e-5
WEIGHT_DECAY = 0.01

NUM_WORKERS = 2

PATIENCE = 7

MODEL_NAME = "nvidia/mit-b2"

DATASET_DIR = Path(
    "/content/clinderma_workspace/processed/"
    "pigmentation_multisource_final"
)

CHECKPOINT_DIR = Path(
    "/content/drive/MyDrive/Clinderma_Pigmentation/"
    "training/checkpoints/segformer_b2_multisource_v2"
)

RESULTS_DIR = Path(
    "/content/drive/MyDrive/Clinderma_Pigmentation/"
    "training/results"
)

CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

LAST_CHECKPOINT = CHECKPOINT_DIR / "last_state.pt"
BEST_MODEL = CHECKPOINT_DIR / "best_model.pt"

HISTORY_FILE = RESULTS_DIR / "segformer_b2_multisource_v2_history.json"
FINAL_RESULTS_FILE = RESULTS_DIR / "segformer_b2_multisource_v2_final_results.json"


# =============================================================================
# REPRODUCIBILITY
# =============================================================================

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")


# =============================================================================
# HEADER
# =============================================================================

print("=" * 80)
print("CLINDERMA STAGE 8L")
print("SEGFORMER-B2 — MULTI-SOURCE SEGMENTATION TRAINING")
print("ORIENTATION-SAFE + RESOLUTION-SAFE METRICS")
print("=" * 80)

print()
print("TARGET CLASSES")

for idx, name in CLASS_NAMES.items():
    print(f"  {idx}: {name}")

print()
print("Dataset:")
print(f"  {DATASET_DIR}")

print()
print("Device:")
print(f"  {DEVICE}")

if torch.cuda.is_available():
    print(f"  GPU: {torch.cuda.get_device_name(0)}")


# =============================================================================
# DATASET
# =============================================================================

class ClindermaSegmentationDataset(Dataset):

    def __init__(self, split_dir, processor):

        self.split_dir = Path(split_dir)
        self.processor = processor

        self.image_dir = self.split_dir / "images"
        self.mask_dir = self.split_dir / "masks"

        if not self.image_dir.exists():
            raise RuntimeError(
                f"Image directory not found:\n{self.image_dir}"
            )

        if not self.mask_dir.exists():
            raise RuntimeError(
                f"Mask directory not found:\n{self.mask_dir}"
            )

        image_extensions = {
            ".jpg",
            ".jpeg",
            ".png",
            ".bmp",
            ".webp"
        }

        self.images = sorted(
            [
                p for p in self.image_dir.iterdir()
                if p.suffix.lower() in image_extensions
            ]
        )

        if len(self.images) == 0:
            raise RuntimeError(
                f"No images found in {self.image_dir}"
            )

        self.pairs = []

        for image_path in self.images:

            mask_path = self.mask_dir / f"{image_path.stem}.png"

            if not mask_path.exists():

                raise RuntimeError(
                    f"Missing mask for image:\n"
                    f"{image_path}\n"
                    f"Expected:\n"
                    f"{mask_path}"
                )

            self.pairs.append(
                (image_path, mask_path)
            )

    def __len__(self):
        return len(self.pairs)

    def __getitem__(self, idx):

        image_path, mask_path = self.pairs[idx]

        image = Image.open(image_path).convert("RGB")
        mask = Image.open(mask_path)

        # ---------------------------------------------------------
        # Respect physical image orientation.
        # ---------------------------------------------------------

        try:
            image = ImageOps.exif_transpose(image)
        except Exception:
            pass

        image_np = np.array(image)

        mask_np = np.array(mask)

        # ---------------------------------------------------------
        # Ensure mask contains only valid class IDs.
        # ---------------------------------------------------------

        if mask_np.ndim != 2:
            raise RuntimeError(
                f"Mask must be single-channel:\n{mask_path}"
            )

        unique_values = np.unique(mask_np)

        invalid_values = [
            int(v)
            for v in unique_values
            if int(v) < 0 or int(v) >= NUM_CLASSES
        ]

        if invalid_values:
            raise RuntimeError(
                f"Invalid class IDs {invalid_values} "
                f"found in mask:\n{mask_path}"
            )

        # ---------------------------------------------------------
        # Resize image and mask together.
        #
        # Image:
        #   bilinear
        #
        # Mask:
        #   nearest-neighbor
        # ---------------------------------------------------------

        image_pil = Image.fromarray(image_np)

        mask_pil = Image.fromarray(
            mask_np.astype(np.uint8),
            mode="L"
        )

        image_pil = image_pil.resize(
            (IMAGE_SIZE, IMAGE_SIZE),
            Image.Resampling.BILINEAR
        )

        mask_pil = mask_pil.resize(
            (IMAGE_SIZE, IMAGE_SIZE),
            Image.Resampling.NEAREST
        )

        # ---------------------------------------------------------
        # SegFormer processor
        # ---------------------------------------------------------

        encoded = self.processor(
            images=image_pil,
            segmentation_maps=mask_pil,
            return_tensors="pt",
        )

        pixel_values = encoded["pixel_values"].squeeze(0)

        labels = encoded["labels"].squeeze(0)

        labels = labels.long()

        # Processor may return another spatial resolution.
        # Force labels to IMAGE_SIZE x IMAGE_SIZE.
        if labels.shape[-2:] != (IMAGE_SIZE, IMAGE_SIZE):

            labels = F.interpolate(
                labels.unsqueeze(0).unsqueeze(0).float(),
                size=(IMAGE_SIZE, IMAGE_SIZE),
                mode="nearest",
            ).squeeze(0).squeeze(0).long()

        return {
            "pixel_values": pixel_values,
            "labels": labels,
        }


# =============================================================================
# PROCESSOR
# =============================================================================

print()
print("=" * 80)
print("LOADING SEGFORMER PROCESSOR")
print("=" * 80)

processor = SegformerImageProcessor.from_pretrained(
    MODEL_NAME,
    do_resize=True,
    size={
        "height": IMAGE_SIZE,
        "width": IMAGE_SIZE,
    },
    do_reduce_labels=False,
)

print("Processor loaded.")


# =============================================================================
# DATASETS
# =============================================================================

print()
print("=" * 80)
print("CREATING DATASETS")
print("=" * 80)

train_dataset = ClindermaSegmentationDataset(
    DATASET_DIR / "train",
    processor
)

valid_dataset = ClindermaSegmentationDataset(
    DATASET_DIR / "valid",
    processor
)

test_dataset = ClindermaSegmentationDataset(
    DATASET_DIR / "test",
    processor
)

print(f"Train: {len(train_dataset)}")
print(f"Valid: {len(valid_dataset)}")
print(f"Test : {len(test_dataset)}")


# =============================================================================
# DATALOADERS
# =============================================================================

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
    drop_last=False,
)

valid_loader = DataLoader(
    valid_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
    drop_last=False,
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
    drop_last=False,
)

print()
print("DataLoaders created.")

print(f"Train batches: {len(train_loader)}")
print(f"Valid batches: {len(valid_loader)}")
print(f"Test batches : {len(test_loader)}")


# =============================================================================
# MODEL
# =============================================================================

print()
print("=" * 80)
print("LOADING PRETRAINED SEGFORMER-B2")
print("=" * 80)

model = SegformerForSemanticSegmentation.from_pretrained(
    MODEL_NAME,
    num_labels=NUM_CLASSES,
    id2label=CLASS_NAMES,
    label2id={
        name: idx
        for idx, name in CLASS_NAMES.items()
    },
    ignore_mismatched_sizes=True,
)

model.to(DEVICE)

print()
print("Model loaded.")

total_parameters = sum(
    p.numel()
    for p in model.parameters()
)

trainable_parameters = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print(f"Total parameters     : {total_parameters:,}")
print(f"Trainable parameters : {trainable_parameters:,}")


# =============================================================================
# OPTIMIZER
# =============================================================================

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY,
)

print()
print("Optimizer:")
print("  AdamW")
print(f"  Learning rate: {LEARNING_RATE}")
print(f"  Weight decay : {WEIGHT_DECAY}")


# =============================================================================
# LEARNING RATE SCHEDULER
# =============================================================================

total_update_steps = math.ceil(
    len(train_loader) / GRADIENT_ACCUMULATION
) * MAX_EPOCHS

warmup_steps = (
    math.ceil(len(train_loader) / GRADIENT_ACCUMULATION)
    * WARMUP_EPOCHS
)


def lr_lambda(current_step):

    if current_step < warmup_steps:

        return float(current_step + 1) / max(
            1,
            warmup_steps
        )

    progress = (
        current_step - warmup_steps
    ) / max(
        1,
        total_update_steps - warmup_steps
    )

    return max(
        0.0,
        0.5 * (
            1.0 + math.cos(math.pi * progress)
        )
    )


scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer,
    lr_lambda
)


# =============================================================================
# MIXED PRECISION
# =============================================================================

USE_AMP = torch.cuda.is_available()

if USE_AMP:

    scaler = torch.amp.GradScaler(
        "cuda"
    )

else:

    scaler = None


print()
print("Training configuration:")
print(f"  Maximum epochs       : {MAX_EPOCHS}")
print(f"  Batch size           : {BATCH_SIZE}")
print(f"  Gradient accumulation: {GRADIENT_ACCUMULATION}")
print(f"  Effective batch size : {EFFECTIVE_BATCH_SIZE}")
print(f"  Mixed precision      : {USE_AMP}")


# =============================================================================
# CONFUSION MATRIX
# =============================================================================

def update_confusion_matrix(
    confusion,
    predictions,
    labels,
    num_classes
):

    predictions = predictions.detach().view(-1).long()
    labels = labels.detach().view(-1).long()

    valid = (
        (labels >= 0)
        & (labels < num_classes)
        & (predictions >= 0)
        & (predictions < num_classes)
    )

    predictions = predictions[valid]
    labels = labels[valid]

    indices = (
        labels * num_classes
        + predictions
    )

    bincount = torch.bincount(
        indices,
        minlength=num_classes * num_classes
    )

    confusion += bincount.reshape(
        num_classes,
        num_classes
    ).cpu()


# =============================================================================
# METRICS
# =============================================================================

def calculate_metrics(confusion):

    confusion = confusion.float()

    intersection = torch.diag(confusion)

    ground_truth = confusion.sum(dim=1)

    predicted = confusion.sum(dim=0)

    union = (
        ground_truth
        + predicted
        - intersection
    )

    iou = intersection / union.clamp(
        min=1e-8
    )

    dice = (
        2 * intersection
        / (
            ground_truth
            + predicted
        ).clamp(min=1e-8)
    )

    # -------------------------------------------------------------
    # Background is included in the global metric because this is
    # how the previous pipeline reported mIoU/Dice.
    # -------------------------------------------------------------

    miou = iou.mean().item()

    mean_dice = dice.mean().item()

    return (
        miou,
        mean_dice,
        iou.numpy(),
        dice.numpy()
    )


# =============================================================================
# CRITICAL RESOLUTION-SAFE METRIC FUNCTION
# =============================================================================

def prepare_predictions_for_metrics(
    logits,
    labels
):

    # -------------------------------------------------------------
    # SegFormer-B2 normally produces lower-resolution logits.
    #
    # Example:
    #
    # logits:
    #   [2, 6, 128, 128]
    #
    # labels:
    #   [2, 512, 512]
    #
    # Before:
    #
    #   argmax(logits) -> 128 x 128
    #
    # compared with:
    #
    #   labels -> 512 x 512
    #
    # This causes:
    #
    #   IndexError
    #
    # Therefore logits are resized BEFORE argmax.
    # -------------------------------------------------------------

    target_size = labels.shape[-2:]

    if logits.shape[-2:] != target_size:

        logits = F.interpolate(
            logits,
            size=target_size,
            mode="bilinear",
            align_corners=False,
        )

    predictions = torch.argmax(
        logits,
        dim=1
    )

    return predictions


# =============================================================================
# SMOKE TEST
# =============================================================================

print()
print("=" * 80)
print("GPU FORWARD-PASS SMOKE TEST")
print("=" * 80)

sample_batch = next(
    iter(train_loader)
)

sample_pixels = sample_batch[
    "pixel_values"
].to(DEVICE)

sample_labels = sample_batch[
    "labels"
].to(DEVICE)

with torch.no_grad():

    sample_output = model(
        pixel_values=sample_pixels,
        labels=sample_labels
    )

print(
    f"Input  : {tuple(sample_pixels.shape)}"
)

print(
    f"Labels : {tuple(sample_labels.shape)}"
)

print(
    f"Logits : {tuple(sample_output.logits.shape)}"
)

print(
    f"Smoke-test loss : "
    f"{sample_output.loss.item():.6f}"
)


# =============================================================================
# CHECKPOINT LOADING
# =============================================================================

start_epoch = 1
best_val_dice = -1.0
epochs_without_improvement = 0
global_step = 0

history = []

if LAST_CHECKPOINT.exists():

    print()
    print("=" * 80)
    print("CHECKPOINT FOUND")
    print("=" * 80)

    checkpoint = torch.load(
        LAST_CHECKPOINT,
        map_location=DEVICE,
        weights_only=False
    )

    model.load_state_dict(
        checkpoint["model_state"]
    )

    optimizer.load_state_dict(
        checkpoint["optimizer_state"]
    )

    scheduler.load_state_dict(
        checkpoint["scheduler_state"]
    )

    if USE_AMP and checkpoint.get(
        "scaler_state"
    ) is not None:

        scaler.load_state_dict(
            checkpoint["scaler_state"]
        )

    start_epoch = (
        checkpoint["epoch"] + 1
    )

    best_val_dice = checkpoint.get(
        "best_val_dice",
        -1.0
    )

    epochs_without_improvement = checkpoint.get(
        "epochs_without_improvement",
        0
    )

    global_step = checkpoint.get(
        "global_step",
        0
    )

    history = checkpoint.get(
        "history",
        []
    )

    print(
        f"Resuming from epoch "
        f"{checkpoint['epoch']}"
    )

    print(
        f"Best validation Dice: "
        f"{best_val_dice:.5f}"
    )

else:

    print()
    print("No previous checkpoint found.")
    print("Starting from pretrained SegFormer-B2.")


# =============================================================================
# TRAINING FUNCTION
# =============================================================================

def train_one_epoch():

    model.train()

    running_loss = 0.0

    optimizer.zero_grad(
        set_to_none=True
    )

    for batch_idx, batch in enumerate(
        train_loader
    ):

        pixel_values = batch[
            "pixel_values"
        ].to(
            DEVICE,
            non_blocking=True
        )

        labels = batch[
            "labels"
        ].to(
            DEVICE,
            non_blocking=True
        )

        if USE_AMP:

            with torch.autocast(
                device_type="cuda",
                dtype=torch.float16
            ):

                outputs = model(
                    pixel_values=pixel_values,
                    labels=labels
                )

                loss = outputs.loss

                loss_for_backward = (
                    loss
                    / GRADIENT_ACCUMULATION
                )

            scaler.scale(
                loss_for_backward
            ).backward()

        else:

            outputs = model(
                pixel_values=pixel_values,
                labels=labels
            )

            loss = outputs.loss

            loss_for_backward = (
                loss
                / GRADIENT_ACCUMULATION
            )

            loss_for_backward.backward()

        running_loss += loss.item()

        should_update = (
            (batch_idx + 1)
            % GRADIENT_ACCUMULATION
            == 0
        ) or (
            batch_idx + 1
            == len(train_loader)
        )

        if should_update:

            if USE_AMP:

                scaler.unscale_(
                    optimizer
                )

                torch.nn.utils.clip_grad_norm_(
                    model.parameters(),
                    max_norm=1.0
                )

                scaler.step(
                    optimizer
                )

                scaler.update()

            else:

                torch.nn.utils.clip_grad_norm_(
                    model.parameters(),
                    max_norm=1.0
                )

                optimizer.step()

            optimizer.zero_grad(
                set_to_none=True
            )

            scheduler.step()

    return (
        running_loss
        / len(train_loader)
    )


# =============================================================================
# VALIDATION FUNCTION
# =============================================================================

def evaluate(loader):

    model.eval()

    running_loss = 0.0

    confusion = torch.zeros(
        (
            NUM_CLASSES,
            NUM_CLASSES
        ),
        dtype=torch.int64
    )

    with torch.no_grad():

        for batch in loader:

            pixel_values = batch[
                "pixel_values"
            ].to(
                DEVICE,
                non_blocking=True
            )

            labels = batch[
                "labels"
            ].to(
                DEVICE,
                non_blocking=True
            )

            if USE_AMP:

                with torch.autocast(
                    device_type="cuda",
                    dtype=torch.float16
                ):

                    outputs = model(
                        pixel_values=pixel_values,
                        labels=labels
                    )

            else:

                outputs = model(
                    pixel_values=pixel_values,
                    labels=labels
                )

            loss = outputs.loss

            running_loss += loss.item()

            # -----------------------------------------------------
            # IMPORTANT:
            # Resize logits 128x128 -> labels 512x512
            # BEFORE calculating predictions.
            # -----------------------------------------------------

            predictions = (
                prepare_predictions_for_metrics(
                    outputs.logits,
                    labels
                )
            )

            update_confusion_matrix(
                confusion,
                predictions,
                labels,
                NUM_CLASSES
            )

    avg_loss = (
        running_loss
        / len(loader)
    )

    miou, dice, ious, dices = (
        calculate_metrics(confusion)
    )

    return (
        avg_loss,
        miou,
        dice,
        ious,
        dices
    )


# =============================================================================
# TRAINING LOOP
# =============================================================================

print()
print("=" * 80)
print("TRAINING START")
print("=" * 80)

print(
    f"Maximum epochs       : {MAX_EPOCHS}"
)

print(
    f"Starting epoch       : {start_epoch}"
)

print(
    f"Batch size           : {BATCH_SIZE}"
)

print(
    f"Effective batch size : "
    f"{EFFECTIVE_BATCH_SIZE}"
)

print(
    f"Learning rate        : "
    f"{LEARNING_RATE}"
)

print(
    f"Checkpoint directory : "
    f"{CHECKPOINT_DIR}"
)


training_start_time = time.time()

for epoch in range(
    start_epoch,
    MAX_EPOCHS + 1
):

    epoch_start = time.time()

    print()
    print("-" * 80)
    print(
        f"EPOCH {epoch}/{MAX_EPOCHS}"
    )
    print("-" * 80)

    # -------------------------------------------------------------
    # TRAIN
    # -------------------------------------------------------------

    train_loss = train_one_epoch()

    # -------------------------------------------------------------
    # VALIDATION
    # -------------------------------------------------------------

    (
        val_loss,
        val_miou,
        val_dice,
        val_ious,
        val_dices
    ) = evaluate(
        valid_loader
    )

    epoch_time = (
        time.time()
        - epoch_start
    )

    current_lr = (
        optimizer.param_groups[0]["lr"]
    )

    print(
        f"Train loss : {train_loss:.5f}"
    )

    print(
        f"Val loss   : {val_loss:.5f}"
    )

    print(
        f"Val mIoU   : {val_miou:.5f}"
    )

    print(
        f"Val Dice   : {val_dice:.5f}"
    )

    print()

    print("Per-class IoU:")

    for class_id, class_name in CLASS_NAMES.items():

        print(
            f"  {class_name:<12}: "
            f"{val_ious[class_id]:.5f}"
        )

    print()

    print("Per-class Dice:")

    for class_id, class_name in CLASS_NAMES.items():

        print(
            f"  {class_name:<12}: "
            f"{val_dices[class_id]:.5f}"
        )

    print()

    print(
        f"Learning rate: "
        f"{current_lr:.8f}"
    )

    print(
        f"Epoch time: "
        f"{epoch_time / 60:.2f} min"
    )

    # -------------------------------------------------------------
    # HISTORY
    # -------------------------------------------------------------

    epoch_record = {

        "epoch": epoch,

        "train_loss": float(
            train_loss
        ),

        "val_loss": float(
            val_loss
        ),

        "val_miou": float(
            val_miou
        ),

        "val_dice": float(
            val_dice
        ),

        "learning_rate": float(
            current_lr
        ),

        "epoch_time_seconds": float(
            epoch_time
        ),

        "class_iou": {
            CLASS_NAMES[i]: float(
                val_ious[i]
            )
            for i in range(NUM_CLASSES)
        },

        "class_dice": {
            CLASS_NAMES[i]: float(
                val_dices[i]
            )
            for i in range(NUM_CLASSES)
        },
    }

    history.append(
        epoch_record
    )

    # -------------------------------------------------------------
    # BEST MODEL
    # -------------------------------------------------------------

    is_best = (
        val_dice > best_val_dice
    )

    if is_best:

        best_val_dice = val_dice

        epochs_without_improvement = 0

        torch.save(
            model.state_dict(),
            BEST_MODEL
        )

        print()
        print("✓ NEW BEST MODEL")

        print(
            f"Best validation Dice: "
            f"{best_val_dice:.5f}"
        )

    else:

        epochs_without_improvement += 1

        print()
        print("No improvement.")

        print(
            f"Epochs without improvement: "
            f"{epochs_without_improvement}"
        )

    # -------------------------------------------------------------
    # CHECKPOINT
    # -------------------------------------------------------------

    checkpoint = {

        "epoch": epoch,

        "model_state": model.state_dict(),

        "optimizer_state": optimizer.state_dict(),

        "scheduler_state": scheduler.state_dict(),

        "scaler_state": (
            scaler.state_dict()
            if USE_AMP
            else None
        ),

        "best_val_dice": best_val_dice,

        "epochs_without_improvement":
            epochs_without_improvement,

        "global_step": global_step,

        "history": history,

        "config": {

            "model_name": MODEL_NAME,

            "num_classes": NUM_CLASSES,

            "class_names": CLASS_NAMES,

            "image_size": IMAGE_SIZE,

            "batch_size": BATCH_SIZE,

            "gradient_accumulation":
                GRADIENT_ACCUMULATION,

            "learning_rate":
                LEARNING_RATE,

            "weight_decay":
                WEIGHT_DECAY,

            "max_epochs":
                MAX_EPOCHS,
        }
    }

    torch.save(
        checkpoint,
        LAST_CHECKPOINT
    )

    with open(
        HISTORY_FILE,
        "w"
    ) as f:

        json.dump(
            history,
            f,
            indent=2
        )

    print()
    print(
        "✓ Checkpoint saved to Drive"
    )

    # -------------------------------------------------------------
    # EARLY STOPPING
    # -------------------------------------------------------------

    if (
        epochs_without_improvement
        >= PATIENCE
    ):

        print()
        print(
            "Early stopping triggered."
        )

        break


# =============================================================================
# TRAINING COMPLETE
# =============================================================================

total_training_time = (
    time.time()
    - training_start_time
)

print()
print("=" * 80)
print("TRAINING LOOP COMPLETE")
print("=" * 80)

print(
    f"Best validation Dice: "
    f"{best_val_dice:.5f}"
)

print(
    f"Training time: "
    f"{total_training_time / 3600:.2f} hours"
)

print(
    f"Last checkpoint: "
    f"{LAST_CHECKPOINT}"
)

print(
    f"Best checkpoint: "
    f"{BEST_MODEL}"
)

print(
    f"History: "
    f"{HISTORY_FILE}"
)


# =============================================================================
# LOAD BEST MODEL
# =============================================================================

print()
print("=" * 80)
print("LOADING BEST CHECKPOINT FOR FINAL TEST EVALUATION")
print("=" * 80)

best_state = torch.load(
    BEST_MODEL,
    map_location=DEVICE,
    weights_only=False
)

# Supports both:
# 1. raw model state_dict
# 2. checkpoint dictionary

if (
    isinstance(best_state, dict)
    and "model_state" in best_state
):

    model.load_state_dict(
        best_state["model_state"]
    )

else:

    model.load_state_dict(
        best_state
    )

model.to(DEVICE)


# =============================================================================
# FINAL HELD-OUT TEST
# =============================================================================

print()
print("=" * 80)
print("FINAL HELD-OUT TEST EVALUATION")
print("=" * 80)

(
    test_loss,
    test_miou,
    test_dice,
    test_ious,
    test_dices
) = evaluate(
    test_loader
)

print()
print(
    f"Test loss : {test_loss:.5f}"
)

print(
    f"Test mIoU : {test_miou:.5f}"
)

print(
    f"Test Dice : {test_dice:.5f}"
)

print()
print("Class IoU:")

for class_id, class_name in CLASS_NAMES.items():

    print(
        f"  {class_name:<12}: "
        f"{test_ious[class_id]:.5f}"
    )

print()
print("Class Dice:")

for class_id, class_name in CLASS_NAMES.items():

    print(
        f"  {class_name:<12}: "
        f"{test_dices[class_id]:.5f}"
    )


# =============================================================================
# FINAL RESULTS
# =============================================================================

final_results = {

    "model": MODEL_NAME,

    "num_classes": NUM_CLASSES,

    "class_names": CLASS_NAMES,

    "image_size": IMAGE_SIZE,

    "best_validation_dice": float(
        best_val_dice
    ),

    "test_loss": float(
        test_loss
    ),

    "test_miou": float(
        test_miou
    ),

    "test_dice": float(
        test_dice
    ),

    "test_class_iou": {
        CLASS_NAMES[i]: float(
            test_ious[i]
        )
        for i in range(NUM_CLASSES)
    },

    "test_class_dice": {
        CLASS_NAMES[i]: float(
            test_dices[i]
        )
        for i in range(NUM_CLASSES)
    },

    "training_configuration": {

        "batch_size": BATCH_SIZE,

        "gradient_accumulation":
            GRADIENT_ACCUMULATION,

        "effective_batch_size":
            EFFECTIVE_BATCH_SIZE,

        "learning_rate":
            LEARNING_RATE,

        "weight_decay":
            WEIGHT_DECAY,

        "max_epochs":
            MAX_EPOCHS,

        "patience":
            PATIENCE,

        "mixed_precision":
            USE_AMP,
    },

    "dataset": str(
        DATASET_DIR
    ),

    "best_model": str(
        BEST_MODEL
    ),
}

with open(
    FINAL_RESULTS_FILE,
    "w"
) as f:

    json.dump(
        final_results,
        f,
        indent=2
    )


# =============================================================================
# COMPLETE
# =============================================================================

print()
print("=" * 80)
print("STAGE 8L COMPLETE")
print("=" * 80)

print()
print("BEST VALIDATION DICE")
print(
    f"  {best_val_dice:.5f}"
)

print()
print("FINAL TEST RESULTS")

print(
    f"  Test mIoU : {test_miou:.5f}"
)

print(
    f"  Test Dice : {test_dice:.5f}"
)

print()
print("BEST MODEL")
print(
    BEST_MODEL
)

print()
print("FINAL RESULTS")
print(
    FINAL_RESULTS_FILE
)

print()
print("=" * 80)
```
